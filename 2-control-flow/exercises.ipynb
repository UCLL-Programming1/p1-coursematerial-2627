## Exercise 2.4: Rock paper scissors

Time to build a classic: a two-player game of rock paper scissors. Both players
secretly enter their choice, the program decides who won the round, and it keeps a
running score until the players decide to stop.

Here is how a full game should work:

- The program prints `Welcome to Rock Paper Scissors!`.
- It then plays rounds, one after another, until player 1 chooses to quit:
  - Player 1 enters `rock`, `paper`, `scissors`, or `Q` to quit.
  - If player 1 quits, the game stops.
  - Otherwise player 2 enters their choice (`rock`, `paper` or `scissors`).
  - The program announces the result of the round: `It's a tie!`,
    `Player 1 wins this round!`, or `Player 2 wins this round!`. Remember:
    rock beats scissors, scissors beats paper, and paper beats rock.
  - The winner of the round gets a point, and the program prints the current score
    as `Score - <score 1> - <score 2>`.
- When player 1 quits, the program prints `Thanks for playing!`.

The choices are not case-sensitive: `Rock`, `ROCK` and `rock` all mean the same
thing, and `q` quits just like `Q`.

The program will run as follows:

```none
Welcome to Rock Paper Scissors!
Player 1, enter your choice (rock/paper/scissors) or (Q)uit: > rock
Player 2, enter your choice (rock/paper/scissors): > scissors
Player 1 wins this round!
Score - Player 1: 1, Player 2: 0
Player 1, enter your choice (rock/paper/scissors) or (Q)uit: > paper
Player 2, enter your choice (rock/paper/scissors): > paper
It's a tie!
Score - Player 1: 1, Player 2: 0
Player 1, enter your choice (rock/paper/scissors) or (Q)uit: > Q
Thanks for playing!
```
where `rock`, `scissors`, `paper`, `paper` and `Q` are input provided by the user.

## Exercise 2.5: Hangman

Let's play Hangman!

In this exercise you'll build the word-guessing game hangman. One player (the
gamemaster) secretly picks a word. The other player tries to discover it by guessing
one letter at a time. Every letter they guess that appears in the word is revealed;
every letter that doesn't, costs them a life. The player starts with `11` lives. Guess the
whole word before the lives run out and you win, otherwise you lose.

Here is how the full game works:

- The program prints `Welcome to hangman!` and asks the gamemaster for the word.
- It prints a blank line and `Ready to play!`.
- Then it repeats the following until the game is over. Each time around it shows:
  - a blank line,
  - `Already guessed: ` followed by every letter guessed so far,
  - `Word to guess: ` followed by the word with revealed letters shown and the rest
    shown as `*`,
  - `<lives> lives left`,
  - another blank line,
  - and then it asks `Guess a letter: > `.
- For each guess:
  - If it isn't exactly one character, the program prints
    `Your guess needs to be exactly one character!` and nothing else changes.
  - If that letter was already guessed before, it prints `You already guessed <letter>!`
    and nothing else changes.
  - Otherwise the letter is added to the guessed letters. If it appears in the word,
    the program prints `Good guess!`; if not, it prints `Miss!` and the player loses
    a life.
- The player **wins** as soon as the whole word is revealed: the program prints
  `You win!` and stops. The player **loses** when their lives reach `0`: the program
  prints `You lose!` and then `The correct word was: <word>`, and stops.

A short example game (the word is `hi`):

```none
Welcome to hangman!
Gamemaster, which word should be guessed: > hi

Ready to play!

Already guessed:
Word to guess: **
11 lives left

Guess a letter: > z
Miss!

Already guessed: z
Word to guess: **
10 lives left

Guess a letter: > ha
Your guess needs to be exactly one character!

Already guessed: z
Word to guess: **
10 lives left

Guess a letter: > h
Good guess!

Already guessed: zh
Word to guess: h*
10 lives left

Guess a letter: > i
Good guess!
You win!
```
where `hi`, `z`, `h` and `i` are input provided by the user.

Or in the case the player runs out of lives (we're skipping to the end of the game):
```none
...
Already guessed: qwertyuiop
Word to guess: *******
1 lives left

Guess a letter: > z
Miss!
You lose!
The correct word was: hangman
```
