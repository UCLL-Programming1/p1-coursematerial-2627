# Executing code line-by-line (Debugging)

So far, we've been executing our programs all at once, which is how you typically run programs. In this chapter we're going to take a look at how we can execute our programs one line at a time, and see why this can be useful.

> Note: In some of the code examples that follow, there might be mistakes. This is intentional. These mistakes will serve to demonstrate the usefulness of what we're about to explain.

## Introduction Example

Consider a possible solution for the chapter 1 milestone exercise `highest-card-improved`:

````
print("Let's see who has the highest card!")
card_1 = input("Player 1, enter your card: > ")
card_2 = input("Player 2, enter your card: > ")

if card_1 == "Jack":
    card_1 = "11"
if card_2 == "Jack":
    card_2 = "11"

if card_1 == "Queen":
    card_1 = "12"
if card_2 == "Queen":
    card_2 = "12"

if card_1 == "King":
    card_1 = "13"
if card_2 == "King":
    card_2 = "13"

if card_1 > card_2:
    print("Player 1 has the highest card!")
elif card_1 < card_2:
    print("Player 2 has the highest card!")
else:
    print("Both players are holding the same value card!")
````

As a reminder, this code should allow us to enter two cards, for example `5` and `9`, and it should tell us which of them is the highest:
```none
Let's see who has the highest card!
Player 1, enter your card: > 5
Player 2, enter your card: > 9
Player 2 has the highest card!
```

We do this by first letting players enter a number:

````
print("Let's see who has the highest card!")
card_1 = input("Player 1, enter your card: > ")
card_2 = input("Player 2, enter your card: > ")
````

and finally comparing these cards, to print out the relevant message:

````
if card_1 > card_2:
    print("Player 1 has the highest card!")
elif card_1 < card_2:
    print("Player 2 has the highest card!")
else:
    print("Both players are holding the same value card!")
````

- if the value of `card_1` is higher (`>`) than that of `card_2`, we print `"Player 1 has the highest card!"`
- if that is not the case, but instead  the value of `card_1` is smaller (`<`) than that of `card_2`, we print (`"Player 2, enter your card: > "`)
- if it's neither of the previous cases, that means that the card values are **equal**, so we print `"Both players are holding the same value card!"`

This would work great for cards `1` - `10`, but we of course also have "face cards" (`"Jack"`, `"Queen"` or `"King"`), and in order to be able to compare these to the number cards, we simply assign them a value that corresponds to the value they have:

````
if card_1 == "Jack":
    card_1 = "11"
if card_2 == "Jack":
    card_2 = "11"

if card_1 == "Queen":
    card_1 = "12"
if card_2 == "Queen":
    card_2 = "12"

if card_1 == "King":
    card_1 = "13"
if card_2 == "King":
    card_2 = "13"
````

This way a `"Jack"` is worth `"11"`, a `"Queen"` `"12"` and a `"King"` `"13"`.

Let's run this code a couple of times, to verify that it works as intended.

First, run it with the value `2` and `9` for player 1 and 2 respectively. You would get the following output:
```none
Let's see who has the highest card!
Player 1, enter your card: > 2
Player 2, enter your card: > 9
Player 2 has the highest card!
```

Great, so far it seems to be doing what we expect!

Now, let's run it again, this time letting player 1 have the higher card, for example with values `8` and `5`:
```none
Let's see who has the highest card!
Player 1, enter your card: > 8
Player 2, enter your card: > 5
Player 1 has the highest card!
```

It still works!

So far we've only ran the code with number values. Let's also try it with one of them being a 'face card' (Jack, King or Queen), for example `3` and `Queen`:
```none
Let's see who has the highest card!
Player 1, enter your card: > 3
Player 2, enter your card: > Queen
Player 1 has the highest card!
```

Ouch, it turns out that our code isn't completely correct after all...

## Finding a mistake in our code

This means there is a mistake in our code.

At this point you might have already spotted it. If you haven't, don't feel bad. We made this mistake challenging to spot on purpose, hoping that you wouldn't figure it out immediately. If you did already find it, for the following explanation try to pretend that you didn't :).

So, assuming we didn't already find the mistake, how do we go about finding it?

Of course we could try to go over the code again, line by line, see if we can figure out what we did wrong, but we just did that already, when explaining what each part of the code was doing.

The difficulty of finding such mistakes in code is that **it could be anywhere**, and we could spend a lot of time looking at parts of the code that don't actually have the mistake in them.

This is an inherent problem that we have when we execute the program in its entirety:
- we start the program
- we see the output when it's done

This means that the mistake could be anywhere between the start and the end of the program (when we get the output)

Is there a better way? Fortunately yes.

## Executing code line by line: how?

Instead of executing code all at once, we can instead execute it one line at a time, allowing us to check whether that line did what we expected it to do. We call executing the code line by line **Debugging**. Let's see how this works.

Doing this requires a couple of steps. At first it may seem overly complicated, but at the end you will see that all of these steps serve a purpose.

> Note: In what follows we are going to describe how to execute the code line by line for the highest card example. All of this will make more sense if you just immediately follow along with the steps yourself!

### Step 1. Place a 'breakpoint'

The first step you have to do is to put a breakpoint. This is the point at which the execution will **pause**. You can place a breakpoint by hovering over the different lines of code to the lefthandside. A little red circle should appear. Click the on it to place the breakpoint. For now, we will just put it on the first line of the code.

![Breakpoint](../.assets/breakpoint.png)

### 2. Run in 'debugging mode'

After having put the breakpoint, instead of running the code by clicking on the `play`-button, we actually click on the little dropdown to the right of it:

![Debug Dropdown](../.assets/debug-dropdown.png)

This should display the option `Debug Cell`, which we can then click.

### 3. Debugging Mode

This will launch **Debugging Mode**. It should look something like this:

![Debugging Mode](../.assets/debugging-mode.png)

So far, it might not look very interesting, but observe the changes in VS Code (circled in blue):
- The line of code we put the breakpoint on (the little red circle) is now **highlighted in yellow**. This is the **line of code** we're **about to execute**. (so we haven't executed it yet)
- To the left, we now have a window with the title `VARIABLES` (though right now there is not much to see there)
- On the top (left) of your screen, there is a new little toolbar with some buttons we can click (we'll see what those do later)

### 4. Execute a single line with `Step into`

So let's click on of the buttons in the toolbar. Click the third icon, it should say `Step into`:

![Step Into](../.assets/step-into.png)

If you clicked on it, you should be able to see the following changes:
- There is some new information in the left window `special variables`,`function variables` etc. We will still ignore those for now.
- The output of the first line of code is visible under the cell as output (`Let's see who has the highest card!`). This is because the **first line of code has executed**
- The second line of the code is now highlighted in yellow. As the first line of code has executed, this line is now the **next line of code that will be executed**.

Let's do this another time.

This time, the program doesn't immediately jump to the next line. This is because on this line, we have to provide input. The program is waiting for that:

![Waiting for Input](../.assets/waiting-for-input.png)

For example, we could enter `Jack`, and press `enter`.

After doing this, the program does jump to the next line, as we can see by the third line now being highlighted. However we see another change: **The variable `card_1`** is now **displayed** in the `VARIABLES` menu, together with **the value it is assigned to**.

![Debug Variable](../.assets/debug-variable.png)

### Usefulness: more information

This is incredibly helpful to us! Though it **may have seemed obvious** that `card_1` was going to refer to this value, **now we know for sure**. We know for sure that this step of the code did what it was supposed to do. Later, when writing programs where variables will change values more often, this will be even more helpful, as we will be able to verify that the values they get will be what we expect them to be.

> - **When running a program entirely**, we only have information about **its output**.
> - **When running a program one line at a time**, we also have information about **its state after every single line**.

As we will see next, having that extra information will help us to reason better about our program.

## How to find the mistake in our code?

If at this point, it's not clear for you what the problem with the code is, you might feel that this is an impossible situation: "Comparisons don't work the way we just learned 1 week ago??!!".

**This is a common feeling to have** when programming. It's part of the journey.

Let's see what you can do, when you feel like there is nothing left to do.

### Always keep challenging your assumptions

One of the key things you can do is to try to **split your assumptions up in even smaller pieces and validate those**. Keep throwing stuff at the wall and see what sticks.

For example, `card_1` refers to the value 11 and `card_2` to the value 8, so if `card_1 > card_2` is `False`, this means that `11 > 8` is `False` too right?? Let's check this assumption by typing this expression into the debug console:

![11 > 8](../.assets/11_greater_than_8.png)

What?! How can we explain this? On the one hand we should be relieved that we at least know how comparisons work, but on the other hand does this mean that we don't know how variables work? Clearly that can't be the case. We've worked with variables a lot already and we know that **using the variable name instead of the value it is assigned to is equivalent, that it should always result in the exact same result**.

But wait a minute... Did we use the exact same value that our variable was referring to in our comparison? When we look at our variables window:

![Variables Close up](../.assets/variables_close_up.png)

and compare that to what we typed in our debug console:

![Debug Console Close up](../.assets/debug-console-close-up.png)

We see one little difference. Can you spot it?

In our variable window, our values are written as `'11'` and `'8'`, while in the debug console we wrote `11` and `8` (without the quotes (`'`)). Does this make a difference?

Well let's find out. Let's type the expression with quotes in our debug window as well:

![Expression Difference](../.assets/expression-difference.png)

### After pinpointing the cause

Voila! It seems like we're onto something. Apparently `'11' > '8'` and `11 > 8` are not the same thing and the version with quotes doesn't work.

So why is that?

At this point **the debugging mode can no longer help us**. But it did **point us at the key point** that seems to break our program. Now we have to rely on our Python knowledge to find the explanation and fix our code.

So what is the difference between `'11' > '8'` and `11 > 8`? As you might know the version with quotes (`'`) is **comparing strings**, while the version without it is **comparing numbers**. So why do they not work the same way?

Well as we saw (very briefly) in section 8 of last week (on `if`-`else` statements), the comparison operators (`>`, `<`, `>=` etc) compare strings **lexicographically**, meaning that a string that would appear **earlier** in the alphabet is **'smaller'** than a string that would appear later in the alphabet. The example we saw was:

````
"Adam" < "Eve"
````

where `"Adam"` is 'smaller' than `"Eve"` (because `A` appears in the alphabet before `E`).

So when we compare the string values:

````
"11" > "8"
````

What we are asking is "Does `1` appear later in the alphabet than `8`?"

The answer is **no**.

## Fixing our code

When we understand what is wrong with our code, we can go on to fix it.

Since the wrong behaviour is due to using strings instead of numbers, we should adapt our program to use numbers instead.

So what do we need to change specifically? Two things:
1. We need to use numbers instead of strings, so we should drop all the quotes `"`.
2. We know that all the values coming out of the `input()` function calls are strings by default (even if we type in something that could represent a number), so before comparing them at the end, we need to convert them to integers (using `int()`).

Making these two changes, we get the following code:

````
print("Let's see who has the highest card!")
card_1 = input("Player 1, enter your card: > ")
card_2 = input("Player 2, enter your card: > ")

if card_1 == "Jack":
    card_1 = 11
if card_2 == "Jack":
    card_2 = 11

if card_1 == "Queen":
    card_1 = 12
if card_2 == "Queen":
    card_2 = 12

if card_1 == "King":
    card_1 = 13
if card_2 == "King":
    card_2 = 13

card_1 = int(card_1)
card_2 = int(card_2)

if card_1 > card_2:
    print("Player 1 has the highest card!")
elif card_1 < card_2:
    print("Player 2 has the highest card!")
else:
    print("Both players are holding the same value card!")
````

## Other debugging options

### `Resume`

In the Debugging toolbar, we only used `Step Into` for now. However, as we saw earlier, there are other buttons that we can click. The first one, `Continue`, is also particularly useful.

![Continue](../.assets/continue.png)

To see what it does, put your breakpoint of the first line of the code and run the program in debugging mode. Then when it pauses on the first line, click the `Continue` button. What happens?

Instead of going to the next line, the code will simply 'resume', meaning that it continues to run until it reaches the end.

### Setting (multiple) breakpoints

So far we've always set our breakpoint on the first line. What would happen if you would set the breakpoint later on? (for example on the line `if card_1 > card_2):`)

![Later Breakpoint](../.assets/later_breakpoint.png)

Well the program no longer pauses on the first line, but on the line you put the breakpoint on instead (after you gave the card inputs for the different players).

This can be useful if you know for sure the first part of your code works fine, and you only want to start executing the code line by line at a specific point where you suspect the mistake to be.

You can also put down multiple breakpoints. The debugger will pause at the first one as always, but if you then select `Resume`, it will start running the program again until it pauses at the second breakpoint you put down.

### `Step Over` and other options

Apart from `Resume` and `Step Into`, there some other options as well. We won't use them for now, but will still give a short explanation for those of you who are interested.

The second button is `Step Over`

![Step Over](../.assets/step_over.png)

If you try it out, you will probably not notice any difference between `Step Over` and `Step Into`. For what we've learned about Python so far, this is indeed the case. Later we will learn about things where there actually is a difference between the two.

So while you could use any of the two, we recommend to use `Step Into`.

The other buttons are:
- `Step Out` (button 4), which also doesn't do anything special for us yet (you'll notice it behaves a lot like `Resume`)
- `Restart` (button 5) merely restarts the program.
- `Disconnect` (button 6) disconnects the debugger, finishing the program as a regular run (not even stopping for other breakpoints, like `Resume` does)

So far we've seen how to debug (execute a program line by line) the code of in Jupyter notebook cell. However, most of the code you'll be writing will be in a Python (`.py`) file.

In Python file you can put a breakpoint in the exact same way. Simply hover over the line to the left of the line numbers. A red button will appear that you can again click:

The debug mode works exactly the same, with

![Python Debug Buttons](../.assets/python-debug-buttons.png)

## Final notes about debugging

The most important thing to understand about running your program line-by-line like this **does not change** how the program behaves. If you run the program in the regular way, it follows exactly the same steps and does exactly the same thing as when you do it line by line. Doing it line-by-line, you simply get to see how it does it.

The term **debugging** refers both to the general process of trying to find and fix mistakes in your code and to executing our code line by line to achieve this (since that is what you will use it for most in practice).

However, even if you are not looking for mistakes in your code, executing your code line by line is also very useful when you just started out programming to understand exactly how your code runs. **We recommend to do this as much as you can!**
