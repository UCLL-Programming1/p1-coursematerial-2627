# Truthy and falsy

## Values as conditions

Take a look at the following piece of code. What will it do? Can it even run at all? Run the cell to find out.

1. The first time you run the cell, enter a name.
1. The second time, simply enter without running anything.

````
name = input("What is your name? > ")

if name:
    print("Hello, " + name + "!")
else:
    print("You didn't enter a name.")
````

Having done this, you will have noticed that `if name:` apparently tells the difference between the two cases (entering a name or not entering a name) — even though `name` is a string, not a boolean `True` or `False` value.

The condition `if name` seems to behave the same as the condition `if name != ""` (`name` will contain the empty string if enter without typing any value).

So the code above behaves the same as:

````
name = input("What is your name? > ")

if name != "":
    print("Hello, " + name + "!")
else:
    print("You didn't enter a name.")
````

How is this possible?

## `bool()`

Well it turns out that if you put any value in an `if`-condition, Python will try to convert that value to a boolean. So how does Python do this? By using the `bool()` function.

As you may remember from our section on arithmetic and [data types](../1-basic-syntax/5-arithmetic-types.ipynb), we can convert values of different types to one another. The one we probably use the most often is `int()` to convert a string to an integer.

Just as have the functions `int()`, `float()` and `str()` to convert values to integers, floats and strings respectively, there is also a function `bool()` that converts values to boolean values (`True` or `False`).

And it turns out this function is very lenient in what it accepts as `True`-like or `False`-like values. You can see for yourself:

````
print(bool(""))       # False — empty string
print(bool("Alice"))  # True  — non-empty string
print(bool(0))        # False — zero
print(bool(3))        # True  — non-zero number
print(bool(-3))       # True  — negative, but still non-zero number
````

We can see that `bool()` converts:

- all strings to `True`, **except for the empty string `""`** which which it converts to `False`.
- **all non-zero numbers** to `True`, and only `0` to `False`.

You can test any value for yourself to verify that this is the case.

The result of `bool()` can be summarized in this table:

|| `True` | `False` |
|---|---|---|
|Numbers| Any non-zero number | `0` |
|Strings| Any non-empty string | `""` (empty string) |

We call values that `bool()` converts to `True` **truthy** and values that are converted to `False` **Falsy**.

### Truthy != `True`

Note that we are not saying a non-empty string is equal to the value `True`. For example:

````
print("Alice" == True)
````

will print `False`.

What we are saying is that the `bool()` function will convert it to the value `True`:

````
print(bool("Alice") == True)
````

## Truthy and falsy in `if`-conditions

So why are we making such a big fuss about this `bool()` function?

Because whenever you write an `if`-statement, Python actually uses the `bool()` function under the hood to make sure your `if`-condition gets transformed to a boolean value.

So when you write

````
name = "James"
if name:
    print("Hello " + name + "!")
````

What Python is really (implicitly) doing is:

````
name = "James"
if bool(name):
    print("Hello " + name + "!")
````

Because we saw above the `bool()` function can transform any value into a boolean value, this means **you can use any value as an `if`-condition**.

## Bugs that are hard to spot

Though it might seem like a cool thing that Python behaves very flexibly in this way, this behaviour can sometimes also result in **bugs that are hard to spot**.

Let's take a look at the following code that is supposed to check whether someone is either 18 or 21 years old:

````
age = int(input("Enter your age: > "))

if age == 18 or 21:
    print("It's time for an extravagant birthday party this year!")
````

This looks reasonable, but run it with an age like `30` and you'll find it *always* prints the message. How can this possibly be the case?

Intuitively, you might think that the above condition is the same as:

```python .noeval
if age == 18 or age == 21:
```

But actually, how python reads this is:

```python .noeval
if (age == 18) or (21):
```

To Python there are two expressions that it needs to combine with the `or`-operator, `age == 18` and `21`, and if one of them is not a boolean value already, logical operators like `and` and `or` will also implicitly use the `bool()` function to convert these values to booleans.

So the way Python looks at this condition is actually

```python .noeval
if (age == 18) or bool(21):
```

and as we now know, `bool(21)` will always be converted to `True` (because it's a non-zero number), so the whole condition is always `True`, no matter what `age` is.

This is why it's important to always repeat the value you're comparing, on both sides of `and` or `or`:

````
if age == 18 or age == 21:
    print("You're either 18 or 21.")
````

## The takeaway

You don't need to be able to *write* code that relies on truthi- and falsiness, if you feel more comfortable with conditions that are explicit.

However, it's important to be aware of how this works, as it might show up in your code as bugs. Later, you might also encounter this behaviour in other people's code, which is also why it is important to know how this works.

## Extra: `1` and `0` are literally `True` and `False`

There's one more quirk worth knowing. In the beginning of this section we explained that a value being 'truthy' or 'falsy' does not mean this value **is equal to** `True` or `False`. For example:

````
"Alice" == True
````

is `False`, while

````
bool("Alice") == True
````

is `True`.

There are two exceptions to this: `1` and `0` are **actually equal to** `True` and `False`. You can think of this as and `True` and `False` actually being just the numbers `1` and `0` wearing a different name. You can confirm this:

````
print(1 == True)   # True
print(0 == False)  # True
````

But this is *not* the case for arbitrary truthy or falsy values — only `1` and `0` specifically:

````
print("hello" == True)  # False
print(2 == True)        # False
````

So while `bool("hello")` gives `True`, `"hello"` is not literally equal to `True`. This special relationship only holds between `1`/`0` and `True`/`False`.

Again, we don't tell you about this behaviour because you should use it in your code (and you really shouldn't), but just to give you a heads-up for when you might encounter this behaviour by accident in your own code.
