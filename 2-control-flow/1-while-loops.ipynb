# While Loop

## Improving our guessing game

You may remember the first game we ever coded in this course:

````
guess = input("Guess what animal I'm thinking of: > ")
if guess == "dolphin":
    print("You guessed correctly!")
else:
    print("Nope, try again!")
````

To make the following explanation slightly simpler, we're going to **invert** our condition:

````
guess = input("Guess what animal I'm thinking of: > ")
if guess != "dolphin":
    print("Nope, try again!")
else:
    print("You guessed correctly!")
````

As you can see, we inverted `guess == "dolphin"` to `guess != "dolphin"`, which makes this condition **True in all cases that the previous condition was false and vice versa**. However, because we also switched around our print-statements, the functionality is exactly the same. If you need to, you can run this program a couple of times or think about this a little until you're convinced this version works the same way the previous version did.

While this game technically works the way it's intended, it's not ideal: if you guess incorrectly, instead of automatically being able to guess again, you have to restart the program. In this small game that's probably acceptable, but imagine having to restart Call of Duty every time you get killed.

We can improve this by using - instead of `if`, which only executes the code **once** - the `while`-**loop**, which allows us to execute the same piece of code **many times**, **until a stop condition is met**.

````
guess = input("Guess what animal I'm thinking of: > ")
while guess != "dolphin":
    print("Nope, try again!")
    guess = input("Guess what animal I'm thinking of: > ")
print("You guessed correctly!")
````

As you can see by running this program a couple of times (and submitting some wrong guesses), This program does not finish after the first guess. Instead, it keeps looping over the indented block of code (the lines of code starting with 4 spaces). So how does this work?

## While explained

### While vs If

`while` works a lot like the `if`-statement: it evaluates its boolean condition

- **if the condition is True**, the indented block of code following it (which we call the body of the `while`-statement) is executed.
- **if the condition is False**, the indented block of code is not executed. So far, this works just like `if`. However, this is where the similarities end.

### While as a loop

After executing the indented block (in case the condition was true), **instead of** immediately proceeding with the code that follows the indented block (as `if` would do), `while` **does something different**: it jumps back to the top of the `while`-statement and it starts execution from there again:

- It evaluates the boolean condition again
  - If the boolean condition is still true, it executes the indented body again
    - After executing the indented body, it jumps back to the top of the `while`-statement.
  - If the boolean condition is false, it **does not execute the body** (and therefore will not re-evaluate the while-condition).

In other words, the indented block **keeps executing as long as the condition of the `while`-statement is true**. We call the `while`-statement **a loop**, because it loops over the same code multiple times.

### In our guessing game

Knowing this, let's break down the flow of our new guessing game line-by-line. You can check this yourself by executing the code step-by-step as explained in the previous section.

````
guess = input("Guess what animal I'm thinking of: > ")
while guess != "dolphin":
    print("Nope, try again!")
    guess = input("Guess what animal I'm thinking of: > ")
print("You guessed correctly!")
````

1. First the user is asked for a first guess. After this line the variable `guess` will contain the String value of this guess, (Let's say we guess `"shark"`). Then the code proceeds to the next line.
1. On line 2 we have the `while`-statement: `while guess != "dolphin":` this line checks whether the value of `guess` is not equal to the string `"dolphin"`, If it's not equal, the code proceeds to the body of the `while`-loop. If it is equal, the body is simply skipped. As we assume we guessed `"shark"`, we enter the body
   1. first the first line of the body is executed so the message is printed that we should guess again.
   1. Then one line 2 of the body, the code asks for input again. On this line the user can guess a new animal and this new animal will be stored in `guess` (for example `"dog"`)
   1. After the final line of the body, the code jumps back to `while guess != "dolphin"`, which repeats everything starting from step 2 (but with `guess` now containing the new value).
1. Only after the user inputs `"dolphin"` and the condition `guess != "dolphin"` evaluates to False, will the `while`-loop skip the body and proceed to the final line `print("you guessed correctly!"`)

## Counting down

Let's revisit an example of one of our first lectures where we printed out a countdown sequence:

````
print(10)
print(9)
print(8)
print(7)
print(6)
print(5)
print(4)
print(3)
print(2)
print(1)
print("Lift off!")
````

This can be written in a much shorter way using a while-loop:

````
counter = 10
while counter > 0:
    print(counter)
    counter = counter - 1
print("Lift off!")
````

Let's break down what happens:

- We initialize `counter` to `10`. This is the number that will be printed over and over.
- We arrive at the `while` loop. The condition `counter > 0` is evaluated; this yields `True`. (as 10 is bigger than 0) Because the condition is true, the loop's body is executed.
  - The value of `counter` is printed, i.e., `10`.
  - `counter` gets decremented by `1` and is now equal to `9`.
- Execution jumps back to the `while` line.
- The condition is re-evaluated: is `counter` still bigger than `0`? The answer is yes, causing the loop body to be executed a second time.
  - The value of `counter` (`9`) is printed.
  - `counter` gets decremented from `9` to `8`.
- We jump back to the `while` line. This process keeps getting repeated:
  - `8` is printed and `counter` goes from `8` to `7`.
  - `7` is printed and `counter` goes from `7` to `6`.
  - ...
  - `2` is printed and `counter` goes from `2` to `1`.
  - `1` is printed and `counter` goes from `1` to `0`.
- `counter` has reached `0`. The condition is, once again, evaluated, but this time it yields `False` (as 0 is not bigger than 0).
- Because the condition is not true anymore, the loop ends, i.e., the loop body won't get executed again.
- We end up at the last `print`, which outputs `"Lift off!"`.

## Why use `while`?

You should use while **when you want to repeat a specific part of the code** several times, **until a certain condition has been met**.

Using while has many advantages:

- It allows to write code that you simply cannot write without it (for example letting our guessing game run until we guessed correctly)
- Even when it is technically possible to write code without a `while`-loop, often your code can be a lot shorter with it (eg our lift-off example)

## When is the while condition re-evaluated?

In the previous examples, the variable that is used in the `while`-condition is only changed at the very end of the body.

1. In our guessing game, `guess` gets a new value at on the last line `guess = input(...)`
1. In the lift-off example, the counter is also only decremented on the last line `counter = counter - 1`

What would happen if we change the variable used in the condition at the start or somewhere in the middle of the body, **changing the condition**, like in the following example:(guess what the output will be before running the code)

````
counter = 0

while counter < 2:   # condition is checked here
    print("Start of loop, counter = " + str(counter))
    counter = counter + 1
    print("End of loop, x = " + str(counter))
````

As you can see, even though the variable `counter` is equal to 2 at the end of the last execution of the loop (and therefore the condition `counter < 2` is no longer true), the final print-statement is still executed. This is because the while-condition is only re-evaluated **after the entire body is executed**. In other words, the body of a while loop will execute completely before the condition is checked again.

## Infinite loops

Since a while loop executes until its condition is false, there is a very easy way to let the while-loop execute forever: (press the 'stop' button to stop the code)

````
counter = 0
while True:
    counter += 1
    print("This loop has executed " + str(counter) + " times.")
````

You might want to do this on purpose (eg letting a game loop forever until the user quits it), but most of the time this happens by mistake. For example, imagine we forgot to decrement the counter variable in our lift-off code:

````
counter = 10
while counter > 0:
    print(counter)
print("Lift off!")
````

As we didn't decrement the counter, it will always remain 10, and therefore the condition `counter > 0` will also always remain true, resulting in the behaviour you can see.
