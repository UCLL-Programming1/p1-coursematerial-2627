# Application loops

## Programs that keep running

In one of the previous sections we upgraded our guessing game such that it would not stop until we guessed correctly. Think about most programs: the calculator on your phone, a cash machine at the bank, a music player. These programs have something in common: they don't just do one thing and then quit. After you finish one action, they patiently wait for you to do the next thing — until you actively decide to close or exit them.

We've already seen a small taste of this with our guessing game, which kept asking for a new guess until you got it right. In this notebook we are going to take that idea one step further and build a small program that:

- shows the user a **menu** of things they can do
- **keeps running** and showing that menu after each action
- only **stops** when the user chooses to quit

We'll build on one of the previous exercises, the **fortune teller** (at the end of the topic [nested-if-elif](../1-basic-syntax/8-nested-if-elif.ipynb)). If you didn't make already make this exercise, you could try to do that first.

## A single-use fortune teller

In one of the previous exercises we made the following fortune teller:

````
print("Welcome to the Fortune Teller!")
print()

sign = input("Enter your star sign: > ")

if sign == "Aries":
    print("A bold opportunity is heading your way. Say yes.")
elif sign == "Taurus":
    print("Good things come to those who wait. Keep waiting.")
elif sign == "Gemini":
    print("You will soon hear from an old friend. Maybe two.")
elif sign == "Cancer":
    print("Trust your instincts today. Especially about lunch.")
elif sign == "Leo":
    print("The spotlight is yours. Try not to trip.")
elif sign == "Virgo":
    print("Perfection is overrated. Ship it.")
elif sign == "Libra":
    print("A decision you've been avoiding can wait no longer.")
elif sign == "Scorpio":
    print("Someone is thinking about you right now. Weird, right?")
elif sign == "Sagittarius":
    print("An adventure is closer than you think. Pack light.")
elif sign == "Capricorn":
    print("Hard work pays off. Eventually. Unless your coworker is the boss's son, he will get promoted over you.")
elif sign == "Aquarius":
    print("Your most original idea yet is just around the corner.")
elif sign == "Pisces":
    print("Go with the flow today. The flow knows.")
else:
    print("The stars do not recognise that sign. Check your spelling.")
````

## A fortune teller that keeps going

So what we really want is for the program to ask: *"Would you like another fortune, or are you done?"* — and to keep going until the user decides to quit.

Such a program could run as follows:

- It prints the same welcome message as before
- Then it presents us with our options: asking for a fortune, or simply quitting the application
  - if we ask for a fortune, we have to put in our star sign as before, with the program printing out or fortune
  - if we quit, the application simply stops
- It repeats the previous step until we choose to quit

It could have the following output:

## Adding the loop - Option 1

We can make the program by wrapping the whole program (apart from the welcome message) in a `while`-loop. At the start of each loop we show the user a small menu with their options. When they choose to quit, the loop stops.

Here is one of the ways you could have implemented this:

````
print("Welcome to the Fortune Teller!")
print()

quit = False
while quit == False:
    choice = input("What would you like to do? (F)ortune, (Q)uit: > ")
    if choice == "Q":
        quit = True
    elif choice == "F":
        sign = input("Enter your star sign: > ")

        if sign == "Aries":
            print("A bold opportunity is heading your way. Say yes.")
        elif sign == "Taurus":
            print("Good things come to those who wait. Keep waiting.")
        elif sign == "Gemini":
            print("You will soon hear from an old friend. Maybe two.")
        elif sign == "Cancer":
            print("Trust your instincts today. Especially about lunch.")
        elif sign == "Leo":
            print("The spotlight is yours. Try not to trip.")
        elif sign == "Virgo":
            print("Perfection is overrated. Ship it.")
        elif sign == "Libra":
            print("A decision you've been avoiding can wait no longer.")
        elif sign == "Scorpio":
            print("Someone is thinking about you right now. Weird, right?")
        elif sign == "Sagittarius":
            print("An adventure is closer than you think. Pack light.")
        elif sign == "Capricorn":
            print("Hard work pays off. Eventually.")
        elif sign == "Aquarius":
            print("Your most original idea yet is just around the corner.")
        elif sign == "Pisces":
            print("Go with the flow today. The flow knows.")
        else:
            print("The stars do not recognise that sign. Check your spelling.")
    else:
        print("Unknown option. Please enter F or Q.")
    print()

print("The stars bid you farewell.")
````

Run this code a couple of times to verify it works as intended. Notice how to code will keeping going until you enter `Q`.

## Adding the loop - Option 2

An alternative way of implementing the same thing would be as follows:

````
print("Welcome to the Fortune Teller!")
print()

choice = input("What would you like to do? (F)ortune, (Q)uit: > ")

while choice != "Q":
    if choice == "F":
        sign = input("Enter your star sign: > ")

        if sign == "Aries":
            print("A bold opportunity is heading your way. Say yes.")
        elif sign == "Taurus":
            print("Good things come to those who wait. Keep waiting.")
        elif sign == "Gemini":
            print("You will soon hear from an old friend. Maybe two.")
        elif sign == "Cancer":
            print("Trust your instincts today. Especially about lunch.")
        elif sign == "Leo":
            print("The spotlight is yours. Try not to trip.")
        elif sign == "Virgo":
            print("Perfection is overrated. Ship it.")
        elif sign == "Libra":
            print("A decision you've been avoiding can wait no longer.")
        elif sign == "Scorpio":
            print("Someone is thinking about you right now. Weird, right?")
        elif sign == "Sagittarius":
            print("An adventure is closer than you think. Pack light.")
        elif sign == "Capricorn":
            print("Hard work pays off. Eventually.")
        elif sign == "Aquarius":
            print("Your most original idea yet is just around the corner.")
        elif sign == "Pisces":
            print("Go with the flow today. The flow knows.")
        else:
            print("The stars do not recognise that sign. Check your spelling.")
    else:
        print("Unknown option. Please enter F or Q.")

    print()

    choice = input("What would you like to do? (F)ortune, (Q)uit: > ")

print("The stars bid you farewell.")
````

Again, you can run the code a couple of times to see it behaves the same.

### Option 2 explained

Again, let's focus on the parts that are new, and leave out the code that was the same in our single use fortune teller:

```python .noeval
# printing the welcoming message (same as before)

choice = input("What would you like to do? (F)ortune, (Q)uit: > ")

while choice != "Q":
    if choice == "F":
        # inputting your star sign and getting your fortune (same as before)
    else:
        print("Unknown option. Please enter F or Q.")

    print()

    choice = input("What would you like to do? (F)ortune, (Q)uit: > ")

# printing the bye message (same as before)
```

Let's break it down into the individual parts again:

**1.** the first choice

````
choice = input("What would you like to do? (F)ortune, (Q)uit: > ")
````

In this implementation we first ask the user for their choice **outside of the loop**. The reason we do this is so that we don't have to create the extra variable `quit` to keep track of whether we want to exit the loop or not, as we can simply compare the choice of the user directly in our `while`-condition. The fact that this statement is outside of the loop means it won't get repeated (but we will work around that later in our code).

______________________________________________________________________

**2.** the `while`-loop

```python .noeval
while choice != "Q":
```

Again we have a `while`-loop that will repeat our fortune-telling code. In this case however, the condition can simply be whether the last choice the user entered is equal to `Q`. In our first implementation (option 1), this was not an option, as the user wouldn't have entered a choice yet before entering the loop the first time. (Take a moment to see if you understand this part).

______________________________________________________________________

**3.** the `if-else` block As in option 1, we again have an `if-else`-block to do the corresponding action based on the users choice. The difference here is that the program doesn't have to take into account whether the user chose `Q`, because the loop would have ended already if they did. (take a moment to understand why that is the case).

```python .noeval
    if choice == "F":
        # inputting your star sign and getting your fortune (same as before)
    else:
        print("Unknown option. Please enter F or Q.")
```

The rest of this `if-else`-block is the same as in option 1.

______________________________________________________________________

**4.** an extra choice input

```python .noeval
    choice = input("What would you like to do? (F)ortune, (Q)uit: > ")
```

If this line wouldn't be present in the code, the loop would repeat again and again, with the first input the user gave (either `F` or `Q` and this `choice` would never change, as it was set before the loop even started). So to give the user the chance to enter a new choice every time the loop runs, we have put this statement here at the end of the loop. If the user then chooses `Q`, the `while`-condition will be `False` and the loop will end. If the user chooses `F`, the loop will be repeated.

## Conclusions

1. In order to create programs that keep going until the users chooses to end the program, we have to wrap the original program in a loop.
1. There are 2 approaches we can use to determine the `while`-condition
   1. Either we create a variable specifically for tracking whether we should exit the loop or not (corresponding to option 1 above)
   1. Or we try to express the loop-condition in a more direct way in relation to other variables of values that already exist in our program. (corresponding to option 2 above) In this case we'll sometimes have to repeat a statement to make sure the loop also works the very first time we enter it (`choice = input(...)` above)
