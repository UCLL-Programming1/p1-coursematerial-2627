# Strings revisited

So far, we've been using strings to let our programs output text to the user and get some input from them in return.

This has been working quite well, but have you noticed that sometimes our programs are a little easy to break?

- In our fortune teller program, type `"aries"` instead of `"Aries"` and suddenly it no longer recognises your sign.
- If we'd write an expense-sharing app and you have to split 10 euro's three ways, Python would print `3.333333333` instead of `€3.33`.

These are the little things that make our programs feel a bit rough to use. In this section we're going to learn a couple of new ways to manipulate strings, to make our programs feel more polished.

By now you know the basics: you can

- combine strings with `+`
- compare them with `==` or `<`
- convert them to numbers with `int()` or `float()`
- print or collect them with `print()` and `input()`.

Let's see what else we can do.

## String operators

You already know `+` for concatenation and the comparison operators. But there are two more string operators worth knowing.

### Repetition: `*`

Imagine you want to print a separator line between sections of output — a row of dashes, say. You could write:

````
print("--------------------------------------------------")
````

But counting dashes by hand is tedious and error-prone. The `*` operator lets you **repeat a string a given number of times**:

````
print("-" * 50)
````

This produces exactly the same output. The `*` operator works with any string and any positive integer:

````
print("ha" * 3)   # hahaha
print("*" * 10)   # **********
````

### Membership: `in`

The `in` operator lets you **check whether one string appears inside another**:

````
print("A" in "Alice")    # True
print("z" in "Alice")    # False
print("ice" in "Alice")  # True
````

This is handy when you want to check whether a user's response contains a particular word, without forcing them to type an exact match:

````
response = input("Do you want to continue? > ")
if "yes" in response:
    print("Great, let's keep going!")
````

This would match `"yes"`, `"yes please"`, `"oh yes definitely"` — anything containing the word.

> Note: as is the case with `==` and `!=` is **case sensitive**, so `"yes" in "Yes, please"` (with a capital letter) would evaluate to `False`!

### Overview of string operators

With the two new operators we've just learned, we can update the table of all the string operators we've covered so far:

| Syntax         | Result     | Description              |
| -------------- | ---------- | ------------------------ |
| `"ab" + "cd"`  | `"abcd"`   | Concatenation            |
| `"ab" * 3`     | `"ababab"` | Repetition               |
| `"a" in "abc"` | `True`     | Membership               |
| `"a" == "a"`   | `True`     | Equality                 |
| `"a" != "a"`   | `False`    | Inequality               |
| `"a" < "b"`    | `True`     | Lexicographic comparison |
| `"a" <= "b"`   | `True`     | Lexicographic comparison |
| `"a" > "b"`    | `False`    | Lexicographic comparison |
| `"a" >= "b"`   | `False`    | Lexicographic comparison |

## String interpolation (f-strings)

By now you've probably written something like this:

````
name = "Alice"
age = 30
print("Hello " + name + ", you are " + str(age) + " years old.")
````

It works, but it's a little awkward. Every time you want to add a variable, we have to split up the string and add a `+`. We also can't forget to convert our numbers to strings with `str()` (if you don't remember why, take away the `str()` in the previous example and run the cell again).

There is a better way to do this.

**F-strings** let you embed variables directly inside a string. You write the string as normal, but with an `f` in front of the opening quote. Anywhere you want to insert a variable, you put its name inside curly braces `{}`:

````
name = "Alice"
age = 30
print(f"Hello {name}, you are {age} years old.")
````

This produces exactly the same output — Python fills in the values for you, and converts numbers to strings automatically. No more `str()` calls.

You can put any expression inside the curly braces, not just variable names:

````
x = 5
age = 18
print(f"5 times 3 is {x * 3}.")
print(f"Next year you will be {age + 1} years old.")
````

Don't forget the `f` in front of the string or the curly braces within it, otherwise Python won't recognize your variables!

````
x = 5
age = 18
print("5 times 3 is {x * 3}.")
print(f"Next year you will be age + 1 years old.")
````

### Formatting options

F-strings have one more trick: **formatting options**. By adding a colon `:` after the variable name inside the curly braces, you can control exactly how the value is displayed.

**Decimal places** — use `:.2f` to show a float with exactly 2 decimal places:

```
price = 3.14159265358979
print(f"The price is {price:.2f} euros.")   # The price is 3.14 euros.
```

This is how you get `3.14` instead of `3.14159...`.

**Zero padding** — use `:02d` to display an integer with at least 2 digits, padding with a zero on the left if needed:

````
hour = 9
minute = 5
print(f"The time is {hour:02d}:{minute:02d}.")   # The time is 09:05.
````

**Text alignment** — use `<` (left), `>` (right), or `^` (centre), followed by a number for the field width:

````
print(f"{'left':<10}|")    # left      |
print(f"{'right':>10}|")   #      right|
print(f"{'centre':^10}|")  #   centre  |
````

This is useful when you want output to line up neatly in columns.

> F-strings are the modern, recommended way to embed variables in strings in Python. They are faster and more readable than concatenation with `+`, and handle type conversions automatically.

## String indexing and slicing

In the exercise `Name Reader` in the section about [application loops](3-application-loop.ipynb) we had to use `name[0]` to get the first letter of a name. We said we'd cover this properly later - that time has come.

### Indexing

Every character in a string has a numbered position called its **index**. The first character is at index `0` (yes, we start counting at `0`, not `1`), the second at index `1`, and so on:

````
word = "Python"
print(word[0])   # P
print(word[1])   # y
print(word[5])   # n
````

You can also count from the **end** of the string using negative indices. Index `-1` gives the last character, `-2` the second-to-last, and so on:

````
word = "Python"
print(word[-1])   # n
print(word[-2])   # o
````

> If you try to access an index that doesn't exist — for example `word[100]` on a six-character string — Python will raise an error.

````
word = "Python"
print(word[100])   # error!
````

### Slicing

To extract more than one character at a time, you can use **slicing**. The syntax is `s[start:stop]`, where `start` is the index to begin at, and `stop` is the index to stop *before* (so the character at index `stop` is not included!):

````
word = "Python"
print(word[0:3])   # Pyt  (characters at positions 0, 1, 2)
print(word[2:5])   # tho  (characters at positions 2, 3, 4)
````

You can leave out `start` to begin from the very beginning, or leave out `stop` to go all the way to the end:

````
print(word[:3])   # Pyt  (from the start, up to but not including index 3)
print(word[3:])   # hon  (from index 3 to the end)
````

> The character at the `stop` index is **not included**. `word[0:3]` gives you positions 0, 1, and 2 — not 3.

## String methods

### What is a method?

You already know standalone functions like `print()`, `input()`, `int()`, and `len()`. These you call by writing the function name and passing your value as an argument: `len(my_string)`.

A **method** is similar to a function, but it belongs to a specific type. Instead of passing the string as an argument, you write the string first, then a dot, then the method name, like so:

````
name = "Alice"
print(name.lower())   # alice
````

The `.lower()` method returns **a copy** of the original string, but will **all letters converted to lowercase**.

As you can see, the value before the dot (`name` here) is still the thing being operated on. It just goes in a different position. Other than that, methods work just like regular functions.

One important thing to keep in mind: **string methods never modify the original string**. This catches a lot of people out. For example:

````
name = "Alice"
name.upper()
print(name)   # still "Alice", not "ALICE"
````

Here `name.upper()` returns a new, fully uppercase string, but we are not assigning that result to any variable. If we want to keep it we should assign to a variable:

````
name = "Alice"
name_upper = name.upper()
print(name_upper)   # ALICE
print(name) # Alice
````

### Overview of useful string methods

Python comes with many built-in methods for strings.. Here is a selection of the most useful ones for the upcoming exercises:

| Method                 | Description                                                       |
| ---------------------- | ----------------------------------------------------------------- |
| `s.lower()`            | Returns a lowercase copy of `s`                                   |
| `s.upper()`            | Returns an uppercase copy of `s`                                  |
| `s.find(subs)`         | Returns the index of `subs` in `s`                                |
| `s.startswith(prefix)` | Checks if `s` starts with `prefix`                                |
| `s.endswith(suffix)`   | Checks if `s` ends with `suffix`                                  |
| `s.strip()`            | Returns a copy with whitespace removed from both ends             |
| `s.ljust(width, fill)` | Returns a left justified copy of size `width` padded with `fill`  |
| `s.rjust(width, fill)` | Returns a right justified copy of size `width` padded with `fill` |

A full list of string methods can be found [online](https://docs.python.org/3/library/stdtypes.html#string-methods), or by running `help(str)` in your Python shell.
