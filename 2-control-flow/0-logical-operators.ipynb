# Logical operators

## Combining conditions

In the last section we saw how we can make our programs do one thing in one case (if a certain condition is met) and another thing in another case. We saw the example:

````
guess = input("Guess what animal I'm thinking of: > ")
if guess == "dolphin":
    print("You guessed correctly!")
else:
    print("Nope, try again!")
````

In this case `guess == "dolphin"` was the condition that we checked. However, sometimes we want to check multiple conditions.

For example, we might want to make another guessing game, where I'm thinking of more than one animal and if you guess any one of them, you win.

Let's say the animals I'm thinking of are `"dolphin"` and `"cat"`. We now need to check two conditions:

1. is your guess equal to `"dolphin"`?
1. is your guess equal to `"cat"`?

If any of these two conditions is `True`, the game should print you win. In the other case you lose.

So basically we need to be able to combine conditions into one single condition. In real life we do this a lot as well. People might say things like:

- If **Colruyt is closed** or if **they don't have any more apples**, just go to Carrefour instead
- If **Barcelona have to sell one more player** or if **they can't find a replacement for Messi**, they will have a hard time this season
- If **I would have trained more** or if **I had been just a bit younger**, I would have no problem running this 10K.
- ...

In the statements above, each time there are two conditions. If even just one of them is true, the whole condition is true and the final instruction applies.

## The `or`-operator

In English, this is what the the word **or** does. In Python, we use `or` as well, we call it **a logical operator**.

We use it as follows:

````
guess = input("Guess what animal I'm thinking of: > ")
if guess == "dolphin" or guess == "cat":
    print("You guessed correctly!")
else:
    print("Nope, try again!")
````

When you run this code a couple of times, you'll find that this does exactly what we want it to. If we guess `"cat"` or if we guess `"dolphin"`, it prints out that we guess correctly, otherwise (else), it prints we should try again.

### Another example: Student discount

Let's say we want to write a script to determine whether someone gets the 'Youth discount'. You get the youth discount when you're either:

- under 18

or

- a student

This program could look like this:

````
age = int(input("How old are you? > "))
student = input("Are you a student? (yes/no): > ")

if age < 18 or student == "yes":
    print("You get the youth student discount!")
else:
    print("Sorry, no discount for you.")
````

which behaves as we specified above (run it until you're convinced)

In this example there are are four combinations that can occur:

|  | Case | Student discount? |
|------|-----------|-------------------|
| 1 | You’re under 18 **and** you’re a student (e.g. a 14-year-old in high school) | ✅ Yes |
| 2 | You’re under 18 **but not** a student (e.g. a 17-year-old who graduated early) | ✅ Yes |
| 3 | You’re **not** under 18 **but still** a student (e.g. a 20-year-old in college) | ✅ Yes |
| 4 | You’re **not** under 18 **and not** a student (e.g. a 31-year-old lecturer at UCLL) | ❌ No |

So out of the 4 possible combination there is only one outcome where you don't get the student discount.

We could represent this more more concisely a table like this:

| Case | under 18 | a student | Student discount|
|------|----------|------------------|----------|
| 1    | Yes     | Yes             | Yes ✅  |
| 2    | Yes     | No            | Yes ✅  |
| 3    | No    | Yes             | Yes ✅  |
| 4    | No    | No            | No ❌ |

Or in Python terms:

| Case | age < 18 | student == "yes" | age < 18 **or** student == "yes" |
|------|----------|------------------|----------|
| 1    | True     | True             | True ✅  |
| 2    | True     | False            | True ✅  |
| 3    | False    | True             | True ✅  |
| 4    | False    | False            | False ❌ |

Looking at these tables you can see that `or` just takes in two boolean values:

1. `age <= 18` (which is either `True` or `False`)
1. `student == "yes"` (which is also `True` or `False`)

and also evaluates to a boolean result, `True` or `False`.

This might be even more obvious by running the following cell:

````
age = 14
student = 'yes'

age < 18 or student == 'yes'
````

### The `or` 'truth table'

In fact, the table above is universally true for the `or`-operator, no matter what the two conditions are:

| Case | Condition 1 | Condition 2 | Result (Condition 1 or Condition 2)|
|------|----------|------------------|----------|
| 1    | True     | True             | True ✅  |
| 2    | True     | False            | True ✅  |
| 3    | False    | True             | True ✅  |
| 4    | False    | False            | False ❌ |

whether Condition 1 could represent "being under 18" or "that our guess was dolphin" (or any other condition), and where Condition 2 could represent "being a student" or "our guess being cat" (or any other condition),

- if **both** conditions are True, the combined condition will be `True`
- if **only one** of these conditions are `True`, the combined condition will still be `True`
- and only if **both** of these conditions are `False`, will the combined condition be `False`

The four combinations and their results can also be directly evaluated in Python:

````
True or True
````

````
True or False
````

````
False or True
````

````
False or False
````

## The `and`-operator

Imagine the requirements for getting a student discount are actually different from what we specified above. Instead of being under 18 **or** being student, you actually have to be both under 18 **and** you have to be a student. So if you're student, but not under 18 (the example of the 20 year-old college student), you no longer get a student discount. Neither does the 17 year-old early graduate. The new truth table for this situation looks like:

| Case | under 18 | a student | Student discount|
|------|----------|------------------|----------|
| 1    | True     | True             | True ✅  |
| 2    | True     | False            | False ❌ |
| 3    | False    | True             | False ❌ |
| 4    | False    | False            | False ❌ |

As you see, this table is far more strict: instead of the **3 out of 4** cases where the result is `True`, now **only 1** out of 4 cases is `True`.

As you might guess from the title of this section, we also have a logical operator to represent this situation: the `and`-operator. It again comes from the way we talk about these situations in natural language, eg "For you to get this discount, you have to be under 18 **and** you have to be a student.

Our new program that satisfies the change in requirements is exactly the same as the previous one, except for the operator used:

````
age = int(input("How old are you? > "))
student = input("Are you a student? (yes/no): > ")

if age < 18 and student == "yes":
    print("You get the youth student discount!")
else:
    print("Sorry, no discount for you.")
````

### The `and` 'truth table'

As with `or`, it doesn't really matter what the two boolean conditions represent, the result of the operator will always be the same:

| Case | Condition 1 | Condition 2 | Result (Condition 1 and Condition 2)|
|------|----------|------------------|----------|
| 1    | True     | True             | True ✅  |
| 2    | True     | False            | False ❌ |
| 3    | False    | True             | False ❌ |
| 4    | False    | False            | False ❌ |

and again you can verify this directly in Python:

````
True and True
````

````
True and False
````

etc

## The `not`-operator

The final operator is the easiest one, it serves to check whether a condition is **not** true. It a **unary** operator, which means it only operates on **a single value** and it is for any boolean expression what `!=` is for `==`. It simply inverts all `True` values to `False` values, and `False` values to `True` values:

````
15 < 18
````

````
not 15 < 18
````

Of course, a lot of the time you can invert the operation itself, for example: `not (15 < 18)` will give the same result as `15 >= 18`. Whether one or the other is the most natural to write, depends from situation to situation.
