## Exercise 5.1: Expense Tracker

Keeping track of your expenses is an important part of managing your finances. In this exercise, you will build a simple expense tracker that stores expenses in a file and allows the user to view useful information about their spending.

All expenses are stored in a file called `expenses.txt`. 

Each line in the file contains:
- A category
- An amount

Both values are separated by a comma.

Example:

```none
Food,15
Transport,4
Food,8
Games,20
```

An expense is a specific occurrence where you spent money: if I have lunch and spend €15, I add an entry for `Food,15` to the list. 
If then buy groceries for dinner and I pay €8, I add another entry for `Food` but this time for €8, thus resulting in `Food,8`.

For your expense tracker, you need to write the following functions:
- `load_expenses(filename)` which reads all expenses from the file and returns them as a list
- `save_expenses(filename, expenses)` which writes all expenses back to the file
- `total_spent(expenses)` which returns the total amount spent across all expenses.
- `amount_for_category(expenses, category)` which returns the total amount spent in a specific category.
- `largest_expense(expenses)` which returns the largest individual expense (so not the category total)

Write an application that allows the user to interact with the expenses tracker.
When the program starts, it should load all expenses from `expenses.txt`.
The user can then choose one of the following options: 

```none

(L) List expenses
(T) Total spent
(C) Category total
(H) Highest expense

```

The menu should with all the options to perform should appear when you start the program.

## Exercise 5.2: Connect Four

Connect Four is a classical game where two players take turns dropping pieces into a vertical game board.
When a piece is dropped into a column, it falls down as far as possible and lands on
top of any pieces that are already there.

In our programmed version of the game, one player uses `X` pieces and the other uses `O` pieces.

The goal is to get **four of your own pieces in a row**:
- horizontally,
- vertically,
- or diagonally.

The game board consists of **7 columns** and **6 rows**.

Initially the board is empty:

```none
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
```

Here is how the game works:

- The program starts with an empty board.
- Player `X` takes the first turn.
- The game repeats the following until someone wins:
  - The board is printed.
  - The current player is asked:
    ```none
    Player X, choose a column (1-7):
    ```
    or
    ```none
    Player O, choose a column (1-7):
    ```
  - If the column number is invalid (not between `1` and `7`), print:
    ```none
    Invalid column!
    ```
    and ask the same player again.
  - If the column is full, print:
    ```none
    That column is full!
    ```
    and ask the same player again.
  - Otherwise, place the piece in the chosen column.
  - After placing the piece:
    - If the current player has four in a row, print:
      ```none
      Player X wins!
      ```
      (or Player O) and end the game.
    - Otherwise the turn switches to the other player.

The board should always be printed using spaces between the symbols.

For example:

```none
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
X O - - - - -
```

## Exercise 5.2a: Connect Four : the game board
Before we can determine who wins a game of Connect Four, we first need to build the game board and allow players to place pieces on it.

In this exercise, you will implement the main game loop together with a few helper functions.

The board consists of **7 columns** and **6 rows**.

An empty board looks as follows:

```none
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
```

Player `X` starts the game. After every move, the turn switches between player `X` and player `O`.

A player chooses a column to drop a piece into. The piece should fall down to the lowest available position in that column.

For example, if the board looks like:

```none
- - - - - - -
- - - - - - -
- - - - - - -
- - - - - - -
O - - - - - -
X - - - - - -
```

and player `X` drops a piece in column `1`, the board becomes:

```none
- - - - - - -
- - - - - - -
- - - - - - -
X - - - - - -
O - - - - - -
X - - - - - -
```

First, you need to write some core functions that you will be using a lot in the game:
- `print_board(board)` that prints the board in its current state
- `drop_piece(board, column, piece)` that drops a piece (`"X"` or `"O"`) into a column and modifies the board. The function should return `True` if it was successfully placed and `False` if the column was already full.
- `board_full(board)` which checks if the board is full (all the columns have been fully filled up). Returns `True` if full, else `False`.

Also write the main program which does the following:

1. Creates an empty Connect Four board.
2. Starts with player `X`.
3. Repeatedly:
   - Prints the board.
   - Asks the current player for a column number between `1` and `7`.
   - Drops a piece into the chosen column.
   - If the column is full, print:

     ```none
     That column is full!
     ```

     and let the same player try again.
   - Otherwise switch to the other player.

4. The game ends when the board is full.

When the board becomes full, print:

```none
The board is full!
```

and stop the program.

**Do not worry about determining a winner yet.**
That functionality will be added in Part **02b**.
