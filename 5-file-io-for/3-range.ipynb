# Range

When we've introduced the `for`-loop, we've seen how it allows us to more elegantly iterate over an **iterable**. However, sometimes we just want to do an iteration for a given number of steps, instead of iterating over a list for example. Let's say we want to print "Hi" 10 times. In a `while`-loop we would do this as such:

````
i = 0
i_max = 10
while i < i_max:
    print("Hi")
    i += 1
````

In other words, the loop body (printing "Hi" and adding 1 to i) is executed for every value from `0` to `i_max`. With our `for`-loop, the loop body is called for every entry in an iterable, so we would need an iterable that allows us to do something a given amount of times. Fortunately, python has exactly that, an iterable called **range**.

Our previous example would then become:

````
i_max = 10
for i in range(0, i_max):
    print("Hi")
````

The loop shown above makes `i` go through all values of the `range(0, i_max)`, i.e., `0`, `1`, `2`, … `i_max - 1`. Notice that this code is yet again more concise than the `while`-loop example. It's also less error prone, as we don't need to manually manage the incrementing of the index anymore. Note: The variable name `i` can be freely chosen.

### `IMPORTANT`

Note that `range(start, stop)` does not include `stop` itself.

````
for i in range(0, 3):
    print(i)
````

`range`has some more features than the case we've seen thus far. To most straightforward use case is as follows:

- `range(start, stop)` represents all integers from `start` to `stop` (exluding `stop` itself).

We can also introduce a "step size":

- `range(start, stop, step)` goes from `start` to `stop` with increments of `step`.

For example, let's say we only want to print the even numbers between 0 and 10 (including 0 and excluding 10). For this we need to make steps of two:

````
for i in range(0, 10, 2):
    print(i)
````

- Negative `step` allows you to count backwards.

````
for i in range(10, 0, -2):
    print(i)
````

Note: When we go backwards, we notice we print 10 but not 0, while in the other direction we printed 0 and not 10. This is because the first value of your range is included and the second value is excluded, so always bear this in mind when using ranges.

For the typical usecase of starting your range at 0, there is a more concise way to create a range:

- `range(stop)` is the same as `range(0, stop)`.

````
for i in range(3):
    print(i)
````

from random import randrange

def deal_card(deck):
    random_index = randrange(len(deck))
    return deck.pop(random_index)
