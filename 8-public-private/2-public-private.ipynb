# Private Attributes

## Enforcing setter methods: non-public attributes

With our setter methods, we now have a way to consistently make sure we don't accidentally set invalid values to our attributes. However, there is still one downside: **we have to remember to use them**. For example as we just saw, in our class `Course` if we use the setter method, we will get an error when we pass in invalid value:

````
programming_2.set_day_of_week("Every day")
````

However, we can still technically do this:

````
programming_2.day_of_week = "Every day"
print(programming_2.day_of_week)
````

Of course **we are not supposed to**, but in real life, when writing code for lots of projects, **it's easy to forget**.

- classes can have hundreds of lines of code, where the setter methods are not immediately visible
- you might not touch a specific piece of code for months or years, forgetting that you wrote these setter methods for it
- you might not have been the one writing the class in the first place, as real life projects are created by many different programmers

### What can we do?

Well, we could **name our attributes** in a way that **makes it very obvious** that a specific attribute **should not by modified directly**, that instead we should use the setter method to change it, something like: `dont_change_this_attribute_without_setter_day_of_week` and `dont_change_this_attribute_without_setter_slot` where the first part of our attribute name is `dont_change_this_attribute_without_setter_` followed by the actual attribute name. If you would then write:

````
programming_2.dont_change_this_attribute_without_setter_day_of_week = "Every day"
````

it would still work, but it makes it really obvious that you're not supposed to do this.

### Single underscore `_`: non-public attributes

Of course having to to start all of your attributes with `dont_change_this_attribute_without_setter_` is a little bit tedious and hard to read, so instead Python developers came up with a different convention: **starting the attribute name with a single underscore `_`**, so for example `_day_of_week`, and `_slot`.

This **single underscore** doesn't change how your code behaves, but it's simply there to tell other programmers (and future-you) that **they should not use this attribute directly** (instead relying on its setter method).

Our new `Course` class will look as follows:

````
class Course:
    def __init__(self, name, day_of_week, slot):
        self.name = name
        self.set_day_of_week(day_of_week)
        self.set_slot(slot)

    def set_day_of_week(self, new_day_of_week):
        if new_day_of_week not in ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]:
            raise ValueError(f'"{new_day_of_week}" is not a weekday, choose "Monday", "Tuesday", "Wednesday", "Thursday" or "Friday"')
        self._day_of_week = new_day_of_week

    def set_slot(self, new_slot):
        if new_slot not in [1, 2, 3, 4]:
            raise ValueError(f'{new_slot} is not a valid slot, choose 1 (8h-10h), 2 (10h-12h), 3 (13h-15h) or 4 (15h-17h) instead')
        self._slot = new_slot

programming_2 = Course("Programming 2", "Monday", 1)
````

And of course we can still modify the `_day_of_week` attribute directly:

````
programming_2._day_of_week = "Every day"
print(programming_2._day_of_week)
````

but we **now know we're not supposed to**.

> Comparison to other programming languages: If you have experience with other programming languages, you might know of languages where this concept of "private attributes" are (strictly) enforced (such as Java). In Python this is not the case. The idea is that if you don't want to follow best practices, you are probably going to find a way to do so (for example in Java you can use something called 'reflection' to access private fields anyway). What this convention is for, is **to prevent you** from breaking best practices **by mistake**, not to prevent you to do it altogether, if you really want to do it.

> Best practices: At this point it may also seem weird to you that we add all of this complexity just "not to make mistakes by accident". However, as you gain experience programming, you will notice that anything that **catches your (little) mistakes early saves you a lot of time looking for them later**\*.

## 'getter' methods

Let's take another look at our `Course` class:

````
class Course:
    def __init__(self, name, day_of_week, slot):
        self.name = name
        self.set_day_of_week(day_of_week)
        self.set_slot(slot)

    def set_day_of_week(self, new_day_of_week):
        if new_day_of_week not in ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]:
            raise ValueError(f'"{new_day_of_week}" is not a weekday, choose "Monday", "Tuesday", "Wednesday", "Thursday" or "Friday"')
        self._day_of_week = new_day_of_week

    def set_slot(self, new_slot):
        if new_slot not in [1, 2, 3, 4]:
            raise ValueError(f'{new_slot} is not a valid slot, choose 1 (8h-10h), 2 (10h-12h), 3 (13h-15h) or 4 (15h-17h) instead')
        self._slot = new_slot

programming_2 = Course("Programming 2", "Monday", 1)
````

One thing you may have realized, is that if we want to print any of its attributes, we have to use the new names, starting with the underscores. Instead of:

````
print(programming_2.day_of_week) # this won't work because there is no attribute `day_of_week`
````

we have to do:

````
print(programming_2._day_of_week) # with the leading underscore
````

But this is not great, because the whole point of having these private attributes is that we're not supposed to use them directly.

To get around this, we have to create other methods that allow us to get the value of the private attributes:

````
class Course:
    def __init__(self, name, day_of_week, slot):
        self.name = name
        self.set_day_of_week(day_of_week)
        self.set_slot(slot)

    def get_day_of_week(self):
        return self._day_of_week

    def set_day_of_week(self, new_day_of_week):
        if new_day_of_week not in ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]:
            raise ValueError(f'"{new_day_of_week}" is not a weekday, choose "Monday", "Tuesday", "Wednesday", "Thursday" or "Friday"')
        self._day_of_week = new_day_of_week

    def get_slot(self):
        return self._slot

    def set_slot(self, new_slot):
        if new_slot not in [1, 2, 3, 4]:
            raise ValueError(f'{new_slot} is not a valid slot, choose 1 (8h-10h), 2 (10h-12h), 3 (13h-15h) or 4 (15h-17h) instead')
        self._slot = new_slot

programming_2 = Course("Programming 2", "Monday", 1)
````

As you can see, we added two methods `get_day_of_week()` and `get_slot()`. The only purpose of these two methods is to get the values of these private attributes, so that, once we created objects of this class, we can **access its attributes without having to access the private attributes directly**, like so:

````
print(programming_2.get_day_of_week())
print(programming_2.get_slot())
````

This code no longer relies on accessing private attributes, instead using public methods.

At his point, you might think to yourself: Then what's the whole point? If we are going to be able to access these private
attributes through public methods, why not just make the attribute public in the first place? That would save us a lot of
seemingly unnecessary trouble.

Well the difference is that with these new getter-methods, we **still can't change** the value of the private attribute
**without using the setter-method**, we can only **see its value**. For example, we can't do the following:

````
programming_2.get_day_of_week() = "Monday"
````

So we are still protecting ourselves from accidentally changing attributes in ways we were not supposed to, which was
exactly the purpose of making these attributes private.

## getter-setter pairs

Very often, when creating private attributes, like we did above, we will create both a getter method (to get the attributes value) and a setter method (to set it in a safe way). We call these two methods a getter-setter pair.

For the `Pokemon` class from the start of the chapter, this would look as follows:

````
class Pokemon:
    def __init__(self, name, hp, moves):
        self.name = name
        self.moves = moves
        self._hp = hp
        self.max_hp = hp

    def get_hp(self):
        return self._hp

    def set_hp(self, new_hp):
        self._hp = new_hp
        if self._hp < 0:
            self._hp = 0
````
