# Private methods

To close off this chapter, note that not only can you make attributes private, you can make **methods private as well**.

For example:

````
class Thermostat:
    def __init__(self, celsius):
        self.celsius = celsius

    def display(self):
        print(f"{self.celsius}°C / {self._to_fahrenheit()}°F")

    def _to_fahrenheit(self):
        return self.celsius * 9/5 + 32
````

In this example, there is a method `_to_fahrenheit()` that simply converts the value of the `celsius` attribute to fahrenheit.

This didn't **have to** be a method specifically, but there are some reasons that you might want to do it anyway:

- maybe you want to use this conversion in some other places as well, making it convenient that you can simply call the method
- turning this piece of code makes it immediately clear what this piece of code does, while it would probably be a little bit harder to understand what `print(f"{self.celsius}°C / {self.celsius * 9/5 + 32}°F")` would do

So **why make this method private?**

The argument for making this method private is that for the purpose of this (imaginary) class, this method is simply there for us to help us implementing other methods, we don't necessarily want users of our class to use this method themselves.

> Note: The argument for making methods private is probably not as intuitive right now as the argument for private attributes. That is ok. At this point, you don't really need to know when to make methods private or not. We tell you about this simply to let you know it is possible and to give you some broad idea of why it would be useful. You will learn about this in more detail later in other courses.
