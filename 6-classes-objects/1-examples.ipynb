## Note

Write a class `Note`, which will have a `title` and `text` (which are both strings) and some `tags` (which should be a list of strings). The class should be able to be used as follows:

````
note_1 = Note("Use of Classes", "Classes can be used to make my code more readable, by having descriptive attributes", ['Python', 'Programming', 'Classes'])
print(note_1.title) # Use of Classes
print(note_1.text) # Classes can be used to make my code more readable, by having descriptive attributes
print(note_1.tags) # ['Python', 'Programming', 'Classes']
````

### Solution

````
class Note:
    def __init__(self, title, text, tags):
        self.title = title
        self.text = text
        self.tags = tags

note_1 = Note("Use of Classes", "Classes can be used to make my code more readable, by having descriptive attributes", ['Python', 'Programming', 'Classes'])
print(note_1.title) # Use of Classes
print(note_1.text) # Classes can be used to make my code more readable, by having descriptive attributes
print(note_1.tags) # ['Python', 'Programming', 'Classes']
````

One particular thing you might notice about this class is that all the constructor is doing is **assign its arguments to corresponding attributes**. This is quite common for smaller classes.

Note: it is import to realize that in the above example `title` is **not the same as** `self.title` (which is why we have to explicitly assign the value of `title` to `self.title`)

- `title` is just a regular **function argument**. After the constructor finishes, it will **cease to exist**
- `self.title` is an **attribute**, which is **stored on the object `self`** (in this case the object that will be assigned to `note_1`). After the constructor finishes, its value will be available using `note_1.title`

## Person

Write a class `Person`. A person should have a `first_name`, `last_name`, `full_name` and `initials`. However, when creating the object we should only have to pass the first and last name, with `full_name` and `initials` being determined automatically. For example:

````
person_1 = Person("James", "Bond")
print(f"first name: {person_1.first_name}") # James
print(f"last name: {person_1.last_name}") # Bond
print(f"full name: {person_1.full_name}") # James Bond
print(f"initials: {person_1.initials}") # JB
````

### Solution

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

person_1 = Person("James", "Bond")
print(f"first name: {person_1.first_name}") # James
print(f"last name: {person_1.last_name}") # Bond
print(f"full name: {person_1.full_name}") # James Bond
print(f"initials: {person_1.initials}") # JB
````

In this example the first two attributes are simple, their values are simply taken from the two arguments that we pass to the function.

However, the other two attributes **are derived** from those first two arguments.
