#### Our Classes

**`Card`** is created from a string like `"DA"` and has a `suit` and a `rank`:

```python .noeval
card = Card("DA")
print(card.suit)  # D
print(card.rank)  # A
```

- `is_playable(discard_pile)` returns whether this card may be played on the given discard pile:
  it matches if it has the same suit as the pile's current suit, or the same rank as the
  pile's top card, or if it is an 8.
- `__repr__()` returns the card as it is printed in a hand, for example `DA`.

**`Deck`** is created without any arguments and holds all 52 cards in its `cards` attribute:

- `deal_card()` removes a random card from the deck and returns it
- `deal_cards(number)` returns a list of `number` cards, dealt one by one
- `is_empty()` returns whether the deck has run out of cards

**`DiscardPile`** is created with the card that starts the pile. It keeps its `cards` and
the `current_suit` that has to be matched:

- `top_card()` returns the card lying face-up on top
- `add(card, suit)` puts a card on top of the pile and stores `suit` as the new

- `refill_deck(deck)` puts every card of the pile except the top one back into the deck

**`Player`** is created with the cards of their hand, which it keeps in `cards`:

- `find_card(string_value)` returns the card of this hand whose string value is
  `string_value`, or `None` if the player doesn't have that card
- `play_turn(deck, discard_pile)` plays one full turn

Finally, `main()` creates the deck, deals both players five cards, starts the discard pile
and lets the two players take turns until one of them has no cards left.

## Exercise 6.1a: Setting up the game

As a first step, let's rewrite how the game is set up, creating the deck,
giving each player their starting hand, and initializing the discard pile.

### `Card`

Of course before we can start creating a deck of cards, the very first change
to our code will be to create a class `Card` (which will be the same as the
class `PlayingCard` from the theory sections).

Its basic functionality is that it can be created by passing a string
representation of the card, and that we can then inspect its `suit` and
`rank` using the corresponding attributes:

```python .noeval
card = Card("DA")
print(card.suit)  # D
print(card.rank)  # A
```

We should also give it an appropriate `__repr__` implementation, such that when
we print the object itself we get:

```python .noeval
card = Card("DA")
print(card) # DA
```

instead of `<__main__.Card object at 0x10cb90050>`, which we would get if we
didn't implement the `__repr__` method.


### Creating the deck - `Deck`

The first step in our program is to create a deck of cards. In our previous
version of Crazy Eights, we used the function `create_deck()` to do this,
creating a list of cards. However, a deck of cards is a very specific list,
with some very specific functionality (for example dealing cards at random), so
it deserves its own class!

Its basic functionality should be that when created, it should have an attribute
`cards` that is a list of all the cards still in the deck:

```python .noeval
deck = Deck()
print(deck.cards) # [HA, H2, H3, ..., SJ, SQ, SK]
print(len(deck.cards)) # 52
```

Hint: the existing function `create_deck()` already does most of this work
(though you will have to change it to use our new `Card` objects instead of
strings). The big change is where all of this code should be put such that it is
executed when the object is 'constructed'.

### Dealing cards

After we created the deck, we deal the cards to the players. Right now,
`deal_card(deck)` and `deal_cards(deck, number)` are functions. However,
because these functions only really make sense to use with a `Deck` object,
let's turn them into **methods** of the `Deck` class, such that they can be
used as follows:

```python .noeval
deck = Deck()
card = deck.deal_card()
hand = deck.deal_cards(5)
print(card) # C8 (or any other card, depending on what random card you got dealt)
print(hand) # [C9, S10, SK, H7, C3] (or any other sequence of cards, depending on what cards you got dealt)
```

### Discard pile - `DiscardPile`

The discard pile is another "list" of cards, like a deck. However, it behaves
very differently from a deck, as you're not supposed to deal cards from it. In
addition the discard pile always shows its top card (which a deck does not do).
Because it behaves differently, it deserves its own class.

When creating it, we should always give it a single card, which will be the
starting card of the discard pile. Like a deck, it will keep track of all its
cards in an attribute `cards` (which should be a list).

Additionally we should always be able to ask it its **top card**, which we will
allow by giving it a method `top_card()`.

So putting all of this together, we should be able to use this class as follows:

```python .noeval
discard_pile = DiscardPile(Card("SA"))
print(discard_pile.cards) # [SA]
print(discard_pile.top_card()) # SA

# Then, when adding more cards to the discard pile:
discard_pile.cards.append(Card("C8"))
discard_pile.cards.append(Card("HK"))
print(discard_pile.top_card()) # HK
```

### `Player`

In our previous version of the game, players were represented by their "hands",
being a list of cards.

However, because players need to keep track of their hands and play turns, they
also deserve their own class.

Initially it will be very simple: it will just keep track of the cards the
player has in their hand. When creating a player, we should give them their
starting cards:

```python .noeval
deck = Deck()
player = Player(deck.deal_cards(5))
print(player.cards) # [C9, S10, SK, H7, C3] (or any other sequence of cards, depending on what cards you got dealt)
```

## Exercise 6.1b: Playing a turn

As we've set up the game, let's now play a single turn. As it's the player that

Find out what changes you need to make to get this method working.

### `card in cards` - `find_card`

One of the bigger changes you will see is that the check
```python .noeval
elif move not in player_cards:
```
no longer works. Why is that?

Well, `move` is a string value that represents a card, for example `"C8"`. Originally, `player_cards` was a list of
strings too, so checking whether the string `move` was in the list of strings `player_cards` could simply be done by
checking `move in player_cards`.

However, while `move` is still a string, `player_cards` is now a list of `Card` objects, so by definition `move` (as a
string) will never be in this list.

To deal with this, write a new method `find_card(move)` on the class `Player`. This method should find the
`Card` object represented by the string `move`. If there is no such card in the `Player`'s hand, it returns `None`.

The method should be used as follows:
```python .noeval
player = Player([Card("C8"), Card("SK"), Card("H9")])
print(player.find_card("SK")) # SK
print(player.find_card("C8")) # C8
print(player.find_card("S7")) # None
```

Then use this method in the `play_turn` method to replace the check `elif move not in player_cards` with something
that works instead.

### `is_playable`

Secondly, as we saw in the theory section, the `is_playable` function should be adapted to work with our new `Card` objects.

However, here we will go one step further. In the theory section, we simply turned the function
```python .noeval
def is_playable(card, top_card, top_suit):
    return card[0] == top_suit or card[1:] == top_card[1:] or card[1:] == "8"
```
into a **new function**:

```python .noeval
def is_playable(card, top_card, top_suit):
    return card.suit == top_suit or card.rank == top_card.rank or card.rank == "8"
```

However, this function is clearly functionality that is only applicable to `Card` objects (the first argument should
always be an object of the `Card` class), so let's turn this function into **a method** on the class `Card` instead.

As such, it would be used as follows:

```python .noeval
ace_of_spades = Card("SA")
eight_of_clubs = Card("C8")
print(ace_of_spades.is_playable(eight_of_clubs, "C")) # False
```

## Exercise 6.1d: Letting the deck and the discard pile do their own work

We now have a working game again. We used classes to do so. So are we done?

### `.is_empty()` and `.refill_deck(deck)`

Take another look at `play_turn`, and in particular at what it
does when a player draws the last card of the deck:

```python .noeval
if len(deck.cards) == 0:
    while len(discard_pile.cards) > 1:
        deck.cards.append(discard_pile.cards.pop(0))
```

This block of code consists of two parts:
- checking whether the deck is empty by comparing its length to `0`
- removing all the cards but the last one from the discard pile (using `pop()`) and appending them back to the deck

Nothing about this is wrong, but this is a lot of code in a method of the class `Player`
to deal with the logic of `Deck` and `DiscardPile`. What tends to make code easier to read
is that when we're looking at the methods in one class, we don't have to understand too much about
how other classes work exactly.

Imagine this code would instead be:
```python .noeval
if deck.is_empty():
    discard_pile.refill_deck(deck)
```

This is a lot easier to read, and makes it immediately clear that if the deck is empty, we'll
simply refill it using the discard pile. How exactly this is done might not concern us. But if
it does, we can of course still check out these methods to see how they work.

Of course the methods used above don't exist yet, but we can create them!

So create the methods `is_empty()` and `refill_deck(deck)` on the appropriate classes such that

## Exercise 6.1e: Cleaning up what's left

We are almost done. However, there is one last peculiarity that we could still resolve in our code.

### The current suit

The current suit is first determined in `main()` and passed to `play_turn()`.
After that each successive call to `play_turn` needs the updated value of the
current suit. When we originally wrote this program without classes, the best
way to do that was to return that updated value, and then pass it back again as
an argument to the next call of `play_turn()`.

However, one of the advantages of using classes and objects is that we can
store information in them. For example, we don't have to pass the top card of
the discard pile as an explicit argument to `play_turn()`, because the
`DiscardPile` object that we pass as an argument already has this information.

Can we do the same for the current suit?

We can, we just need to determine where it would make the most sense to save it.

As the current suit is most often determined by the top card of the discard
pile, we could argue that `DiscardPile` should keep track of this information.
So let's do that.

Create an attribute `current_suit` on the class `DiscardPile` that keeps track
of this value. With this attribute created, you can make the following changes
to the code:
- when you put a card on the discard pile, update the `current_suit` of the discard pile as well.
- turn `.play_turn(deck, discard_pile, current_suit)` into `.play_turn(deck, discard_pile)` (making the necessary changes everywhere to make this work)
- remove the variable `current_suit` from `main()`

### `DiscardPile.add()`

After making the previous changes, we are now in a situation where every time we put
a card on the discard pile we should update its attribute `current_suit` as well.

Since we will **always** need to do these two things together, it makes sense
to create a method for it.

Create a method `add(card, suit)` on the class `DiscardPile` that accepts a card
and a suit argument, and appends the card to its `cards` and stores `suit`
in its `current_suit` attribute. It should be used as follows:

```python .noeval
pile = DiscardPile(Card("C3"))
pile.add(Card("H8"), "S")      # an 8 was played and the player chose spades
print(pile.top_card())         # H8
print(pile.current_suit)       # S
```

### `.is_playable()`

Having made the previous changes, `.is_playable()` is now in a somewhat interesting
position: it takes the arguments `top_card` and `current_suit`, but both of
these arguments are pieces of information that are now held by the same
`DiscardPile` object.

So instead of passing these arguments separately, we can simply pass
the entire `DiscardPile` object and derive these values from that object.

Turn `.is_playable(top_card, current_suit)` into **`.is_playable(discard_pile)`**,
which should still contain the same logic, but getting the necessary information
from the `DiscardPile` argument instead:

```python .noeval
pile = DiscardPile(Card("C3"))
pile.add(Card("H8"), "C")
print(Card("C10").is_playable(pile))  # True  - same suit
print(Card("H7").is_playable(pile))   # False - neither the suit nor the rank matches
```

#### A variable number of players

So how do we support more players in our game? Before reading on, try to think about this challenge yourself first. If
you feel like you have an idea, try it out and see what happens. Then if you're all out of ideas, come back here for
some inspiration.

The first idea that might come to mind is to simply repeat the code that we already have for two players, but now
for however many players we get. For two players we have the loop:

```python .noeval
game_over = False
while not game_over:
    print("Player 1 Turn!")
    player_1.play_turn(deck, discard_pile)
    if not player_1.cards:
        print("Player 1 wins!")
        game_over = True
    else:
        print("Player 2 Turn!")
        player_2.play_turn(deck, discard_pile)
        if not player_2.cards:
            print("Player 2 wins!")
            game_over = True
```

So if we have 4 players, we could write:

```python .noeval
game_over = False
while not game_over:
    print("Player 1 Turn!")
    player_1.play_turn(deck, discard_pile)
    if not player_1.cards:
        print("Player 1 wins!")
        game_over = True
    else:
        print("Player 2 Turn!")
        player_2.play_turn(deck, discard_pile)
        if not player_2.cards:
            print("Player 2 wins!")
            game_over = True
        else:
            print("Player 3 Turn!")
            player_3.play_turn(deck, discard_pile)
            if not player_3.cards:
                print("Player 3 wins!")
                game_over = True
            else:
                print("Player 4 Turn!")
                player_4.play_turn(deck, discard_pile)
                if not player_4.cards:
                    print("Player 4 wins!")
                    game_over = True
```

It's a bit long, but does the trick, right?

Well, actually no. Apart from this code not being a very elegant solution (long, a lot of code repetition, many nested
`if`-`else` statements), it also **simply doesn't work**.

Note that the assignment doesn't ask you to make a **4-player** game, nor a **3-player** game. It asks for a **variable
number** of players. That means that we **simply don't know in advance** how many players we will have. If we wrote this
code and the user playing it enters `5`, we don't have enough `if`-blocks.

So we need a different system.

If you want to think about it a little more before getting the final hint, this is the time! If you want the final hint,
you can read the next paragraph.

Though we won't give you the complete solution outright, the last hint we will give is that whenever you don't know in advance
how many of something (in this case `Player`s) you will have, you can't store them in a set of variables (as you don't
know how many you need to create up front). Instead, you need to store them in something that is flexible in the number
of things it allows you to store.

## Exercise 6.3: Crazy Eights: keeping score

What good is winning if you can't rub it in people's faces that you won more than them?

So let's expand our game one final time, this time allowing it to keep track of the number of
times everyone won.

Of course we know that whenever a program finishes, it 'forgets' everything that happened,
starting with a blank slate the next time it starts. So to save the scores over time, we will
keep them saved in a file.

Next to the program there is a file [scores.txt](milestone-exercises/scores.txt), which will hold the wins of
all the players in all the games they've played. This file will have one line for every player,
where each line is of the form `{player name}:{number of games won}`, for example:

```none
Alice:3
Bob:1
```

Note that players who didn't win any games are not in the file.

This will allow the game to keep track of all the wins, and print this information when playing a game:

```none
How many players are playing? > 3
Enter the name for player 1: > Alice
Current number of wins for Alice: 3
Enter the name for player 2: > Bob
Current number of wins for Bob: 1
Enter the name for player 3: > Carol
Current number of wins for Carol: 0
Alice's turn!
Top card: CK
Top suit: C
Your hand: [D8, H10, C2, S6, H4]
Play card or draw (D): > d8
Choose suit (H D C S): > S

... (the rest of the game is played here)

Alice's turn!
Top card: C4
Top suit: C
Your hand: [C2]
Play card or draw (D): > c2

Alice wins! They now have 4 wins!
```
where `3`, the three names, and the cards and suits typed after them are input provided by the user.

After this game, `scores.txt` would hold:

```none
Alice:4
Bob:1
```

> Note: the cards are dealt at random, so your own game will not look like the one above.

### In `main()`

In `main()` we will need to do a couple of things:
1. read the current score information before the game starts
2. whenever a player enters their name, print out how many games they've won so far
3. when a player wins another game their score should be updated by 1, with their new total being printed
4. at the end of the game, the score information should be written to the file that contains the score information

This is quite a lot of functionality, so let's create a new class that lets us do all of that
easily.

### `Scoreboard`

We will create a class `Scoreboard` that keeps track of the number of wins at all times, and
lets us write that information to a file, and read it again at the start of a new game.

Create this class and implement the following methods:
1. the **constructor**: takes a filename as argument, and initializes the `Scoreboard` based on the information in that file
2. `get_number_of_wins(name)`: returns how many games the player with `name` has won. Somebody who isn't on the scoreboard yet has won `0` games.
3. `add_win(name)`: adds a win for the player with name `name`, adding them to the scoreboard if this was their first.
4. `save()`: writes the whole scoreboard back to the file it read, in the same `name:wins` format it was read from.

In summary the class should be usable as follows:

```python .noeval
scoreboard = Scoreboard("scores.txt")
print(scoreboard.get_number_of_wins("Alice"))  # 3
print(scoreboard.get_number_of_wins("Carol"))  # 0
scoreboard.add_win("Carol")
print(scoreboard.get_number_of_wins("Carol"))  # 1
scoreboard.save()
```
(After this run `scores.txt` should contain a line for `Carol` with `1` win)
