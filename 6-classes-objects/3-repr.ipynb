# `__repr__`

Revisiting the `Person` class we wrote earlier in this chapter, take a look what happens when we try to prints objects of that class directly. What do you think is going to happen?

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

person_1 = Person("James", "Bond")
print(person_1)  # <---- what will happen here?
````

It prints a weird string, something like `<__main__.Person object at 0x10cb90050>`. If you look at it closely it is at least saying that the thing that was printed was a 'Person object'. The weird number (it's a hexadecimal number) that comes after it, is the memory location in which it is stored, but that information is pretty useless to us. This is the **default string represenation of any object**.

If we create and print another object:

````
person_2 = Person("Amelie", "Poulain")
print(person_2)
````

we get a similar output, and at least we see that the number is different, so if we remember that number we could differentiate between those two objects.

However, often times when we are debugging our code to look for mistakes, we want to print intermedary results or use the debugger to see what our current variables are. Getting this hard-to-understand representation is not very convenient when trying to do this.

Fortunately we can change this.

## Implementing the `__repr__` method

To change the string representation of objects of a particular class, we can implement the `__repr__` method. If you include this method in your class, every time Python needs a string representation of an object, it will run this method and use its return value as the string representation.

For example, for the class `Person` we could do:

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

    def __repr__(self):
        return self.full_name

person_1 = Person("James", "Bond")
print(person_1)
person_2 = Person("Amelie", "Poulain")
print(person_2)
````

Now when printing an object of the `Person` class, Python prints the `full_name`, because that is the value that our `__repr__` method returns.

A couple of notes about `__repr__`:

- `__repr__` is another one of these methods that starts with two underscores, just like the constructor `__init__`. In Python, methods that start and end with two underscores are called **'dunder'** (from double underscore) methods. You will almost never call these methods explicitly yourself, but Python uses them under the hood for very specific purposes: in the case of `__init__` to **init**ialize the object and in the case of `__repr__` to get a string **repr**esentation of the object (hence their names).
- `__repr__` only has one argument, `self`. To create the string representation, we can create any string using any of the attributes that are stored on the object.

## `__repr__` convention and guidelines

Of course we could let `__repr__` return any string value, for example:

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

    def __repr__(self):
        return f"first name: {self.first_name}, last name: {self.last_name}"

person_1 = Person("James", "Bond")
print(person_1)
person_2 = Person("Amelie", "Poulain")
print(person_2)
````

But typically we want to include the following things:

- **The class** the object belongs to (if you don't do this, it might look like person_1 is just a string)
- Enough information to make the object **uniquely identifiable**. In the case of `Person` this would mean that at least `first_name` and `last_name` should be included

### The `__repr__` convention

The Pythonic convention on how to implement `__repr__` is to return a string that represents the code to create the object, for example for the class `Person`:

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

    def __repr__(self):
        return f'Person("{self.first_name}", "{self.last_name}")'

person_1 = Person("James", "Bond")
print(person_1)
person_2 = Person("Amelie", "Poulain")
print(person_2)
````

As, you can see the printed values `Person("James", "Bond")` and `Person("Amelie", "Poulain")` are exactly how we created these objects in the first place. Using this convention makes it clear what class the object belongs to and about which object specifically we are talking about.

> Note: This is just a convention. You don't have to use it this way, but we tell you this so that you (1) know about it and (2) have a format to use if you don't have a lot of inspiration yourself. But you **don't have to** stick to this convention at all times

## `__str__`

When reading code that comes from other sources, you might encounter the method `__str__`, that seems to do exactly the same thing as `__repr__`:

````
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.full_name = f"{first_name} {last_name}"
        self.initials = f"{first_name[0]}{last_name[0]}"

    def __str__(self):
        return f'Person("{self.first_name}", "{self.last_name}")'

person_1 = Person("James", "Bond")
print(person_1)
person_2 = Person("Amelie", "Poulain")
print(person_2)
````

However, there is a subtle difference that you can observe while running the following cell:

````
print([person_1, person_2])
````

If you run the previous two cells you'll see that for objects that are part of a list (and also in some other places such as the debugging tools) `__str__` is not used to convert the object to a string. If you change the cell back to using `__repr__` and run the cells again, you'll see that `__repr__` **does work** in this case as well.

We're not going into detail about why these two similar methods exist (that's for next semester). For now it's good enough to know that using `__repr__` will work in all cases. We are only telling you about `__str__` because we know you might encounter this method in other sources and to let you know you shouldn't use it for now.

## Conclusion

`__repr__` allows us to define a more useful `String` reprentation for objects of our classes, which comes in handy when debugging our code.
