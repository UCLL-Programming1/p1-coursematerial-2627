## Crazy Eights

As a refresher, Crazy Eights is a simple card game that is a lot like UNO, but played with a regular deck of cards. The rules of the game are pretty simple:

- Every player starts with 5 cards in their hand
- Every turn a player must do one of two things:
  - they either play one of their cards and put it face-up on the discard pile
  - if they are unable (or unwilling) to play a card, they have to draw a card
- A player can only play a specific card if that card **matches either the suit** (hearts, spades, diamonds or clubs) **or the rank** (A, 2, 3, 4...) of **the last card that was played**
- The only exceptions to the previous rule are 8's (which are crazy) and can be played on top of any previous card.
- In addition, when playing an 8, the player can choose which suit the next card should have. (So 8's behave like color-changing cards in UNO) This **overrides the suit of the last card that was played**.

### Checking valid plays - `is_playable`

One of the key functions we wrote for this game was the function `is_playable(card, top_card, top_suit)`, which checked
whether you were allowed to play a given card (`card`) on a given top card of the discard pile (`top_card`), having to
match suit `top_suit` (which can be different from the suit of `top_card` due to 'crazy eights'). We wrote this function
as follows:

````
def is_playable(card, top_card, top_suit):
    return card[0] == top_suit or card[1:] == top_card[1:] or card[1:] == "8"
````

Can you still tell what this function was doing?

If you can't, don't feel bad, this function is not very easy to read (we will see later why this is).

Let's take a look at what this function was supposed to do, so we have some more context:

#### String Representations

The first important point to know is that in this code, we decided to represent cards by **Strings**. As we needed to
know both the **rank** and **suit** of the card, we decided to:

- take the first letter of every suit: (H)earts, (D)iamonds, (S)pades and (C)lubs
- take the rank (represented by one or two letter strings: `'A', '2', '3', ..., '10', 'J', 'Q', 'K'`)

And add them together in a single string, such that we get string representations such as:

- `'HA'`: (A)ce of (H)earts
- `'D3'`: 3 of (D)iamonds
- `'C10'`: 10 of (C)lubs

#### `is_playable` functionality

With these representations, the function `is_playable` is supposed to work as follows:

````
is_playable('C10', 'C3', 'C') # True  - The card 'C10' (10 of clubs) has the same suit as the `top suit` (Clubs)
is_playable('D9', 'S9', 'S')  # True  - The card 'D9' (9 of diamonds) has the same rank as the card 'S9' (9 of spades)
is_playable('H8', 'SK', 'S')  # True  - 8 of hearts (H8) does not have the same suit as the top suit, nor the same rank as the king of spades (SK), but 8's are crazy, they can be played on top of anything
is_playable('H7', 'SK', 'D')  # False - 7 of hearts (H7) does not have the same suit as the top suit, nor the rank as the king of spades (SK), so cannot be played on top of it.
````

As you can see here – as is specified in the rules of the game – there are three ways that allow a card to be played on top of another:

- either it has the same suit as the top suit
- or it has the same rank as the top card
- or it's simply an `8`

These are exactly the cases the function `is_playable` is checking:

**Checking the suit**: `card[0] == top_suit` - The suit is always the first letter of the 'card', so we compare that first letter with the value `top_suit`

**Checking the rank**: `card[1:] == top_card[1:]` - The rank is whatever comes after the suit, most of the time being a single character (the exception being the rank `'10'`). To select the rest of the string after the first character we use slicing, eg `string[1:]`. We do this both for `card` and `top_card` to compare these values with one another.

**Checking whether the rank of the card is an 8**: `card[1:] == '8'` - In this case we don't need to compare `card` to `top_card` or `top_suit`, we use the same slicing of the last condition to check whether the rank of `card` is equal to `'8'`.

These conditions are combined together with the `or`-operator, such that the result will be `True` if just any one of these conditions is `True`.

## Cards as strings?

Now that we know what the function `is_playable` does, let's take another look at it:

````
def is_playable(card, top_card, top_suit):
    return card[0] == top_suit or card[1:] == top_card[1:] or card[1:] == "8"
````

Were you able to tell what this function did before we did the explanation?

If your answer is no, you're not alone. This is also why we had to spend so much time explaining it.

One of the reasons why this code is a little hard to understand is that, to understand it, you need to know a lot of things:

1. the python slicing syntax `string[1:]`
2. the python syntax for comparing values and the `or`-operator to join these comparisons into a single value.
3. the rules of crazy eights
4. that `card` and `top_card` are strings
5. that the strings are in the format `"{suit}{rank}"` where suit is one character and rank consists of one or more characters

And while we can expect people looking at this code to know point (1), (2) and (3), point (4) and (5) are not really that obvious.

So what makes this code hard to understand is that **we're using `String`s** that are normally used to represent a single piece of text **to represent something that is not text**, namely **playing cards**, which are not random strings of text, but have two very specific characteristics, namely a **suit** and a **rank**.

Apart from `String`s (to represent text), we also learned about different data types already:

- `bool`eans to represent **yes/no**, `True`/`False` values
- `int`egers to represent whole numbers
- `float`s to represent decimal numbers
- `list`s to represent sequences of arbitrary values

What if we also had a `PlayingCard` datatype? A data type that represents playing cards and for which it will be immediately obvious when we are talking about its suit and its rank, something that we could use like:

````
card = PlayingCard("DA")
````

to create a card (similar to `int('10')` to convert a string to an integer, but to convert a string to a `PlayingCard` instead)

and then be able to access the suit and the rank of the card separately, like:

````
print(card.suit) # D
print(card.rank) # A
````

we could then rewrite the `is_playable` function

````
def is_playable(card, top_card, top_suit):
    return card[0] == top_suit or card[1:] == top_card[1:] or card[1:] == "8"
````

to

````
def is_playable(card, top_card, top_suit):
    return card.suit == top_suit or card.rank == top_card.rank or card.rank == "8"
````

which is a lot easier to read (once you're used to this new syntax).

Even if you've never seen this code before, you can see at a glance that the two arguments `card` and `top_card` are apparently this new kind of datatype that have a suit and a rank (and if you're familiar with playing cards in real life you know what that means), that we use to compare the cards to each other.

As you can tell from trying to run this code, this data type does not exist in Python.

But we can create it.

## Classes: custom datatypes

In Python (and in many other programming languages) we call new custom data types that we can create **classes**.

The above `PlayingCard` datatype (class) can be created in Python as follows:

````
class PlayingCard:
    def __init__(self, string_value):
        self.suit = string_value[0]
        self.rank = string_value[1:]

card = PlayingCard("DA")
print(card.suit) # D
print(card.rank) # A
````

As you can see, having this new class definition allows us to create these card 'objects'.

Creating custom datatypes is incredibly common when making programs. Here the example that we use is a new data type for playing cards, but in other programs you might create new data types for:

- `Player`s in games that have more than 1 player, where we might need to keep track of their `name` and `score`
- `Note`s in a note taking app where each note would have a `title`, `text` and perhaps some `tags`
- `Car`s in a racing sim, where each car would have a `name`, `max_speed`, `weight` and other characteristics
- `Pokemon`s in a pokemon simulation game where each pokemon has a `name`, `exp`, `hp`, `moves` etc

Of course which class you need and what attributes they have depend on what information you need to keep track of exactly.

> **Objects and Classes**: We're going to use the terms classes and objects a lot throughout this explanation. So what are classes and objects, and what is the difference between them? Well **a class is a datatype** (like `String` or `int` etc) while **an object is one instance of such a datatype**. For example the string `"James"` is a String **object**, just like `1`, `10` and `42` are all integer objects. A real life analogy is that **'T-shirt' is a type (class)** of clothing, while all the t-shirts you have in your closet are all very specific t-shirts (objects). Or that 'Bike' is a type (class) of vehicle, but that the bike that you drive is a specific bike (object).

> Complicated Syntax? At this point it may seem that this whole class syntax is overly complicated for what we are trying to do. However when we learn more things about classes later, this will all make a lot more sense. For now just sit tight, we'll explain everything one step at a time.

### The class header

On the first line, we are declaring that we are creating a new **class** using the `class` keyword.

- `class` is **for classes** what `def` is **for functions**
- **After** this `class` keyword, we specify the **name of the class** (it can be anything you want, in this case we named our class `PlayingCard`).
  - note: while we use **snake_case** (lowercase letters and underscores) for variable and function names in Python, for classes we typically use **PascalCase** (capitalized words without underscores in between) for class names
- We end the first line with **a colon** `:` (as we do for functions, if-statements, loops etc)

### The class body

The contents of the class (the **class body**) is an indented block. Later we'll see other things that we can write in this body, but in this case the body contains a single element: the **`__init__()` method**, which we also call the **constructor**.

### The constructor `__init__()` as the '`PlayingCard()` function'

When looking at the code card = `PlayingCard("DA")`, you might have expected that we would have to define some kind of function `PlayingCard()` that creates the value that we assign to the variable `card`. But that function seems to be nowhere in sight, or is it?

Actually, when you try to **call a class as if it's a function**, what Python actually does is **call the constructor `__init__()`** that is defined in that class instead.

> We call the method `__init__()` the constructor because it 'constructs' (creates) the object.

But that doesn't seem quite right, there seems to be **a mismatch** in both the **arguments and the return value**:

- `PlayingCard("DA")` only passes along **a single argument** value while `__init__(self, string_value)` **takes two arguments**
- `PlayingCard("DA")` seems like it should return a value (to assign to `card`) while `__init__(self, string_value)` does not return any value.

A detailed explanation of what happens under the hood here would go a little too far, but the easiest way to think about it, is that

- `self` **is the** card **object we are about to create**, but we don't have to pass it to the `__init__()` method explicitly, **Python passes it for us automatically**. Therefore, the only argument we have to pass is `string_value`, for which we passed the value `"DA"`
- Likewise, since the whole purpose of the constructor `__init__()` is to create a new `PlayingCard` object, we **don't have to explicitly return** this object (`self`). **Python returns it automatically** for us **too**.

Python doing all of this stuff automatically may make all of this stuff a little bit harder to understand at first. However once you get used to it, it is actually very convenient.

### Inside of the constructor: attributes

So what happens inside of the constructor? Well because we want to be able to access the `suit` and the `rank` of a card using these specific names, we have to set these values using what we call **attributes**.

**What are attributes?** Attributes are **like variables**, but **tied to one specific object**. In this case we are creating the ace of diamonds so we set:

- the value `'D'` (using `string_value[0]` to select the first character of the argument `"DA"` that we passed) **for the suit**
- the value `'A'` (using `string_value[1:]` to select whatever comes after the first character of the argument `"DA"` that we passed) **for the rank**

These attributes are saved for that specific object we created. So we can print

````
print(card.suit)
print(card.rank)
````

However, when we create many cards, all of them will have different values for their suits and ranks, for example:

````
ten_of_clubs = PlayingCard("C10") # creating the 10 of clubs
ace_of_spades = PlayingCard("SA") # creating the ace of spades
ace_of_clubs = PlayingCard("CA") # creating the ace of clubs
````

and we can print their respective suits and ranks

````
print(ten_of_clubs.rank) # 10
print(ace_of_spades.suit) # S
print(ace_of_clubs.suit) # C
````

Assigning these values in the constructor of our class, **ensures that every object of that class we create, will have a new set of these 2 attributes**.

### `self`

In the beginning it can be confusing what `self` is exactly and how `self.suit = ...` in the constructor results in for example `ace_of_spades.suit`, getting a value and `ace_of_clubs.suit` getting a different value, even though it also uses `self`.

However, as we mention earlier `self` is an argument that **refers to the object currently being created**. So even though it's always called `self`, every time the constructor is called, it refers to a new object being created.

When we run

````
ten_of_clubs = PlayingCard("C10") # creating the 10 of clubs
````

in the constructor `self` will refer to the object that will end up in the variable `ten_of_clubs`.

Likewise, when we run:

````
ace_of_spades = PlayingCard("SA") # creating the ace of spades
````

`self` within the constructor no longer has anything to do with `ten_of_clubs`, it now refers to a new object that is about to assigned to the variable `ace_of_spades`.

> in the constructor `self` is a variable that always refers to **the object being created** and is the value that will be **returned from the constructor**.

If this is a little confusing to you right now, that's ok. It's normal for the concept of `self` to take a while to fully understand.

### `is_playable` with `PlayingCard`

Finally if we use the new function definition that we wrote earlier in this chapter, we can use it on these new `PlayingCard` objects we created.

````
def is_playable(card, top_card, top_suit):
    return card.suit == top_suit or card.rank == top_card.rank or card.rank == "8"

print(is_playable(ten_of_clubs, ace_of_spades, 'S')) # False: no rank or suit in common
print(is_playable(ace_of_clubs, ace_of_spades, 'S')) # True: both have the rank 'Ace'
````

## Modifying attributes after creation

One final thing you need to know about attributes is that they - just like variables - can be modified after they are initialized, for example

````
ace_of_spades.rank = "K"
print(ace_of_spades.rank)
````

Of course for `PlayingCard`s doing this does not make a lot of sense, as a playing card typically does not change after being created. In real life you're not going to take a card, cross out the rank, and write a new one on the card.

But consider a new class `Player`, that we will use to keep track of different players, specifically their names and their scores. This class could look as follows:

````
class Player:
    def __init__(self, player_name):
        self.name = player_name
        self.score = 0

player_1 = Player("James")
print(f"Player name: {player_1.name}, score {player_1.score}")

# after winning a game
player_1.score += 1
print(f"New score: {player_1.score}")
````

A couple of things are noteworthy about this example:

- As we can see we can modify the attributes of objects after they are created
- In this example the attribute `score` is always initialized to `0`. That means that every time we create a `Player` object, its score will always be `0`. In the context of a game this makes sense, as everyone starts a game with zero points, and it's only after playing that this score can increase.
