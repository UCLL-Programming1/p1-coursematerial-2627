# Methods

In this chapter we're going to learn how to make our own methods. However, before we do so, it's good to quickly revisit what a method is exactly.

## Recap: Methods vs Functions

We saw several methods already

- For `String`s:
  - `.upper()`
  - `.startswith()`
  - `.strip()`
  - ...
- for `List`s:
  - `.append()`
  - `.pop()`
  - ...
- for file objects:
  - `.readlines()`
  - `.write()`
  - ...

Likewise, we've seen several functions (`print()`, `input()`, `open()` etc) in addition to having made our own functions as well.

So **what's the difference?**

In essence **a method is a kind of function**

- it has arguments that you need to pass to it when calling
- it can return values (and by default it returns `None`)

However, **when we call a method**, we have to use **slightly different syntax**: instead of writing `function(arg, other_arguments)` for a function, we write `arg.method(other_arguments)` for a method.

- Instead of `upper("abc")` which we'd write if `upper()` were a function, we write `"abc".upper()`
- Instead of `append(list_of_names, "James")` which we'd write if `append()` were a function, we write `list_of_names.append("James")`
- etc

So for a method we write **the first argument before the method name** (separated by a period `.`). The other arguments are passed within the parentheses as usual.

So why do these do these two kinds of functions both exist, even though they're almost the same thing?

One difference is that the first argument of methods always needs to be of a specific class:

- The first (and only) argument of `.upper()` needs to be a `String`
- The first argument of `.append()` needs to be a `list`
- The first argument of `.write()` needs to be a file object.

These **methods are simply not defined for other types of arguments**, which you can easily verify by trying to call them on anything else

````
a_number = 42
a_number.upper()
````

or

````
a_string = "Hello"
a_string.append("world") # should not work
````

And these errors are clear: integers do not have an `upper` attribute and strings don't have an `append` attribute.

Functions, on the other hand, can sometimes work for different values

````
print("Hello")
print(42)
print(True)
````

However, there are also functions that only work for specific arguments (`open()` for example always needs a string as its first argument to tell it which file it should open).

In reality, these two concepts come from two different styles (paradigms) of programming **functional programming** and **object-oriented programming**. Typically we use a mixture of **both functions and methods**. In some cases it's very obvious what you should use, while in other cases it depends on your style of programming. It's a little bit like how in natural language there are often several words that (almost) mean the same thing, it often just depends on your taste which one you would use.

## Method definitions

So how can we create our own methods?

As mentioned before, a method **is a function tied to a specific class** (the class of its first argument). That's why you have to define a method **inside of a class**.

For example, imagine we want a method `.get_card_value()` for our `PlayingCard` class from the introduction example that returns what the value of the card is (where `'A'` is worth `1`, `2`-`10` are worth their respective number, and `'J'`, `'Q'` and `'K'` are all worth `10`).

We wrote this as a function when implementing **Blackjack** in our chapter on functions:

````
def get_card_value(card):
    if card == 'A':
        card_value = 1
    elif card == 'K' or card == 'Q' or card == 'J' or card == '10':
        card_value = 10
    else:
        card_value = int(card)
    return card_value

print(get_card_value('A')) # 1
print(get_card_value('5')) # 5
print(get_card_value('K')) # 10
````

where the argument `card` was simply a `String`.

We can easily convert this function into a method by defining it inside the `PlayingCard` class:

````
class PlayingCard:
    def __init__(self, string_value):
        self.suit = string_value[0]
        self.rank = string_value[1:]

    def get_card_value(card):
        if card.rank == 'A':
            card_value = 1
        elif card.rank == 'K' or card.rank == 'Q' or card.rank == 'J' or card.rank == '10':
            card_value = 10
        else:
            card_value = int(card.rank)
        return card_value

ace_of_spades = PlayingCard("SA")
five_of_clubs = PlayingCard("C5")
king_of_diamonds = PlayingCard("DK")

print(ace_of_spades.get_card_value()) # 1
print(five_of_clubs.get_card_value()) # 5
print(king_of_diamonds.get_card_value()) # 10
````

As you can see, we simply moved the function inside of the class, which turned it into a method. We had to change two things:

- since `card` inside of this method no longer just refered to a `String`, but now refers to a `PlayingCard` object, we have to select its `rank` instead of just comparing it directly to the strings `'A'`, `'K'`, `'Q'` etc
- when calling the method
  - in order to call the method we ofcourse first need to create `PlayingCard` objects we can call the method on
  - what was the first (and in this case only) argument of the function moves to the spot in front of the method (right before the `.`)

## `self` as the first argument of methods

In the above example, the name of the argument of the method was `card`, which is a fairly good name, because it represents a `PlayingCard`.

However, we could call this argument anything, just as we can choose the names for our function arguments. In the case of methods, there is a convention to use in Python though, and that is to call the first argument `self`, so to write:

````
class PlayingCard:
    def __init__(self, string_value):
        self.suit = string_value[0]
        self.rank = string_value[1:]

    def get_card_value(self):
        if self.rank == 'A':
            card_value = 1
        elif self.rank == 'K' or self.rank == 'Q' or self.rank == 'J' or self.rank == '10':
            card_value = 10
        else:
            card_value = int(self.rank)
        return card_value

ace_of_spades = PlayingCard("SA")
print(ace_of_spades.get_card_value()) # 10
````

This does not change the behaviour of the method, as we simply changed what the argument is called, and this argument still refers to whatever value was specified before the period `.` when calling the method.

### Wasn't `self` the object about to be created??

At this point you might be confused, as we previously said that **in the constructor** `self` referred to **the object** that was about **to be created**. And for the constructor this is indeed the case. However the **constructor is just another method**, and the **first argument of any method** is **always an object of the class the method is located in**. And (by convention) we always call this argument `self`. The **difference between** the **constructor** and **other methods** is that:

- (as mentioned in the introductory chapter) in the constructor **this argument is passed implicitly by Python** as **the new object to be created**
- For other methods, we have to **pass this argument explicitly**, using the dot notation `object.method()`

## When to use methods?

So now that we know about both methods and functions, when should we use which?

As a rule of thumb, we will use methods over functions if **one of the arguments of the method/function we are writing is an object of a custom class that we created** ourselves. But don't worry too much about this yet, it will become clearer once we get some practice.

> Note that we **can't add methods for built-in classes** such as `String`, `list` etc (as you need to add the methods inside of the class definition).

## Methods that change the object

So far, we've only seen examples of methods that **return a value** (eg the numerical value of a card). However, methods can also **modify the state** the state of the object. An example of this we already know is that the method `.append()` adds an extra element to the list (so it changes the content of the list).

We could have this type of behaviour in our custom classes as well. Consider the following class `SortedList`, which is a new custom type of list, of which the values should always be sorted:

````
class SortedList:

    def __init__(self):
        self.internal_list = []

    def add(self, element):
        index = 0
        while index < len(self.internal_list) and element > self.internal_list[index]:
            index += 1
        self.internal_list.insert(index, element)

    def get_value(self, index):
        return self.internal_list[index]

    def pop(self, index):
        return self.internal_list.pop(index)

    def length(self):
        return len(self.internal_list)

    def values(self):
        return self.internal_list

new_list = SortedList()
new_list.add(1)
new_list.add(6)
new_list.add(3)
new_list.add(7)
new_list.add(1)
print(new_list.values())
````

As you can see from how this class works, it works a little like a list (you can add elements, see the elements (also by index) etc). The big difference is that, when we add a new element, we wrote some code (in the `.add()` method) to make sure it is added at the correct position such that the list is sorted. It's not super important right now to understand this code.

The main take away is that, inside this `add()` method, **we change the object** (we add an new element to the list it uses to keep track of all the elements).

We can also change the object by assigning new values to the attributes of an object. In the above example, we could for example add a `reset()` method that 'empties' the list:

````
class SortedList:

    def __init__(self):
        self.internal_list = []

    def add(self, element):
        index = 0
        while index < len(self.internal_list) and element > self.internal_list[index]:
            index += 1
        self.internal_list.insert(index, element)

    def get_value(self, index):
        return self.internal_list[index]

    def pop(self, index):
        return self.internal_list.pop(index)

    def length(self):
        return len(self.internal_list)

    def values(self):
        return self.internal_list

    def reset(self):
        self.internal_list = [] # we assign a new empty list to the attribute `self.internal_list`

new_list = SortedList()
new_list.add(1)
new_list.add(6)
new_list.add(3)
new_list.add(7)
new_list.add(1)
print(new_list.values())
new_list.reset()
print(new_list.values())
````
