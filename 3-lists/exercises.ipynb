## Exercise 3.2: Bubblesort

Write a program that first asks for a series of numbers to put in a list and which you should then sort using the Bubble Sort algorithm.
Finally, the application should print the sorted list.

The idea behind Bubble Sort is very simple: we look for consecutive elements that are in the wrong order and swap them around. We repeat this until all the elements are in the correct order.

For example, consider the list [2, 4, 1, 3]:

* The first two elements 2 and 4 are correct.
* The next two elements 4 and 1 are wrong. We swap them, the list is now [2, 1, 4, 3].
* We look at the next pair of consecutive elements, 4 and 3. These too must be swapped, giving rise to 2, 1, 3, 4.
* We start again at the beginning. The first two elements 2 and 1 are wrong → [1, 2, 3, 4].
* We find no more unordered pairs. Sorting is complete.

Note: Be sure to implement the Bubble Sort algorithm yourself and not use any pre-existing sort functions to sort your list.

The execution of the program should look as follows:
```none
Give me a number to put in the list, or (Q) quit adding: > 5
Give me a number to put in the list, or (Q) quit adding: > 1
Give me a number to put in the list, or (Q) quit adding: > 23
Give me a number to put in the list, or (Q) quit adding: > 78
Give me a number to put in the list, or (Q) quit adding: > 12
Give me a number to put in the list, or (Q) quit adding: > 7
Give me a number to put in the list, or (Q) quit adding: > Q
After sorting, the list is: [1, 5, 7, 12, 23, 78]
```
