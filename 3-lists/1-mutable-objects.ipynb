# Lists as the first mutable objects

Consider the following code, what will it print?

```
list_1 = [1, 2, 3]
list_2 = list_1
list_1.append(4)
print(list_2)
```

Wait what? Didn't we see in [6-variables](../1-basic-syntax/6-variables.ipynb) that variables don't stay synchronized? Why is the list referenced by the variable `list_2` updated when updating the list referenced by variable `list_1`?

Well simply because it is the same list.

Going back to the analogy of putting labels on boxes when moving. Imagine if you have a box with your coffee machine in it that has two labels on it, "Coffee Machine" and "Sage Bambino". Then you move down the stairs dropping the box and breaking your coffee machine. Both of the labels are on a box with a broken coffee machine it, that's all there is really to it.

Next, let's take a look at some code that looks a lot like it:

```
list_1 = [1, 2, 3]
list_2 = [1, 2, 3]
list_1.append(4)
print(list_2)
```

If you ran the code, you saw that in this case the second list is not updated? Why is this? Isn't this the same list?

In this case, no.

Even though it looks like the variables `list_1` and `list_2` refer to the same list, we actually created two different lists that just look the same. It's as if you buy two PlayStations in the store. You'll have two boxes that look exactly the same. Before taking them home you could put a label with "My PlaySation" one one and on the other one you could put the label "My brother's PlayStation\`. If you then drop the box with the label "My PlayStation", breaking the PlayStation within it, it's only yours that is broken.

## `==` and `is`

So how can we tell whether two variables refer to the same object, or simply refer to different objects that look the same?

The `==`-operator tells us whether objects look the same:

```
list_1 = [1, 2, 3]
list_2 = [1, 2, 3]
print(list_1 == list_2)
```

But to check whether two objects are really different objects, there is the `is`-operator, which return `False` if two variables refer to different objects:

```
list_1 = [1, 2, 3]
list_2 = [1, 2, 3]
print(list_1 is list_2)
```

But if the two variables really refer to the same object, the `is`-operator will return `True`:

```
list_1 = [1, 2, 3]
list_2 = list_1
print(list_1 is list_2)
```

## Why now? (Mutable vs Immutable)

Why are we only starting to talk about this now? Why didn't we run into issues like this before, using strings or numbers?

Well, let's look at a similar example with strings:

```
string_1 = "hello"
string_2 = string_1
```

What can we do to `string_1` that changes its value, such that the value in `string_2` is also changed?

You might say "Well, we can concatenate an extra string to it". So what will the following code print?

```
string_1 = "hello"
string_2 = string_1
string_1 = string_1 + " world"
print(string_2)
```

Why didn't this change the value of `string_2`, as `list_1.append(4)` changed the value of `list_2`?

Retaking the analogy of using sticky notes as labels for variables (if you need a reminder of what the sticky note
analogy was again, you can reread it in our first section on [variables](../1-basic-syntax/6-variables.ipynb)), let's break down step by step what happens:

1. We apply the label `string_1` to the value `"hello"` (Imagine `"hello"` as a piece of card that has the word 'hello' written on it - with the sticky note `string_1` attached to it)
1. Then we apply another label `string_2` to that same card that has the word "hello" on it. This card now has two sticky notes stuck on it.
1. Then on line 3 we first look what object has the label `string_1` on it and we find the card with the word 'hello' on it. Then we take that card and 'concatenate' it with another card that has the word ' world' on it. Of course here they key is to understand what `+` (concatenation) does. We don't change any of the two strings (or in our analogy the cards with their words), we instead take a new card, and on it we write the result of the concatenation of whatever is on the two cards. Then `string_1 = ...` means that we take the label `string_1` and put it on that new card that we just created and as we saw in the first section on variables, that means we have to take that label of of the previous card. **BUT** the label `string_2` still takes on the original card that has 'hello' on it.

In contrast, `list_1.append(4)` doesn't create a whole new list that has the original items + the one we are adding, it simply adds the new element to the existing list.

In Programming, we say that lists are **mutable** - because we can change what they contain - while strings, numbers and boolean values are **immutable**. There is no way you can change a string value itself, you can only create a new string (using an operation that uses the value of an existing string) and than move the label from that existing string to the new one (but any other label still on the existing string, stays on that string).

So why are some types of values mutable and others not? The short answer is that sometimes it's easier to work with values that are immutable, while sometimes it's more convenient to be able to actually change objects. Later we'll see an immutable kind of list (called a **tuple**), and you'll notice that for a 'list-kind-of-value', it's probably more intuitive to work with them if you can change what is in them, as we can do with lists.

However, don't worry about this too much, the most key takeaway there is simply that **Some datatypes (such as integers, floats, booleans and strings) are immutable**, while **other data types (like lists) are mutable**, and that this can
cause **unexpected results** when the same value is **referenced by more than 1 variable**.
