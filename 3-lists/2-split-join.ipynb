# `.split()` and `.join()`

In one of the examples of Python programs we showed in our introduction class, there was one where you needed to move around using commands like `go north` or `go south` but could choose to quit the game by typing `quit`.

It's clear that in this case, when asking for user input, you'd need to separate the different words the user types, to then check wether the first word the command 'go' or the command 'quit', and if the command is 'go' we'd need to look at the second word to see where to move the user.

But how could we split up the string `"go north"` easily? Well that's exactly what the method `.split()` is for.

## `.split()`

The method `.split()` splits up a string in several parts:

````
user_input = "go north"
print(user_input.split())
````

As you can see the string `"no north"` is split up into **a list** containing the strings `"go"` and `"north"`.

You could use those parts to do some logic:

```python .noeval
user_input = "go north"
parts = user_input.split()
if parts[0] == "go":
    if parts[1] == "north"
        ... # Go north
    elif parts[1] == "south":
        ...
    ...
else:
    # quit the game
```

So in general, `.split()` simply splits up a string:

````
print("Hello, what's your name?".split())
print("This is also a sentence".split())
print("I think that by now you probably realize what split does".split())
````

By default `.split()` uses **any whitespace** as the separator (the boundary) to split.

But you can also specify that `.split()` should use a different separator:

````
print("Hello, what's your name?".split(",")) # notice it doesn't get rid of spaces automatically
print("a-b-c-d-e".split("-"))
print("The given separator is not in this string".split(".")) # a list with a single element
````

## `.join()`

`.join()` does exactly the opposite of `.split()`, it takes a separator and a list of strings, and glues those strings to each other, separating all the elements using the separator. Examples will show what it does better than the explanation:

````
print("-".join(["a", "b", "c", "d", "e"]))
print(" ".join(["This", "is", "also", "a", "sentence"]))
````

In the previous section we showed that you can't use `sum()` to concatenate a list of strings:

````
my_friends = ["Bob", "James", "Alice"]
print(sum(my_friends)) # error!
````

But instead, you can use `.join()` to do the same thing!

````
my_friends = ["Bob", "James", "Alice"]
print("".join(my_friends)) # What you'd expect `sum()` to do
````

But of course, if you want them presented in a clean way, you'd probably add a separator:

````
my_friends = ["Bob", "James", "Alice"]
print(", ".join(my_friends))
````
