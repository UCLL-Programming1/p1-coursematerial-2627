# Lists

## Basic todo app

One of the previous exercises was to write the following [todo-app](../2-control-flow/exercises.ipynb):

````
print("Welcome to your todo app!")
todo_focus = "nothing in particular"
choice = ""
while choice != "E":
    choice = input("Do you want to change (C) your todo-item, show (S) your current item or Exit (E) the application? > ")
    if choice == 'C':
        todo_focus = input("Type your todo: > ")
    else:
        if choice == 'S':
            print("Your most important todo is: " + todo_focus)
````

This program keeps track of the single most important todo that we have to be reminded of. When writing this program, you might have had the following thought: Wouldn't it be a lot more useful to keep track of multiple things, like in a todo-list?

We agree whole-heartedly. However to do that, we're going to have to learn about a new feature in python: `List`s. That's exactly what this lesson will be about. (If you're not convinced we need a new Python feature for this, we challenge you try and create a Todo app with multiple todo's without using a `list`)

## Lists in our daily lives

In life, we use lists all the time. We make a list of the groceries we still have to buy or things we still have to do. In school we make lists of students that are registered for a given course and you as a student probably have a list of all the courses you are taking this semester. In short, we use lists whenever we want to **keep track of more than one thing**, though we might not know how many in advance, or this number changes at different times.

Lists, no matter whether it's a todo-list, a grocery list, a list of registered students etc, typically have a couple of things in common:

1. We can look at what is currently on the list
1. We can add new elements (eg adding an extra todo) at the end or at any other position in our list.
1. We can remove elements from the list (eg striking through a todo once we've finished it) ...

## Creating a list

In python, we can create a list with square brackets:

````
my_first_list = ["the first item in my list", "the second one", "the third", "etc"]
my_first_list
````

The previous list is a list of `String`s, but lists can contain any value:

````
list_of_numbers = [1, 2, 3, 4, 5, 6]
list_of_numbers
````

It can contain a mix of different types, and it can even contain other lists:

````
another_list = [1, "Johnny", [1, 2, 3], "star"]
another_list
````

So a list in Python has the following structure:

```python .noeval
name_of_list = [ ... ]
```

where the list is created by using square brackets `[]`, and any values we want in the list are specified in between those brackets.

We could of course create an empty list by not specifying any values

````
empty_list = []
empty_list
````

## Adding items to a list

Of course though creating a list as fixed set of values might be useful in some cases, in most cases we want to add new items to it as we go along. We do this using the `append` **method** ("append" means **adding at the end**).

````
numbers_i_know = [1, 2, 3, 4]
print(numbers_i_know)
numbers_i_know.append(5)
numbers_i_know.append(6)
numbers_i_know.append(10)
print(numbers_i_know)
````

## Removing items from a list

likewise, we can also remove elements from a list. We do this using the `remove` **method**:

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
print(my_todos)
my_todos.remove("Do laundry")
my_todos.remove("Go grocery shopping")
print(my_todos)
````

So just like for strings, there specific built-in methods for lists, of which `append` and `remove` are the first two that we will see. We'll cover some other list methods soon.

## Multiple occurrences

Lists can contain the same value multiple times

````
my_friends = ["Alice", "Bob", "Alice", "James"]
my_friends
````

So what would happen if we would remove the value `"Alice"` from this list? (Guess before you run the code)

````
my_friends = ["Alice", "Bob", "Alice", "James"]
my_friends.remove("Alice")
my_friends
````

As you can see, when you remove a value that occurs multiple times in the list **`remove` only removes the first occurence of that value**.

## Indexing / slicing

Just like with strings, we can use the index to access an item in a list (and just like with strings, we start counting at `0`)

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
print(my_todos[0])
print(my_todos[1])
print(my_todos[2])
````

or, with negative indexes:

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
print(my_todos[-1])
print(my_todos[-2])
print(my_todos[-3])
````

And just like with strings, we can select many values by using **slicing**:

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
first_two_todos = my_todos[0:2]
print(first_two_todos)
````

### Changing elements at an index

Unlike strings, we can also actually **change a value at a specific index**

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
my_todos[0] = "Wake up"
print(my_todos)
````

> Note: you cannot do this with strings, because **strings are immutable**, they cannot be changed. This will be important later.

## Inserting an item at a given index

Instead of just adding an item at the end of a list using `append`, you can also insert an item at a specific index using the method `insert`. All the following items will be shifted one position to the right:

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
my_todos.insert(0, "Wake up")
print(my_todos)
````

## 'Popping' (removing) an item at a given index

Instead of removing an element **by its value** using `remove`, you can also remove an element by its position in the list (its index), using `pop`

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
my_todos.pop(0)
print(my_todos)
````

You can also use `pop` without specifying an index, in which case **it will remove the last element**.

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
my_todos.pop()
print(my_todos)
````

An interesting feature of `pop` is that it **also returns the value it has removed**:

````
my_todos = ["Do laundry", "Go grocery shopping", "Watch TV"]
removed = my_todos.pop()
print(my_todos)
print(f"todo removed from the list: {removed}")
````

(Of course `remove` does not return the value it removes, because you already specify it when calling the method)

## Operators

Some of the operators that we saw earlier are also defined for lists:

### Concatenation `+`

````
print(["a", "b", "c"] + ["e", "f"])
````

### Multiplication `*`

````
print(['a'] * 10)
````

### Membership (in)

Just like for strings, the `in` operator is also defined for lists, it checks whether a specific value is in the list:

````
groceries = ["milk", "apples", "noodles"]
print("bananas" in groceries)
````

However, be careful! `in` doesn't check if elements are in 'nested lists':

````
groceries = ["milk", "apples", ["noodles", "eggs"]]
print("milk" in groceries) # True
print("noodles" in groceries) # False
print(["noodles", "eggs"] in groceries) # True
````

## List functions

Finally there are some built-in functions that work with lists that are good to know about

### `len()`

As for strings, `len()` returns how many elements are in a list:

````
groceries = ["milk", "apples", "noodles"]
print(len(groceries))
````

But again, every element (even nested lists) only count for 1:

````
groceries = ["milk", "apples", ["noodles", "eggs"]]
print(len(groceries)) # 3
````

### `min`/`max`

`min` and `max` will return respectively the minimum or maximum value of a list:

````
some_numbers = [1, 6, 2, 7, 3, 100]
print(min(some_numbers)) # 1
print(max(some_numbers)) # 100
````

In order to do this it compares all of the elements to each other using `<`, so `min` and `max` also works for getting the string that is first or last in the alphabet respectively:

````
my_friends = ["Bob", "James", "Alice"]
print(min(my_friends))
print(max(my_friends))
````

which also means that if you try to get the minimum or maximum value of lists that have elements that are not comparable, Python will raise an error:

````
mixed_list = ["James", 1984, True, [1, 2, 3]]
print(min(mixed_list)) # error!
````

### `sorted`

The function `sorted` will (as its name implies) return a sorted version of the input list:

````
some_numbers = [1, 6, 2, 7, 3, 100]
print(sorted(some_numbers))
````

Note that the original list is unchanged, `sorted` returns a sorted **copy**:

````
some_numbers = [1, 6, 2, 7, 3, 100]
print(sorted(some_numbers)) # sorted
print(some_numbers) # original, unchanged
````

And just like `min` and `max`, `sorted` relies on `<` to compare the values in the list to know which ones should go first, so if items are not comparable, you'll get an error

````
mixed_list = ["James", 1984, True, [1, 2, 3]]
print(sorted(mixed_list)) # error!
````

### `sum`

And if you want to add a bunch of numbers together, you can use `sum`:

````
some_numbers = [1, 6, 2, 7, 3, 100]
print(sum(some_numbers))
````

and though you might expect this to also work on a list of strings, to concatenate them, as that is what `+` does for strings, that is not the case, summing strings will result in an error:

````
my_friends = ["Bob", "James", "Alice"]
print(sum(my_friends)) # error!
````

(later we'll see that we can use a different function, `join()` for this purpose)
