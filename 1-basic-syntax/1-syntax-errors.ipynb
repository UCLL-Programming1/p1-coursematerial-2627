# Syntax errors

When writing your first 'Hello World' program, you might have gotten it right on your first try, or you might have written a typo (eg: `prin("Hello World")`), forgotten about the parentheses `()` (eg: `print "Hello World"`) or maybe the quotes `""` (eg `print(Hello World)`).

If you did any of these things, you will have encountered your very first `Syntax error`. Congratulations again! Getting a Syntax error is one of the key milestones in your programming journey!

If you didn't do any of these things and therefore did not encounter a syntax error yet, we don't want you to feel left out! You can see what we are talking about by simply running the following cell:

````
print "Hello World"
````

Running this cell should result in output that is similar to the following: (it can be a little different)

```text
  Cell In[3],   line 1
    print "Hello World"
          ^
SyntaxError: Missing parentheses in call to 'print'. Did you mean print("Hello World")?
```

## What is a syntax error?

Writing a syntax error in code is like saying a sentence such as "I go store, buy apple." This sentence is not grammatically correct. However in language, because people are pretty smart, they can often still understand sentences that are maybe not entirely correct, as long as they still convey enough meaning. For example, you might have deduced from the above sentence that I am going to the store to buy an apple. Or maybe you didn't, and then I'd have to repeat myself in a way that's more clear.

However, we pointed out in the introduction (`0-intro/0-programming-intro`) that filling in these gaps in instructions comes with a downside: If we allow to computer to fill in gaps in our instructions, they might fill them in wrong! We concluded that we don't want this to happen.

In order to avoid this, we said that the programming language will be defined such that instructions in it only ever have 1 possible interpretation. To make this possible, Python only allows for a limited (but very high) number of instructions.

Python has a `print` command (later we'll call this a function) that prints out any text you pass to it, but it doesn't have a `prin` (without the 't') command. If it were to try to guess what we wrote, it could guess wrong. We don't want Python to do this. Instead, it tells us: "The instructions (code) stopped making sense here, please take a look" and we — as the programmer — get the chance to correct the mistake ourselves, making sure it is done correctly.

Similarly, executing a command such as `print` is always done by letting it end in parentheses `()`. So if you don't do that, instead of guessing that you probably wanted to execute that command, Python will again simply tell you "The program stopped making sense here" and it will stop executing, so that you get the opportunity to fix your own mistake correctly.

At this point, you might not be 100% sold on this idea that Python working this way is actually an advantage. But it is, which is demonstrated by the fact that all programming languages were designed and still used this way. As you gain more experience in programming, this will become more clear to you. For now, if you're not convinced about it, just know that this is how Python and other programming languages work, and worry about the why later.

> A syntax error means that you wrote code that is 'syntactically` not correct, code that does not make up a valid Python program.

## Fixing syntax errors?

As we saw in the above example, when the Python interpreter encounters code that is not syntactically correct, it stops executing.

Fortunately Python tells us where it stopped. It does this by printing out an error message. The message in the above example is:

```text
  Cell In[3],   line 1
    print "Hello World"
          ^
SyntaxError: Missing parentheses in call to 'print'. Did you mean print("Hello World")?
```

The caret `^` points to the spot where the code is that made the program crash. In this case the exact spot is the first double quote `"`. The error message even lays out exactly what's wrong. It tells us "Missing parentheses". So putting all this information together, this error message indicates that on the spot of our first double quote, Python was expecting parentheses, which is exactly what's wrong with our code.

### Not always straightforward

However, it is not always that straightforward to find our mistakes.

Let's take a look at a different mistake you could have made when writing the 'hello world' program. Execute the following (erroneous) cell.

````
print(Hello World)
````

Running this code also results in a Syntax error. The exact wording and suggested solution may vary depending on your Python version. One result you might have gotten is (it depends on your Python version):

```text
  Cell In[5],   line 1
    print(Hello World)
                ^
SyntaxError: invalid syntax
```

As you probably notice, this error message is not so clear. The caret `^` points to the start of `World` while the mistake that we actually made is that we didn't enclose `Hello World` in double quotes (ie "Hello World"). The error message itself does also not clarify anything, it only says "SyntaxError: invalid syntax" (the exact error message can be different on your computer).

Like we said before, Python might try to make a suggestion as to what is wrong, but it doesn't know what you **intend** to make it do, so this suggestion might be way off. What it can reliably do is tell you where the program stopped being a valid program, and it will do that every time it happens.

> An error message doesn't always tell you exactly where the mistake in the code is, it just tells you where the code stopped being a valid Python program. This points you to the area in your code where you have to start looking, but often you have to investigate from there what the mistake in the code actually is.

## Errors are part of programming

When starting out programming for the first time, encountering errors might feel very frustrating. To many it feels like failure, that they should just be able to write code that doesn't have any errors in it!

And while it's true that as you gain experience you will probably make these kinds of mistakes less often, you will almost never write code that is completely error-free on the first attempt. Finding and fixing errors in the code is an inherent part of programming. Even the most senior programmers, with years or even decades of experience, still write code with errors all the time - though the types of mistakes they make might differ.

This is all to say: try not to panic when encountering an error in your code. Simply read the error message and try to figure out where the mistake is. Fix your code and repeat. This is part of the process.
