# Booleans

## Boolean expressions

In the real world, we often say things as:

- Can you go to the store to get the things we need for pancakes? __If the store doesn't have regular milk__, just bring soy milk instead
- Let's try to go to the museum. __If the museum is closed__, we'll just walk around town.
- It's pretty chilly outside. __If it's raining__, we shouldn't forget to take an umbrella.
- In this game, __if you roll a six__, you get another turn.
- Can you open the door? __If it's a delivery__, just sign in my place.
- Everyone can vote, __as long as they are 18 or older__,

All of the emphasized parts:

- (if) the store doesn't have regular milk
- (if) the museum is closed
- (if) it's raining
- (if) you roll a six
- (if) it's a delivery
- (as long as) they are 18 or older

have something in common: each one describes an **uncertain** situation that is either **the case** or is **not the case**.

- either the store has regular milk or it doesn't have regular milk
- either the museum is closed or it's not closed
- either it's raining or it's not raining
- either you roll a six or you don't roll a six
- either it's a delivery or it's not
- a person is either 18+, or they're not In each of these situations there are only be **two possible outcomes:** the expression is **true** (it is the case) **or** it is **false** (it is not the case).

In programming, we have a particular name for these kinds of statement that are either true or false: **boolean expressions**. (If you are wondering how they got their name, they were named after [George Boole](https://en.wikipedia.org/wiki/George_Boole))

## If-statements

A lot of the times we use boolean expressions in this kind of 'in this situation, do this' statements. For example, if we revisit the examples above:

- __If__ (the store doesn't have regular milk), __just bring soy milk instead__
- __If__ (the museum is closed), __we'll just walk around town__
- __If__ (it's a delivery), __just sign in my place__

So **if the boolean expression** (in parentheses) **is true**, we **do a certain action**. For example if it's true that that the store doesn't have regular milk, we bring soy milk.

And, while we often don't state these things explicitely, there is usually (but not always) also an action associated if the boolean expression is not true, but it is often implied:

- If the store doesn't have regular milk, just bring soy milk instead (__BUT__ if the store **does** have regular milk, **bring the regular milk**)
- If the museum is closed, we'll just walk around town (__BUT__ if the museum **is not** closed, **we'll visit the museum**)

However this is not always the case. For instance in the example

- __If it's a delivery, just sign in my place__

it's not clear what we have to do **if it's not** a delivery. (Obviously we won't have to sign, but what do we need to do instead?)

# In Python

Just like we give these 'if-instructions' to people, we can also give these instructions to Python. In Python we do this with the `if-else` statement. Take a look at the following example, run the cell (and make a guess what animal we're thinking of :))

````
guess = input("Guess what animal I'm thinking of: > ")
if guess == "dolphin":
    print("You guessed correctly!")
else:
    print("Nope, try again!")
````

As you might have figured out, the above code makes you guess an animal. The correct answer is **dolphin** and the code will print out whether you guessed correctly or not.

Of course this 'game' is pretty easy when you can just see the code (and therefore the right answer). If you want to make it more challenging, let a friend change the answer, run the cell and scroll this document all the way down (so you can no longer see the code) before letting you guess.

Now, let's dive a little deeper into this example.

## Boolean expressions

Let's start with the easiest part first: What is `guess == "dolphin"` and what does it do?

Well, this is the first example of a **boolean expression** in Python, and it is the result of comparing two values with the **equals-operator** `==` ( which consists of **two** `=` symbols - not to be confused with the assignment-operator, which is only **a single** `=` symbol).

Very basically, `==` checks whether two values are equal to each other. Let's see some examples

````
'bear' == 'dolphin'
````

````
'dolphin' == 'dolphin'
````

As you can see, if two values are equal, Python returns the value `True` (with a capital T), and when they are not equal, it returns `False` (again with capital F). `True` and `False` are the two **boolean values** and together form the **new data type** `boolean` in Python. Note that these are really a new kind of value, not just another string. You can see this clearly when trying to compare them with a similar string:

````
False == 'False'
````

````
True == 'True'
````

Looping back to our introductory example, if we execute the following cell:

````
input() == 'dolphin'
````

The cell will display `True` if the input we provided is `'dolphin'` and `False` otherwise.

> Note: comparisons using `==` are case-sensitive! (capital letters are not equal to their lower case counterparts) Run the following cells to verify this:

````
'dolphin' == 'Dolphin'
````

````
'Dolphin' == 'Dolphin'
````

````
'dolphin' == 'dOlphIn'
````

````
'a' == 'A'
````

````
'a' == 'a'
````

## If-else statements

Now that we understand how the comparison `guess == 'dolphin'` results in the value `True` or `False`, let's take a look at the other part of the example, the `if-else` structure.

The basic syntax of an `if-else` statement in Python is the following

```python .noeval
if boolean expression:
    # code to execute when expression is true
else:
    # code to execute when expression is false
```

Pay attention to the following parts:

- The first line
  - starts with `if`
  - followed by a **boolean expression** (in this case the comparison `guess == 'dolphin'`)
  - and finally ends with a **semicolon** `:`
- Then we can write whatever code you want to execute **when the boolean expression is true**
  - it follows on the next line
  - **is indented** (this means we let the line start with a couple of spaces, typically 4)
- On the line after the code that is executed when the boolean expression is true
  - we write `else`
  - again followed by **a semicolon** `:`
- So that we can then write code that is executed **when the boolean expression is false**
  - on the line after `else:`
  - should again be **indented**

### Executing multiple lines of code

If you want to execute more than one line of code inside the `if-else` structure, you can do this **as long as you keep all of them indented**

````
guess = input("What animal am I thinking off? > ")
if guess == 'dolphin':
    print("That's right!")
    print("Do you want to play again?")
else:
    print("Unfortunately not")
    print("Give it another try")
````

On the other hand, the following **will not** work, because not all the lines are correctly indented:

````
guess = input("What animal am I thinking off? > ")
if guess == 'dolphin':
    print("That's right!")
print("Do you want to play again?")
else:
    print("Unfortunately not")
    print("Give it another try")
````

### If without else?

Sometimes we want to **do something only when the boolean expression is true**, but **not really when it is false**. for example, if we only want to print something when the guess is correct, but we don't want to print anything when the guess is wrong we can write the following:

````
guess = input("What animal am I thinking off? > ")
if guess == 'dolphin':
    print("That's right!")
````

As you can see by submitting a wrong guess, our program will simply skip over the code in the `if`-block and do nothing.

### Else without if?

While it's possible to leave out the `else`-block. It's definitely **not possible to leave out the `if`-block**. This makes sense, because we specify the boolean expression right after `if`, and `else` means "execute the following code when the boolean expression is false". If there is no boolean expression, `else` has no meaning. Trying to do this will result in a syntax error.

````
guess = input("Guess what animal I'm thinking of? > ")
else:
    print("We'll never get to this point because else without if will result in a syntax error")
````

## More boolean operators

So far the only boolean expressions we've seen are comparisons of the form `value_1 == value_2` (for example `guess == 'dolphin'`).

These expressions used the equals-operator `==`. However, there are a lot more operators than only `==`. We will take a look at the most common ones now:

### Not equals `!=`

First of all there is the `!=`-operator (not equals), which does exactly the opposite of the equals-operator: it checks whether two elements **are not equal** to each other. Wherever `==` results in `True`, `!=` will result in `False` and vice versa. For example:

````
'dolphin' != 'shark'
````

(it is true that `'dolphin'` is not equal to `'shark'`)

````
'dolphin' != 'dolphin'
````

(it is false that `'dolphin'` is not equal to `'dolphin'`)

### `>=`, `>`, `<` and `<=`

We can also compare numbers to each other. This is useful in a lot of cases. A simple example would be: (note that we use `int()` to convert the string into an integer)

````
age = int(input("Are you old enough to have a beer (in Belgium): > "))
if age >= 16:
    print("You legally can, but keep in mind that any consumption of alcohol harms your health!")
else:
    print("Not yet...")
````

In this example, we used the greater-or-equal-than-operator `>=` that compares two numbers and returns `True` if the first number is greater than, or equal to the second number. If that's not the case (the first number is smaller than the second number), it will return `False`:

````
20 >= 5
````

````
20 >= 20
````

````
5 >= 20
````

In addition to the `>=`, the following operators exist:

| Python Syntax | Mathematical Notation | Description |
| :-----------: | :-------------------: | :---------- |
| `==` | $=$ | equal to |
| `!=` | $\neq$ | not equal to |
| `<`  | $<$ | less than |
| `>`  | $>$ | greater than |
| `<=` | $\leq$ | less than or equal to |
| `>=` | $\geq$ | greater than or equal to |

For example, the less-than-operator `<` returns `True` when the first number is **smaller** than the second number. Its result is `False` in all other cases. In fact its result is always the exact opposite from that of the `>=`-operator.

````
5 < 20
````

````
20 < 5
````

````
20 < 20
````

The other comparison-operators do similar things. Experiment with these operators until you feel comfortable with what they do. You will use them a lot.

### Ordering strings?

Can we use the above operators on strings as well? Actually we can. When doing so Python will **compare these string values alphabetically** (where a 'smaller' value appears in the alphabet before a 'bigger' value)

````
'Adam' < 'Eve'
````
