## The `input` function

We mentioned it in the last section: Just as there is a way to print output, there is also a way for your program to ask for input. For this, there is the `input` function, and it can be simply used as follows:

```
input()
```

When you run the above cell, you might notice that if you do nothing, the program doesn't finish quickly like all the pieces of code we've written thus far. Instead it just keeps running while seemingly doing nothing. The reason for this is that this function is waiting for your input. Python does this in different ways depending on where you run your Python code. In a Jupyter notebook like this one, it's asking you with a text box that appears on the top of your screen, which should look something like this:

![Input box](../.assets/input_box.png)

In this box you can write anything you want and press 'Enter'.

As you can see, the code example above shows you the text value you wrote in the input box. This value is shown because - as we saw in the last section - Jupyter notebooks always show you the last value in a cell. However if you want to consistently show these values, you'll have to use `print`:

```
input()
input()
input()
```

```
print(input())
print(input())
print(input())
```

As we saw in the section on `print`, the first example will only show you the last value you inputted, while the second one will print all the values you input.

Note that the value that is shown **is a string** (surrounded by single quotes `''`).

### No function argument?

One thing you might have noticed is `input` **does not take an argument**. Some functions don't take a function argument because they don't need one. The value that matters for `input` is the value that the user provides. Later we'll also see that some function take more than one argument.

> Note that even though we don't need to pass an argument, we **still need to write parentheses `()`**, otherwise the Python interpreter doesn't recognize that we actually want to call (execute) the function. If you run the following code block (which has `input` without parentheses), you'll see it does nothing, except for showing some weird output.

```
input
```

### Printing the input?

At this moment, it may be a little weird that the only thing we're doing with out input is printing it back out again straight away using `print()`. What is the point of that? The reason we're only doing that for now is that printing is the only thing we know how to do. In a short while, we'll learn to do other stuff with our input as well.

### A function that returns a value

Let's look at another example using `input`:

```
print("Hello " + input())
```

Let's dissect what this code does:

1. It prints whatever is between the parentheses following `print`, which is `"Hello " + input()`
1. This expression between the parentheses is a **string concatenation** (addition of two strings) of `"Hello "` and `input()`

So how is possible to concatenate `"Hello "` and `input()`, with `"Hello "` being a string and `input()` being a function?

Basically the `input`-function call **turns into a string** when it is called. It turns into the string containing the text value the user writes in the input box.

So if the value you wrote in the input-box is `Johnny`, the above code is the same as:

```
print("Hello " + "Johnny")
```

When a function "turns into" a new value after executing , we say that that function **returns that value**. So in the above example (again assuming we wrote `Johnny` as input) the `input`-function returns the string `"Johnny"`

This returned value can then be used just as if we wrote that value directly in our program.

### Passing an argument to `input`

Before we said that `input` does not take an argument. This is actually not entirely true, `input` **can** take an argument, but **it is optional**, so you can choose whether or not you want to do that. Then what is this optional argument for? Well simply take a look at the following example:

```
print("Hello " + input("What is your name?"))
```

As you can see the argument is simply used in the input box to show the user what they should be giving as input, which makes it more clear what the purpose of the input box is.

![Input box with prompt](../.assets/input_name.png)

Of course we could write anything. Whatever you write does not impact what is eventually printed, it is just for clarification.

```
print("Hello " + input("Your name please?"))
```

```
print("Hello " + input("Name"))
```
