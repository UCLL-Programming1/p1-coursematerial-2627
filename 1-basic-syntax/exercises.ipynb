In this exercise we will expand on the number guessing game you implemented in the previous exercise. The game will work exactly the same with one key difference: **The player gets two attempts** to guess the number.

So the program will run as follows:
```none

Player, guess the number: > 10
Your guess is too low!
Try again! (You have 1 guess left)
Player, guess the number: > 34
You guessed correctly!
```
or
```none
Game master, enter a number: > 34
Player, guess the number: > 60
Your guess is too high!
Try again! (You have 1 guess left)
Player, guess the number: > 10
Your guess is too low!
You lose!
```
of course, a player could also still guess it right on the first attempt, winning immediately:
```none
Game master, enter a number: > 34

You guessed correctly!
```
