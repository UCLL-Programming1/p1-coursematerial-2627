# Nested if-statements and elif

## Multiple conditions in real life

So far we've only looked at situations with two possible outcomes: either a boolean expression is true, or it isn't. But in real life, we often deal with situations that have **more than two possible outcomes**. Consider the following examples:

- At a theme park: **if** you are younger than 12, you pay the child price. **If** you are between 12 and 64, you pay the adult price. **If** you are 65 or older, you pay the senior price.
- For a school grade: **if** your score is 90 or above, you get an A. **If** it's between 80 and 89, you get a B. **If** it's between 70 and 79, you get a C. And so on.
- When ordering a coffee: **if** you want a small, it costs €2. **If** you want a medium, it costs €3. **If** you want a large, it costs €4.

Each of these situations has **three or more branches** — three or more different things that can happen. A simple `if-else` only gives us two branches. In this notebook we will look at two ways to handle multiple branches: **nested if-statements** and **`elif`**.

## Example: ticket prices

Suppose we are writing a program for a cinema ticket booth. The ticket price depends on the age of the visitor:

- **Younger than 12** → child ticket: €5
- **12 up to (but not including) 65** → adult ticket: €15
- **65 or older** → senior ticket: €8

If you want, you can take a minute here to think about how you would write this program.

Done? Ok, then let's take a look at how we can do this.

We know how to handle two outcomes with `if-else`. Can we somehow combine multiple `if-else` blocks to cover three outcomes?

## A solution that doesn't work

A first solution that you may have come up with, is to simply write a bunch of `if`-statements, like so:

````
age = int(input("Enter your age: > "))

if age < 12:
    print("Child ticket: €5")
if age < 65:
    print("Adult ticket: €15")
else:
    print("Senior ticket: €8")
````

But in this case, what will this code do if you enter `9`, for example? Think about it before running the code.

As you can see, the code does not behave as intended. Because `9` is smaller than both `12` and `65`, it will execute both `if`-statements. So what we want is to only check the second condition `age < 65` in the case the first condition is not `True`. Let's see how we can do this.

## Nested if-statements

One approach is to put our second `if` statement **inside** the `else`-block of our first `if`-statement. This ensures that the second `if`-statement will only be executed if the first `if`-condition is `False`. We can do this because, inside of `if-else` block, we can put any code we want, even other `if-else` statements. We call this **nesting**. Take a look what this would look like:

````
age = int(input("Enter your age: > "))

if age < 12:
    print("Child ticket: €5")
else:
    if age < 65:
        print("Adult ticket: €15")
    else:
        print("Senior ticket: €8")
````

Let's trace through what happens for a few different ages to make sure we understand this:

- If `age` is **8**: the first condition `age < 12` is `True`, so we print `"Child ticket: €5"` and skip the `else` block entirely.
- If `age` is **30**: the first condition `age < 12` is `False`, so we go into the `else` block. Inside that block, we check `age < 65`, which is `True`, so we print `"Adult ticket: €15"`.
- If `age` is **70**: the first condition `age < 12` is `False`, so we go into the `else` block. Inside that block, `age < 65` is also `False`, so we print `"Senior ticket: €8"`.

Notice how the inner `if-else` is **indented** inside the outer `else` block — this is what makes it a **nested** statement. Python knows these two lines belong together because they share the same indentation level.

### Nesting gets messy quickly

Take a look at the following code. It converts grades out of 100 to their letter grades (A, B, C, D, and F). Using nested `if-else` blocks, it would look like this:

````
score = int(input("Enter your score (0-100): > "))

if score >= 90:
    print("Grade: A")
else:
    if score >= 80:
        print("Grade: B")
    else:
        if score >= 70:
            print("Grade: C")
        else:
            if score >= 60:
                print("Grade: D")
            else:
                print("Grade: F")
````

This code works correctly, but look at how far the inner blocks are pushed to the right! With each extra branch, the code gets one more level of indentation. This quickly becomes hard to read and easy to make mistakes in. Luckily, Python gives us a much cleaner way to write this.

## `elif`: a shorthand for 'else if'

Python provides a special keyword `elif` which is short for **'else if'**. It allows us to check an additional condition in the case that the previous condition was `False`, without the need for nesting.

The general syntax is:

```python .noeval
if first condition:
    # code to execute when the first condition is true
elif second condition:
    # code to execute when the first condition is false AND the second condition is true
else:
    # code to execute when all conditions above are false
```

Let's rewrite the cinema ticket example using `elif`:

````
age = int(input("Enter your age: > "))

if age < 12:
    print("Child ticket: €5")
elif age < 65:
    print("Adult ticket: €15")
else:
    print("Senior ticket: €8")
````

This produces exactly the same result as the nested version, but it is much easier to read. All three branches sit at the **same indentation level**, so the structure is immediately clear at a glance.

Here is how Python reads an `if-elif-else` block:

1. Check the `if` condition. If it's `True`, execute that block and **skip everything else**.
1. Otherwise (so only if the first condition was `false`), check the `elif` condition. If it's `True`, execute that block and **skip everything else**.
1. If none of the conditions above were `True`, execute the `else` block.

> Note: just like `if` without `else`, you can also use `elif` without a final `else`. In that case, if none of the conditions match, nothing happens.

### Multiple `elif` branches

You can use **as many `elif` branches as you need**. Let's rewrite the messy nested grade example from before using `elif`:

````
score = int(input("Enter your score (0-100): > "))

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
````

Much better! Even with five branches, the code stays flat and easy to follow. Each condition is checked in order from top to bottom, and as soon as one matches, the rest are skipped.

### What about the order?

Take a look at the following code that looks very similar to the code in the previous section. Will it behave the same. Think about it before running the code. Run the code when you think you know what's going to happen to see if you're right.

````
score = int(input("Enter your score (0-100): > "))

if score >= 60:
    print("Grade: D")
elif score >= 70:
    print("Grade: C")
elif score >= 80:
    print("Grade: B")
elif score >= 90:
    print("Grade: A")
else:
    print("Grade: F")
````

This code doesn't seem to work! Why not? Because Python checks conditions **from top to bottom and stops at the first match**, therefore the order of your `elif` branches matters, since the first conditition that is true will determine which branch will executed. Let's look at an example:

Try entering a score of `95`. You might expect to see `Grade: A`, but instead you get `Grade: D`. Why?

Because Python checks the conditions in order. `95 >= 60` is `True`, so it immediately prints `"Grade: D"` and skips all the remaining `elif` branches — even though `95 >= 90` is also `True`.

The fix is to start with the **most restrictive condition first** (the highest threshold), exactly as we did in the correct version above.

> **Rule of thumb:** when your conditions overlap, always put the most specific (or most restrictive) condition first.

## Summary

In this section we saw 2 things:

1. We can nest new if-else blocks inside of other if-else blocks
1. `elif` is shorthand for `else if`

Here is the full syntax of an `if-elif-else` block:

```python .noeval
if condition_1:
    # executed when condition_1 is True
elif condition_2:
    # executed when condition_1 is False AND condition_2 is True
elif condition_3:
    # executed when condition_1 and condition_2 are False AND condition_3 is True
else:
    # executed when ALL conditions above are False
```

Key things to remember:

- You can have **as many `elif` branches as you need** (including zero)
- The `else` at the end is **optional**
- Python checks conditions **top to bottom** and **stops at the first match**
- `elif` keeps your code **flat and readable**, compared to deeply nested `if-else` blocks
