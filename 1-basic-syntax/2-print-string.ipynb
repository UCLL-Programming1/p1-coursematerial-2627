# Interacting through text

In the introduction section we saw examples of **text-based programs**, programs that **instead of** working with a **graphical user-interface (GUI)**, they work by **interacting with the user through text**.

There are two elements to having our program interact with the user through text:

1. Asking the user for text input
1. Responding to the user with text output

In this section we'll first talk about text output. (Spoiler alert: we already did this). Then in the next section, we'll cover asking the user for input.

## The `print` function and `String`s

Consider again the first piece of code we wrote:

````
print("Hello World")
````

As is probably clear by now, `print` in python prints out whatever text you specify between the parentheses `()`. This is how you let your program give output to the user.

## Functions

In Python, we call `print` a `function`. **Functions are commands** that perform a very specific functionality that you can use in your programs. There are a lot of functions in Python and you'll learn many of them as we progress through this course. At some point you'll even learn how to make new functions yourself.

At this point you might wonder **how these functions work**, how for example the `print`-function achieves to move the pixels on your screen to form these letters. However, the beauty of programming is that **you don't need to know**. Just like you probably don't really know what happens when you press the on-button on your smartphone, or when you press one of its icons on its screen, you don't need to know how these functions work, **you simply need to know how to use them**.

### Calling/executing functions

When we let our program execute a function (as we did above with `print("Hello World")`), we say that we **call the function**. So **calling** is just **a programming term** that means **executing**.

In order to call (execute) this function, we need to write **its name** (in this case `print`), followed by **parentheses** `()`, in between which we specify the **function argument** (the value the function needs to do its thing), in this case `"Hello World"`.

## Strings

In our above example we passed the text `"Hello World"` to `print` in order to output it. A text value like `"Hello World"` is called a `String` (this name comes from 'a string of characters'). `String`s are **pieces of text** our programs work with and are always **enclosed in quotes `""`** so the Python interpreter knows it shouldn't interpret this text as code, but as regular text.

`String` is one of Python's **datatypes**, (we'll see other data types later) it represents textual data.

As you can see, VS Code highlights Strings differently than it does other code. This is called **syntax highlighting** and it is done to make a visual distinction between the different types of elements in our code, to make it easier for us to see what our code does at a glance.

> Instead of double quotes `""`, you could also use single quotes `''` (eg `'Hello World'`). The important thing is that you stay consistent: A String that starts with a double quote needs to end in a double quote and a String that starts with a single quote has to end in a single quote. An expression such as `"Hello World'` (starting with a double quote and ending in a single quote) is not allowed.

### 'Concatenating' strings

Just like numbers, Strings can be 'added' together. We call this **concatenating** strings. We know, way to make a simple operation sound difficult. We swear we didn't choose these names. Concatenating simply means adding one string to the back of the other string:

````
print("Hello " + "World")
````

This is the same as `print("Hello World")`. Note that this operation doesn't magically add any spaces, so adding two words like this:

````
print("Hello" + "World")
````

results in just a single 'word' (`HelloWorld`), with no spaces in between.

Why you would want to do this instead of just writing a single string will become clear later.

## Multiple statements and line by line execution

Of course you can print many different things right behind one another:

````
print("10...")
print("9...")
print("8...")
print("7...")
print("6...")
print("5...")
print("4...")
print("3...")
print("2...")
print("1...")
print("Lift off!")
````

As you can see when executing this code (and as you might expect), **Python simply executes every statement in order, line by line**. The fact that Python executes code line by line is probably pretty intuitive in this example, but keep this in mind for the future!

## Printing the last expression?

If you execute the following two cells, you can see that a result is shown in both cases:

````
print("Hello World")
````

````
"Hello World"
````

So does this mean we don't really need `print` at all? It seems like we can just write whatever we want, and it's just printed out, without having to use the `print` function?

Not quite. In Jupyter notebooks (this kind of document), if the last line of cell you run contains an expression (a string, a number or some other kind of value - such as `"Hello World"` in the above example), it does show you that expression. However, this is only true for the **last** line. To see this, run the following two cells:

````
print("1")
print("2")
print("3")
````

````
"1"
"2"
"3"
````

As you can see, even though we wrote down three strings in the second cell, it's only the string on the last line that is shown to us, while if we use `print` (such as in the first cell), every string is printed.

Also note that what's being shown is slightly different between the two. For the second cell, it shows you the actual `String` (with its enclosing parentheses), while `print` just prints out **the content** of the string (without the parentheses).

This is a peculiarity of Jupyter notebooks. When writing Python programs in `.py` files, you'll always need `print` to show output to the user. To see this execute both [hello_world.py](supporting-material/hello_world.py) and [hello_world_no_print.py](supporting-material/hello_world_no_print.py) and observe the difference (`hello_world_no_print.py` shouldn't have any output)

In this section we saw the `print` function and how we can use this function to output **text** which is represented by the datatype `String`. This is the basic building block we'll use to write text-based programs that interact with users through text.
