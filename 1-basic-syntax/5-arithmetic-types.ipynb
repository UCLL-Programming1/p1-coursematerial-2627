# Basic arithmetic

The most basic thing that a programming language such as Python can do besides taking text as input and outputting some responses is basic arithmetic:

## Operators

### Addition

You can do addition with the `+` operator

```
1 + 1
```

```
254 + 345
```

### Subtraction

subtraction with `-` operator

```
123452 - 2342
```

### Multiplication

multiplication with the `*` operator

```
3 * 4
```

### Division

And division with the `/` operator

```
22 / 2
```

## Order of operations and parentheses `()`

What is the result of the following calculation?

```
3 + 3 * 2
```

The answer is 9, as the first operation to be executed is the multiplication. This is because **multiplication has precedence over addition** (multiplication is done first, before addition).

Python does this just like in mathematics. For this course, we expect you to be familiar with this. If this is not the case, you should brush up on [the order of operations](https://en.wikipedia.org/wiki/Order_of_operations)

Also, just like in mathematics, you can use parentheses `()` to override this order. For example if, in the above example, you want to execute 3 + 3 first and then multiplying that result (6) by 2, you can simply write:

```
(3 + 3) * 2
```

# Data types

In math we often make a distiction between whole numbers (-2, -1, 0, 1, 2, 3, 4, ...) and decimal (or rational) numbers - numbers with a decimal point in them - (0.5, 3.14, 1.618 etc).

Python makes a distinction between these two types of numbers as well. As you can see, addition (+), subtraction (-) and multiplication (\*) result in whole numbers when used on whole numbers, while division results in a decimal number. This makes sense, because division - even between two whole numbers - will sometimes need a decimal number to represent the result. For example:

```
10/4
```

equals 2.5. So in order to be consistent, Python will turn every result of a division into its decimal representation (eg 10.0 in the case of 10).

Of course if you add, subtract or multiply two decimal numbers, you'll also end up with a decimal number:

```
1.3 + 2.4
```

In Python we call whole numbers **integers** and decimal numbers **floats** (from floating point number). For now we don't have to worry too much about this or why it's important, but it's good to be aware that these are two different kinds of numbers.

Integers and floats are what we call **data types**, they represent a specific kind of data (in this case whole and decimal numbers). We've actually already encountered one other data type before in this class - `String` - which represents **textual data**. So a short summary of the data types so far would be:

data type | represents
--- | ---
String | textual data
int | whole numbers
float | decimal numbers

# Type conversions

Run the following cell and input a number (eg 23)

```
input("Input a number please")
```

and observe the difference with simply specifying the value 23 in a cell:

```
23
```

if you pay close attention, you'll see that the values displayed in each case are slightly different. In the first case the values displayed is `'23'` (with surrounding quotes`''` indicating this value is a string), while in the second case the value is `23` with no such quotes indicating this value is an integer.

## `input()` always returns a string

So why is this the case? Well quite simply, the function `input()` will always returns a string. It doesn't interpret the value you put in there. As far as it's concerned, anything typed in this input-box are just a bunch of characters (a string), so that's what it returns, even if some of those strings of characters only contain numbers.

### Why does it matter?

At first glance this doesn't matter, but actually it does. Why?

Well for one, we can only add numbers to numbers and strings to strings:

```
42 + '23'
```

This error is saying that the `+`-operator doesn't support addition between the types `int` and `str` (string). So **a string containing a number** (such as '23') **cannot be added to another number**.

So if we'd ask the user for an input number to show them what the result is of adding 5 to that number (We know this is a pretty simple example), we get the same error:

```
print(input("Enter a number to add 5 to") + 5)
```

The reason Python doesn't let you do this, is because most of the time when you try to add a number to a string, you're probably doing this by accident. Because of this, Python doesn't let you do this and instead wants you to make it explicit that you expect a certain string to actually contain a numerical value.

We can do this by **using functions that convert a value of one datatype into another datatype**:

## The functions `int()`, `float()` and `str()`

For example, if we know a specific string can be interpreted as an integer (a whole number), we can use the method `int()` to convert it to an integer:

```
int('23')
```

We see this function takes the string `'23'` as an argument, and returns the integer value `23`.

This way we can fix our little program that adds 5 to some input, we simply pass the result of the `input` function to the `int()` function (and add 5 to it)

```
print(int(input("Enter a number to add 5 to")) + 5)
```

`int(input())` will return the integer version of the number the user inputs, which allows it to be added to 5.

We can do similar things if we want to make other conversions:

If we expect an string to actually be a float (a decimal number), we can use `float()`:

```
float("34.3")
```

If we expect a floating point number to be an integer, we can make that conversion as well:

```
int(3.0)
```

Or we can convert an integer to a float:

```
float(3)
```

And finally we can also convert numbers back to strings, using the method `str()`:

```
str(3)
```

```
str(23.4)
```

### Converting incorrect values

So what would happen if we try to convert a string to a number, but the string doesn't really represent a number? Guess what will happen before you run the next cell:

```
int("tree")
```

As you may have guessed, if Python doesn't know how to convert the argument you give to it, it will simply raise an error, saying that it can't convert whatever you passed to it. It says this in the slightly complex way "invalid literal for int() with base 10: 'tree'", which means that the string `'tree'` is not a valid number literal in our 10-digit number system.
