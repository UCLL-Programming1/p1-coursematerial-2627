# Variables

A section about moving boxes, using sticky notes as labels and variables in programming.

Imagine you're about to move. You put all of your stuff in boxes and move them (hopefully with some help) into the living room of your new place. Tired of all the moving, you want to make some coffee for yourself and your helpers. You want to unpack your coffee machine, but which box is that in again? If you were a little forward-thinking (or if this is not your first time moving), you might have had the idea to label your boxes with one of the labels being "coffee machine", to be able to easily find your things after the move.

In programming, we also sometimes want to 'label' the values we end up with. In one of the previous sections we saw that the piece of code

````
print("Hello " + input("What's your name? "))
````

would ask the user for their name, which would then be used to print a greeting. For example if your name is `"James"`, it would print:

```none
Hello James
```

However, sometimes we might not want to use the input that the user gives us immediately, instead we might want to save it to use it later. Imagine if instead of immediately printing out the name of user, we first also want to ask them their last name, so that we can print a more detailed greeting. In that case we need to put a 'label' on the values that the user provides, so that we can use those values later on. In Python we can do this as follows:

````
first_name = input("What's your first name? ")
last_name = input("What's your last name? ")
print("Hello, your name is " + last_name + ", " + first_name + " " + last_name + ".")
````

Here, the first value that the user inputs gets the label `first_name`, and the second value they input gets the label `last_name`. We can then use those labels to access the values these labels were applied to.

For example if the user inputs `"James"` as their first name and `"Bond"` as their last name, this will print:

```None
Hello, your name is Bond, James Bond.
```

Likewise, we could rewrite the initial example as:

````
user_name = input("What's your name? ")
print("Hello " + user_name)
````

You can verify that this code works exactly the same as the initial example.

What happens here is that instead of doing something with the return value of `input("What's your name? ")` (being whatever the users writes as input) directly, we first give it the label (`user_name`) and then in the second line we use that label to find the value again and print it out.

If we go back to the analogy of moving boxes, the initial code is like your friend giving you a box and telling you to move it upstairs, while the second case is like your friend putting the coffeemachine in a box in the living room, putting the label "coffee machine" on it and then telling you to take the coffee machine upstairs. In both cases, you did the same thing - moving the box with the coffeemachine upstairs - but in the second case this box simply had a label.

In Python, we call these 'labels' **variables**. So we say that `first_name`, `last_name` and `user_name` are variables.

We can also use variables to put labels on Python values directly:

````
birth_year = 1994
current_year = 2026
print(current_year - birth_year)
````

This may seem unnecessary, and in a way it is. However, it can make the code easier to understand for people reading the code. If we had simply put a print-statement directly

````
print(2026-1994)
````

it may not have been immediately clear that `2026` was referring specifically to the current year or that `1994` was referring to the year that someone was born.

## Assigning and Using variables

In the previous section, we did 2 specific things with variables

- Assigning the values to variable (putting a label on the value)
- Doing something with those labeled values later

We also call the labeling of a value **assigning the value to a variable**. As we saw, we assign a variable following this pattern:

````
variable_name = "Any value that we want a label on"
````

We see that this assignment has three parts:

- the variable name on the left (`variable_name`)
- the `=` symbol (also called the 'assignment operator') in the middle
- the python value we want to label on the right (`"Any value that we want a label on"`)

All variable assignments in Python follow exactly this pattern.

> Important: the variable name is always on the left, the value you want to assign is always on the right. Trying to switch them around will not work! It would be like putting your sticky note in your moving box and trying to use your coffee machine as a label. If you do try it, Python will give you an error

````
"Any value that we want a label on" = variable_name
````

(will not work)

**After we have assigned the value to a variable**, we can access that value using the variable name (the label)

````
print(variable_name)
````

> Important: you can not access a variable before we have assigned it. That would be like looking for the box with the label 'coffee machine' on it before putting the label on the box. Therefore, if you do try to access a variable before assigning it, you will get an error:

````
print(my_name)
my_name = "Some value"
````

(will not work)

### Reassigning variables

Guess what the following code does before running it:

````
my_name = "James"
my_name = "John"
print(my_name)
````

Did it do what you expected? As you can see when you run this code, Python prints the second value. This is because in Python our "labels" (variables) are unique. In real life you can of course make 2 sticky notes with both having the words "coffee machine" on it and stick it on two different boxes. However, this might get confusing, because if your friend tells you to take the 'coffee machine' upstairs, which of the two boxes should you take? To avoid this type of confusion Python does not allow you to create two labels with the same name and put it on two different values. Instead, when you put the same label on a second value, Python automatically takes it off the first one. (So when you put the label "coffee machine" on a second box, it is taken off the first box first).

Taking a label of a value and putting it on a new value instead is what we call **Reassigning a variable**, because we reassign the variable (the 'label') to a new value.

### Assigning a value to multiple variables

Will the following code work? What will it do? Take a guess first and then run the code to see if you're correct:

````
my_name = "James"
my_first_name = my_name
print(my_first_name)
````

So what's going on here? Well actually it's pretty simple. As you can see when running this code, Python prints the original value we assigned. What happens here is that:

1. On the first line we put the label `my_name` on the value `"James"`
1. On the second line we put the label `my_first_name` on whatever value has the label `my_name` (in this case `"James"`) The real life analogy here is that you put the label "coffee machine" on a box. Then your friend asks you to put a new label "Sage Bambino" (this is a type of coffee machine) on the box with the "coffee machine" label. So this box ends up with two labels. This is perfectly fine.

Note that the first variable also still references the value:

````
my_name = "James"
my_first_name = my_name
print(my_name)
````

### Do variables stay synchronized?

Another case you might encounter in your code. What do you think will happen? First make a guess, and then run the code:

````
x = 10
y = x
x = 20
print(y)
````

So what's going on in this case? As you can tell by running the code, the variable `y` sticks with the value `10`. It is not updated because we only move the label `x`. Step by step what happens is:

- We put the label `x` on the value `10`
- We also put the label `y` on whatever value has the label `x` at this point (which at the time of putting the label is the value `10`)
- Then we take the label `x` of the value `10` and put it on the value `20` instead. (but the label `y` stays where it is)

### Impossible Equations?

Let's take a look at a final case we might see. What do you think it will do? Take a guess and run the code:

````
count = 0
count = count + 1
print(count)
````

Wait what? How can `count` be equal to `count + 1`? This is because in Python `count = count + 1` **is not an equality or an equation**, we simply use the `=`-symbol in Python to move a label onto a new value. So in effect **we first calculate what's on the right hand side of the `=`-symbol**, and then **put our label on whatever value comes out of that calculation**. So what happens step by step in this case is:

1. We put the label `count` on the value `0` (or in other words we assign the value `0` to the variable `count`)
1. Then on the second line we first calculate what's on the right hand side: we look whatever value currently has the label `count` (which is the value `0`) and add this value to the value `1`. The result of this addition is also the value `1` and we reassign the variable `count` to this value.
1. Finally we print whatever value currently has the label `count`.

## Augmented assignment

You can also combine an assignment with an operation! The most common example is:

````
counter = 0
print(counter)
counter += 1
print(counter)
counter += 1
print(counter)
counter += 1
print(counter)
````

`counter += 1` is equivalent to `counter = counter + 1` (but written in a shorter way). We can do this for any operator `*=`, `-=`, `/=` (though they are used far less often).

## Variable names

Run the following pieces of code:

````
name = "Johnny"
print("Hello " + name)
````

````
user_name = "Johnny"
print("Hello " + user_name)
````

````
x = "Johnny"
print("Hello " + x)
````

````
name_1 = "Johnny"
print("Hello " + name_1)
````

````
NAME = "Johnny"
print("Hello " + NAME)
````

````
literally_anything = "Johnny"
print("Hello " + literally_anything)
````

As you can see, all of these pieces of code work exactly the same. This is because **the name of the variable has no effect on the functionality** of the code. You can give a variable any name you want!

However, you should try to give your variable names that are **descriptive**. This way it's easier to read the code. This is why we originally gave the variable the name `name`, because it tells people who read the code that this variable points to a name.

### Variable name restrictions

While we just said you can give your variables any name you want, there are actually some rules. These rules are:

1. a variable name can **only** contain **letters** (upper- or lowercase), **numbers** or **underscores** (`_`)
1. a variable name **cannot start with a number** (so it has to start with a letter or underscore)
1. a variable name **cannot be equal to a Python keyword**. We haven't seen any keywords yet but will see what they are later on in the course.

Here are some examples of variable names that are invalid, and will therefore result in an error when you try to run these code pieces:

````
user-name = "Liu"
````

````
$name = "Ahmed"
````

````
3_names = "Johnny, Liu, Ahmed"
````

Note that the error messages don't always specifically call out that the variable name is invalid. However, they do point to the place where we specified an invalid name.

### Naming conventions

In addition to the above rules that are enforced (meaning that they result in errors if you violate them), there are a couple of extra "rules" that people follow by convention. These are not enforced, but you should follow them so it's easier for other people to read your code. These conventions are: * Use descriptive names. * All letters in an identifier should be lowercase. * If an identifier consists of multiple words, these words should be separated using underscores (`_`).

This means that names such as `my_name`, `player_2`, `football_players` etc follow these conventions, but the following do not:

- `i` (not descriptive)
- `My_Name` (not all letters are lowercase)
- `myname` (words are not separated by lowercase)

#### Prompts ending with `> `
You might have found it a little weird that all of the sudden we ended the question to the user with `> `. From
now on we will always do this. The reason for this is purely practical: when showing the output of the program, this
convention makes it immediately obvious what the input of the user is versus what the program printed out.

Compare the above output with the version that wouldn't include `> `:
```none
Welcome to the greeting program!
What is your name? Ben
Hello Ben!
```
In this version it is less clear that `Ben` is user input and not just part of the program output. In larger programs
with more output this will be particularly helpful.

Of course there is nothing special about the character `>` and we might have just as well used any other character
(such as `~` or `#` or anything else), so we of course **have to include it in our prompts explicitely** (for example in
the case above, by writing `input("What is your name ? > ")`).

From now on we will always end our prompts with `> ` when asking the user for input.
