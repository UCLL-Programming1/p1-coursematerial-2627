# Function Execution Mechanics

Let's take a more detailed look at how functions execute exactly.

Up until now in the course, we saw that code usually executes line-by-line, from top to bottom. We encountered two exceptions to this:

1. For `if-elif-else`-blocks, we saw that Python would only execute one of the `if-elif-else` 'branches', and skip the other ones.
1. For `while`-loops, we saw that while the code does execute line-by-line from top to bottom, when it reaches the end of the `while`-block, the execution would jump back to the `while`-condition and if the condition was still true, the `while`-block would be executed again.

Functions are another one of these exceptions. Consider the following example (which is part of the blackjack program we wrote in the previous section):

````
def get_card_value(card):
    if card == 'A':
        card_value = 1
    elif card == 'K' or card == 'Q' or card == 'J' or card == '10':
        card_value = 10
    else:
        card_value = int(card)
    return card_value

value = get_card_value("A")
print(value)
````

In which order will the lines of this cell be executed?

Execute this code step-by-step (If you don't remember how to do that, revisit our section on [executing code step-by-step](../2-control-flow/2-executing-step-by-step.md)), using 'Step Into' to go the next line every time.

We can then see that Python does the following:

1. It starts executing the cell at the first line, reading the function definition. Python will 'remember' that we have defined this function with this specific name.
2. However, because this function does not need to be executed yet, Python simply skips over the entire function body, immediately to line `value = get_card_value("A")`.
3. It is only here that we are going to use the function for the first time.
4. When trying to evaluate `get_card_value("A")`, Python needs to execute the function so it will jump back to **the first line of the function body**, to execute the function.
5. Python will now execute this function line-by-line, with the argument `card` having the value `"A"` (because that is the value that we provided on the line `value = get_card_value("A")`)
6. At the end of the function Python will reach the line `return card_value`. This means the value `card_value` will passed back to the code that called the function.
7. Python jumps back to the place in the code that called the function, so the line `value = get_card_value("A")`. The return value of the function will be assigned to the variable `value`.

Take some time to fully understand how this works, executing the code step-by-step as many times as you need.

## Defining a function != executing a function

Consider the following code. Before running the cell, think about what the code will do:

````
def say_hello(name):
    print(f"hello {name}!")
````

It does nothing? Why is that?

Well the reason is that we have **defined** the function, **but we never called it**, so **the function is not actually executed**. It's like writing down cooking instructions but then never actually giving it to anyone to do, or to write a Python program, but not running it.

Instead, if we actually want to execute the function, we need to do that explicitely:

````
def say_hello(name):
    print(f"hello {name}!")

say_hello("James")
````

This difference between **defining** a function (writing down the instructions) and actually **executing the function** (executing the instructions), is one of the key concepts to understand about functions!

## `return` in depth

Let's take a look a some of the peculiarities of `return`

### No return?

When looking at the function in the previous section, you might have noticed that this function **has no return**. Why is that and is that even allowed?

The reason this is fine, is that functions only need a return statement **if you want to pass a specific value** back to where the function was called. If you don't need to do that, you don't need to return.

In the first function we made ourselves:

````
def get_card_value(card):
    if card == 'A':
        card_value = 1
    elif card == 'K' or card == 'Q' or card == 'J' or card == '10':
        card_value = 10
    else:
        card_value = int(card)
    return card_value
````

We need the return statement to pass the value (an integer between 1 and 13 that represents the value of card letter) back to where we need it.

In the function `say_hello` we don't need it because all we want this function to do is print a message to the terminal.

We've seen this difference in functions already with the first two built-in functions we've learned: `print()` and `input()`

- `print()` does not give a value back to us (it **doesn't return** a value), it simply prints something to the program output
- `input()` does give us a value (the string containing the text the user typed as input), so `input()` **does return** a value.

> Functions that don't need to pass values back don't need return statements. Of course after the whole function body is executed, Python still 'returns' to the point in the code that called the function, that doesn't change.

### Printing vs returning

When learning about functions, sometimes students get confused about the difference between `print()` and `return`, thinking they basically do the same thing.

However, they do very different things:

- `return` stops the function, returning a value to the caller of the function. But it **does not** automatically print this value to the program output.
- `print()` prints a value as output (in the terminal or under the jupyter notebook cell). It **does not** pass that value back to where the function was called This means that:
- without `print()`, you'll have no output in the terminal
- without `return`, you cannot pass a value from within a function outside of it.

You can observe this difference by comparing the output of the following two functions (called `say_hello_print` and `say_hello_return`) that "do the same thing" except the first function prints a value and the second one returns it.

````
def say_hello_print(name):
    print(f"Hello {name}!")

print("say_hello_print:")
result_1 = say_hello_print("James")
print(f"result is: {result_1}")

print()

def say_hello_return(name):
    return f"Hello {name}!"

print("say_hello_return:")
result_2 = say_hello_return("James")
print(f"result is: {result_2}")
````

When you run this code (and it may help to run this code step-by-step), you'll see that the output is:

```none
Function with print():
Hello James!
result is: None

Function with return:
result is: Hello James!
```

We can observe the following things:

1. We see that `say_hello_print` has an extra line that is printed. This is because this function **prints the string `"Hello James!"` within the function body** while the function `say_hello_return` **does not** (it only returns the value).
1. By observing the output line `result is:` for both functions we can see **what value** ends up in the variable `result_1` and `result_2`.
   1. `say_hello_print` **doesn't pass** any value to the `result_1` variable. `result_1` is `None` (we'll see later what `None` means exactly - for now, just think of it as "There is no value")
   1. `say_hello_return` **does pass** a value to the variable `result_2`, namely the string `"Hello James!"`.

> `print()` prints values to the program output, `return` passes values to the point where the function was called.

### Returning early

We've mentioned that `return` ends the function (in addition to passing a value to where the function was called). In the examples so far, this didn't really come into play that much, because the `return`-statement was always on the last line of the function body, when the function was going to end anyway. However, consider the following example:

````
def function_with_early_return():
    print("Start of the function")
    user_input = input("Enter a message that will be returned: > ")
    return user_input
    print("End of the function")

user_input = function_with_early_return()
print(f"The user gave the following input: {user_input}")
````

Running the cell gives the following output: (if you type "hello")

```none
Start of the function
The user gave the following input: > hello
```

As you can see, the line `print("End of the function")` is not executed at all. This is because, like we said, `return` ends the function. So everything that comes after it, is simply not executed. In the cell, you can see that the code also looks a little bit more transparant. If you hover over the code with your mouse, you'll see a pop-over that says "Code is structurally unreachable", which means that the code can never be reached, as there is a `return` statement right before, which ends the function.

## `None`

In the previous section, we saw that a variable got the value `None`, which resulted from the function not returning anything.

`None` is a special value in Python that means there is no value or that a value is missing. The most common way you encounter this value the way we did, by expecting a function to return a value, but the function not returning a value.

This makes it different from values such as the empty string `""` or `0`. The empty string is still a string (it's just empty) and `0` is also still a number, the number `0`. `None` is not the number `0`, `None` means there is no number.

Therefore doing operations with this value almost always results in errors:

````
None + 34
````

````
"hello" + None
````

````
None.startswith("hello")
````

You will most likely encounter some of these errors at some point or another. If you see "NoneType" in an error message, this means that you probably forgot to return a value in one of your functions.

## Function arguments

So how do function arguments work exactly? Let's take another look at the very first function we wrote:

````
def get_card_value(card):
    if card == 'A':
        card_value = 1
    elif card == 'K' or card == 'Q' or card == 'J' or card == '10':
        card_value = 10
    else:
        card_value = int(card)
    return card_value

card_1 = "A"
value_1 = get_card_value(card_1)
print(f"The value of {card_1} is: {value_1}")

card_2 = "K"
value_2 = get_card_value(card_2)
print(f"The value of {card_2} is: {value_2}")
print(f"Total hand value is {value_1 + value_2}!")
````

Fundamentally, every time the function `get_card_value` is called, we pass a value between the parentheses. This value is passed to the function and is assigned to a new variable (the function argument). For example:

- on the line `value_1 = get_card_value(card_1)`, the value that we pass to the function is `card_1`, which is a variable that in turn refers to the value `"K"`
- Python then jumps to first line of the function body `if card == 'A'":`.
- While doing this, it creates a new variable `card` and assigns it to the value we passed to the function. This means that `card` will refer to the same value `card_1` refers to.

Revisiting the stick-note analogy that we used when first explaining [variables](../1-basic-syntax/6-variables.ipynb), The function argument `card` is basically a new label that we put on the value that we pass to the function.

### Argument or no argument

As we already knew from using the functions `print()` and `input()`, some functions like `print()` need an argument to do something useful, while a function like `input()` doesn't need one. We saw that when defining a function, we put the argument name between the parentheses, as is the case with our function `get_card_value()`:

````
def get_card_value(card):
    if card == 'A':
        card_value = 1
    elif card == 'K' or card == 'Q' or card == 'J' or card == '10':
        card_value = 10
    else:
        card_value = int(card)
    return card_value

value = get_card_value("A")
print(f"The value of A is: {value}")
````

But, if we don't any arguments in our function, we simply write no arguments between the parentheses. In one of the previous sections we already did this for the function `function_with_early_return`:

````
def function_with_early_return():
    print("Start of the function")
    user_input = input("Enter a message that will be returned: > ")
    return user_input
    print("End of the function")

user_input = function_with_early_return()
print(f"The user gave the following input: {user_input}")
````

#### Parentheses always needed!

Be careful! While the function has no arguments, we still need to write the parentheses `()` (even when we don't write anything in between). Otherwise Python will raise an error saying that the function doesn't follow the correct function syntax.

For example, executing the following cell, we result in a syntax error (you can tell in advance as VS Code displays its red squigly lines):

````
def function_with_early_return:
    print("Start of the function")
    user_input = input("Enter a message that will be returned: > ")
    return user_input
    print("End of the function")

user_input = function_with_early_return()
print(f"The user gave the following input: {user_input}")
````

### Multiple arguments

You can also create functions that take many arguments. In that case, you need to separate all the arguments names with **commas** (`,`).

For example, if we were to create a function to create an email address given a name and domain, we could write:

````
def make_email(first_name, last_name, domain):
    return first_name.lower() + "." + last_name.lower() + "@" + domain

print(make_email("James", "Bond", "gmail.com"))
````

As you can see, when calling this function, it is **important to specify all of the arguments, in the correct order**.

Python doesn't magically know which argument should correspond to which argument. So if you pass them in the wrong order, the result will not be what you expect:

````
def make_email(first_name, last_name, domain):
    return first_name.lower() + "." + last_name.lower() + "@" + domain

print(make_email("gmail.com", "Bond", "James"))
````

and if you forget an argument the code will result in an error:

````
def make_email(first_name, last_name, domain):
    return first_name.lower() + "." + last_name.lower() + "@" + domain

print(make_email("James", "gmail.com"))
````

This error message tells you that one argument is missing. Note that it says that `domain` is missing because again, Python doesn't know which value is supposed to correspond to which argument, so it simply saying "I see two arguments, so the third one is missing".
