## Exercise 4.1: Tic Tac Toe

Time to build another classic: a two-player game of tic tac toe. Two players take
turns placing their tick on a 3x3 board, and the first to get three of their ticks
in a row (horizontally, vertically or diagonally) wins. If the board fills up before
that happens, it's a draw. When a game ends, the players can choose to play again.

The board has nine spaces, numbered `0` to `8`:

```none
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8
```

Let's take a look how such a program would run, with a short game where player 1 (`Alice`) wins the top row:

```none
Welcome to Tic Tac Toe!

Please enter the name of player 1: > Alice
Alice, choose a tick: > X
Please enter the name of player 2: > Bob
Bob, choose a tick: > O

The current board is:
  |   |
  |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

Alice, it's your turn.
Enter the index of the field you would like to place your tick on: > 0

The current board is:
X |   |
  |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

Bob, it's your turn
Enter the index of the field you would like to place your tick on: > 0
Bad move, needs to be 0-8 and can't be on an occupied space!
Enter the index of the field you would like to place your tick on: > 3

The current board is:
X |   |
O |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

... (More turns are played here)

Alice, it's your turn.
Enter the index of the field you would like to place your tick on: > 2

The current board is:
X | X | X
O | O |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

Alice won!
Do you want to play again (Y/N)? > n
Thank you for playing!
```
where the names, the ticks, the space indices and the final `n` are input provided by
the user.

As you can see from the example, the program works as follows:

- The program prints `Welcome to Tic Tac Toe!` and then asks for both players' names
  and the tick (a single character like `X` or `O`) each of them wants to use.
- A game is then played on an empty board. **Player 1 goes first**, after which the
  players alternate turns. On a player's turn:
  - the program prints `<name>, it's your turn.` and asks for the index of the space
    they want to play,
  - if the space they enter is not a number from `0` to `8`, or is already taken, the
    program prints `Bad move, needs to be 0-8 and can't be on an occupied space!` and
    asks again,
  - once a valid space is given, the player's tick is placed there and the updated
    board is shown.
- After every turn the program checks whether the game is over. A player who just got
  three of their ticks in a row wins: the program prints `<name> won!`. If instead the
  board is now full with no winner, the program prints `It's a draw!`.
- When a game ends, the program asks `Do you want to play again (Y/N)? > `. Anything but
  `n` starts a fresh game on an empty board; `n` stops.
- When the players stop, the program prints `Thank you for playing!`.

## Exercise 4.1a: Tic Tac Toe: the board

The first step in implementing tic-tac-toe is to create an internal representation of the game board to keep track
of where the two players have placed their ticks.

We'll represent the board with a list of 9 elements, which each elements corresponds to the symbol that needs to be
displayed in the corresponding square.

In the orginal assignment, we saw that the players can place their ticks on the squares using the following indices:

```none
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8
```

Therefore, it is only natural that the top-left symbol can be accessed by index `0` in the list that represents our
board, the middle element by index `4` and the bottom-right element by index `8` etc.

In this exercise we'll write the following four functions:

- `initialize_field()` returns a fresh board: a list of nine `" "` spaces.
- `print_field(field)` prints the board as three rows, with the spaces separated by
  ` | `, followed by a blank line.
- `print_indices()` prints the same layout, but filled with the space numbers `0` to
  `8` instead of the board contents. This helps a player see which number belongs to
  which space.
- `print_board_update(field)` shows the whole picture: a blank line, the line
  `The current board is:`, the board itself, the line `The indices of these board
  spaces are:`, and the numbered layout.

### `initialize_field()`

The first function you need to write doesn't do a lot, it simply initializes a new board. This board is a list with 9
spaces. As the players place ticks on the board, these spaces will be replaced by the corresponding symbols.

```python .noeval
field = initialize_field()
print(field)
# Output: [' ', ' ', ' ', ' ', ' ', ' ', ' ', ' ', ' ']
```

### `print_field(field)`
This function takes the field that we have created, and prints it in two-dimensional format that we can see in the game.
(instead of a single line that you get when you simply use the built-in `print()` function)

For example, if we have a field with an `X` on space `0` and an `O` on space `4`:

```python .noeval
print_field(["X", " ", " ", " ", "O", " ", " ", " ", " "])
```
prints:
```none
X |   |
  | O |
  |   |

```
(note the empty line underneath the field)

### `print_indices()`
This function basically does the same as the previous function, but it doesn't take a field argument, as it will always
just print the indices of the squares, instead of the ticks that are on the field:

```python .noeval
print_indices()
```
prints:
```none
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

```

> Hint: while you can code `print_indices()` from scratch, you can make it very short by making use of the `print_field()`
> function you've already defined.

### `print_board_update(field)`

This function should call the previous two functions to print the current state of the board, followed by a reminder
of what indices belong to the different squares. It also starts with an empty line, and it also prints
`The current board is:` and `The indices of these board spaces are:` before printing out the board and the indices
respectively.

So the code
```python .noeval
print_board_update([' ', ' ', ' ', 'X', ' ', ' ', ' ', ' ', ' '])
```
will print exactly:
```none

The current board is:
  |   |
X |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8

```

## Exercise 4.1b: Tic Tac Toe: playing a turn

In the previous exercise we created the functions to create the board and to display it in a clear way. Now let's implement
a single **turn**: asking a player where they want to play, making sure it's a legal move, and placing their tick on the
board.

The part of the game we are going to make corresponds to a single turn like this:

```none
Bob, it's your turn
Enter the index of the field you would like to place your tick on: > 3

The current board is:
X |   |
O |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8
```

or, if the player first enters some faulty input:

```none
Bob, it's your turn
Enter the index of the field you would like to place your tick on: > 0
Bad move, needs to be 0-8 and can't be on an occupied space!
Enter the index of the field you would like to place your tick on: > w
Bad move, needs to be 0-8 and can't be on an occupied space!
Enter the index of the field you would like to place your tick on: > 3

The current board is:
X |   |
O |   |
  |   |

The indices of these board spaces are:
0 | 1 | 2
3 | 4 | 5
6 | 7 | 8
```

A turn consists out of:
- Telling the player it's their turn, with the message `"<player>, it's your turn"`
- Asking the player for input, using the message `"Enter the index of the field you would like to place your tick on: > "`
  - If the player enters wrong input (anything else than `"0"`-`"8"`, or a number that corresponds to a square that
    already has a tick on it - such as square `0` in the above example), the program displays the error message:
    `"Bad move, needs to be 0-8 and can't be on an occupied space!"`. **As long as the user enters wrong input**, the
    program will ask for input again.
- When the player has entered correct input, the program will update and display the new board state (using the
  `print_board_update(board)` function that we wrote in the last exercise)

Start from your solution to the previous exercise (you'll need those board functions
again) and add the following three functions:

- `is_valid_move(move, field)` returns whether `move` is a legal move on `field`. A move
  is legal when it is one of the space numbers `0` to `8` **and** that space is still
  empty. Note that `move` is the text the player typed, so it is a string like `"4"`.
- `ask_player_move()` asks `Enter the index of the field you would like to place your
  tick on: ` and returns exactly what the player typed.
- `play_turn(player, player_tick, field)` plays one full turn for `player`. It prints
  `<player>, it's your turn.`, then asks for a move. As long as the move isn't valid, it
  prints `Bad move, needs to be 0-8 and can't be on an occupied space!` and asks again.
  Once a valid move is given, it places `player_tick` on that space of `field` and shows
  the updated board.

### `is_valid_move(move, field)`

This function returns `True` or `False`, based on whether the typed space is a real, empty space:

```python .noeval
field = initialize_field()
field[4] = "X"

print(is_valid_move("0", field))    # Output: True
print(is_valid_move("4", field))    # Output: False   (space 4 is already taken)
print(is_valid_move("9", field))    # Output: False   (9 is not a space on the board)
print(is_valid_move("hi", field))   # Output: False
```

### `ask_player_move()`

This function shows the prompt and hands back whatever the player typed, (without checking it):

```python .noeval
move = ask_player_move()
# asks: Enter the index of the field you would like to place your tick on:
print(move)   # prints exactly what was typed, for example: 4
```

### `play_turn(player, tick, field)`

This function ties the previous functions together into one full turn. For example when running the code:
```python .noeval

```

or, iff the player first tries some illegal moves, they are asked again until they enter a valid one:

```none

Enter the index of the field you would like to place your tick on: > 4

The current board is:
  |   |
  | X |
  |   |

The indices of these board spaces are:

3 | 4 | 5
6 | 7 | 8

```
where `9`, `x` and `4` are input provided by the user.

## Exercise 4.1d: Tic Tac Toe: checking victories

Currently, our game loops until the board is full, but in reality it should end the moment
someone gets three of their ticks in a row. Before we can add it to the loop, we of course need a way to do this
check.

There are three ways a player can win:
- by occupying all the spaces on any of **the 3 horizontal rows**
- by occupying all the spaces on any of **the 3 vertical rows**
- by occupying all the spaces on any of **the 2 diagonals**
  (from top-left to bottom-right, or from top-right to bottom-left)

Of course, after every turn we could check whether **either one of the players** won the game, but this is a bit
wasteful. We only need to check **whether the player who actually placed a tick in the previous turn** won.

Let's turn this check into functions and split it up in the 3 ways a player can win:
- `win_horizontally(field, player_tick)`: returns `True` if any of the **3 rows** is filled with the symbol `player_tick`, `False` otherwise
- `win_vertically(field, player_tick)`: returns `True` if any of the **3 columns** is filled with the symbol `player_tick`, `False` otherwise
- `win_diagonally(field, player_tick)`: returns `True` if any of the **2 diagonals** is filled with the symbol `player_tick`, `False` otherwise

All of these function have two arguments:
- the `field` on which the game is played
- the tick of the player for whom we are checking whether they won

For example, the function `win_horizontally(field, player_tick)` could be used as follows:
```python .noeval
field = ["O", "O", "O",
         "X", "X", " ",
         " ", " ", " "]
print(win_horizontally(field, 'O')) # True (the whole first row is occupied by O's)
print(win_horizontally(field, 'X')) # False (it's the player with 'O' who won, not the player with 'X')
```

Of course it's not only the first row that results in a victory:
```python .noeval
field = ["X", "X", " ",
         "O", "O", "O",
         " ", " ", " "]
print(win_horizontally(field, 'O')) # True (the whole second row is occupied by O's)
print(win_horizontally(field, 'X')) # False (it's the player with 'O' who won, not the player with 'X')
```

And a player could win vertically as well, for example:
```python .noeval
field = ["X", "X", "O",
         " ", " ", "O",
         " ", " ", "O"]
print(win_vertically(field, 'O')) # True (the whole third column is occupied by O's)
print(win_horizontally(field, 'O')) # False (even though the player with tick 'O' won, they only did so vertically, not horizontally)
```
or diagonally, for example:
```python .noeval
field = ["O", "X", "X",
         " ", "O", "X",
         " ", " ", "O"]
print(win_diagonally(field, 'O')) # True (the diagonal from top-left to bottom-right is filled with O's)
```

- `is_game_over(field, player, player_tick)` ties it all together. It does two things:
  - It returns `True` if the game is over (either by a win or a tie), `False` otherwise
  - If the game is over it prints out how the game ended:
    - If `player` has won (horizontally, vertically or diagonally), it prints `<player> won!`.
    - If the board is full, it prints `It's a draw!`.
    (If the game is not over, it doesn't need to print anything)

For example, on a board where `Bob` has just completed the top row with `O`:

```python .noeval
field = ["O", "O", "O",
         "X", "X", " ",
         " ", " ", " "]
result = is_game_over(field, "Bob", "O")   # prints: Bob won!
print(result) # True (this is the value returned by the function `is_game_over`)
```

On a full board where nobody made a line, any player check gives a draw:

```python .noeval
field = ["X", "X", "O",
         "O", "O", "X",
         "X", "X", "O"]
result = is_game_over(field, "Alice", "X")   # prints: It's a draw!
print(result) # True (the return value of the function is still True in case of a draw)
```

And on a board where the game is still going:

```python .noeval
field = ["X", "O", " ",
         " ", " ", " ",
         " ", " ", " "]
result = is_game_over(field, "Alice", "X")   # prints: (nothing)
print(result) # False (the game is not over)
```
