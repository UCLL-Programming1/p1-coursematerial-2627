## Benchmarking List Operations

We set up some benchmarks to check how fast some operations on lists are. It runs a specific operation on a list 100.000 times. You can run these benchmarks on your own machine:

````
import time

benchmark_list = [5] * 200000

first_bench = time.time()
for i in range(0,100000):
  benchmark_list.insert(0,6)
first_bench_end = time.time()
print(f"Add element at beginning of list: {first_bench_end - first_bench}s")

second_bench = time.time()
for i in range(0,100000):
  benchmark_list.append(6)
second_bench_end = time.time()
print(f"Add element at end of list: {second_bench_end - second_bench}s")

third_bench = time.time()
for i in range(0,100000):
  benchmark_list.pop(0)
third_bench_end = time.time()
print(f"Remove element at beginning of list: {third_bench_end - third_bench}s")

fourth_bench = time.time()
for i in range(0,100000):
  benchmark_list.pop()
fourth_bench_end = time.time()
print(f"Remove element at end of list: {fourth_bench_end - fourth_bench}s")
````

When we ran these benchmarks on one of our machines, we had the following results:

| Operation | Time |
| ------- | ------- |
| Add element at beginning of list | 5.57s |
| Add element at end of list | 0.005s |
| Remove element at beginning of list | 12.0s |
| Remove element at end of list | 0.005s |

#### `INFO`

Note that the absolute times do not mean much: each operation is repeated thousands of times. What matters is how the times relate to each other.

As you can see, operations at the beginning of the list are dramatically slower compared to those taking place at the end of the list. What could be the reason behind this phenomenon?

It's actually very simple: when adding/removing an item at the beginning of a list, all subsequent elements have to be shifted right or left. This overhead does not exist when adding/removing at the end of the list.

## Data Structures

A list is a kind of [*data structure*](https://en.wikipedia.org/wiki/Data_structure). A data structure provides a number of *operations*. For example, a list allows indexing, adding and removing elements, looking for elements, etc.

As shown above, these operations can differ in speed. The [*time complexity*](https://en.wikipedia.org/wiki/Time_complexity) of an operation describes how much time each operation takes. This is actually quite a complicated subject; we won't go into the details here. Instead, we'll simply make the distinction between *slow* operations and *fast* operations. For example, insertion and deletion are *slow* when done at the front of the list, but *fast* when done at the end.

## Membership Testing

Let's examine the operation `x in lst`, i.e., checking whether `x` occurs in `lst`. Is this a slow or a fast operation?

There is no magic or special trickery involved in the operation. In essence, it performs the following algorithm

````
for item in lst:
    if item == x:
        return True
return False
````

So we can yet again set up a benchmark for this, as we did before. We yet again run 100.000 iterations:

````
import time

benchmark_list = [5] * 1000
x = 3

first_bench = time.time()
for i in range(0,100000):
  x_in_list = x in benchmark_list
first_bench_end = time.time()
print(f"Membership testing: {first_bench_end - first_bench}s")
````

On one of our machines, this gives the following result:

| Operation | Time |
| ------- | ------- |
| Membership testing | 0.51s |

In other words, it goes through the list, compares `x` with each item, and if some item turns out to be equal to `x`, `x in lst` returns `True` and `False` otherwise. This is a slow operation: think of having to find a term in an encyclopedia that isn't alphabetically sorted. Compare this benchmark result with the benchmark of adding an element to the end of the list we saw before, and remember that when membership checking we're not even making any changes to the list... Sadly, we cannot do better than this: lists are inherently slow when it comes to membership checking.

## Sets

By now it should be pretty clear that lists have some operations that are fast, while other operations are really slow. Sometimes we need to perform a certain operation many times in our program and we would like to have a faster version of the list for this to manage our sequences of data. Fortunately, there exist other *data structures* than just lists that can improve this for us. Some of these data structures can store and process the sequence of elements in a certain way, making certain operations faster which would otherwise be slow. We won't go into detail on how this works, just remember that you can choose different data structures to make things faster and more efficient. A *set* is such a data structure that we'll have a look at here.

The good news about sets is that the basic operations are very fast.

We have yet again set up the same benchmarks like we performed on lists, performing 100.000 iterations of a certain operation on a set. You can run the benchmarks yourself:

````
import time

benchmark_set = {i for i in range(0, 200000)}

first_bench = time.time()
for i in range(0,100000):
  benchmark_set.add(15222)
first_bench_end = time.time()
print(f"Adding an item: {first_bench_end - first_bench}s")

second_bench = time.time()
for i in range(0,100000):
  benchmark_set.remove(i)
second_bench_end = time.time()
print(f"Removing an item: {second_bench_end - second_bench}s")

benchmark_set = {i for i in range(0,1000)}
x = 562012
third_bench = time.time()
for i in range(0,100000):
  x_in_set = x in benchmark_set
third_bench_end = time.time()
print(f"Membership checking: {third_bench_end - third_bench}s")
````

We get the following results when we run it on one of our machines:

| Operation | Set Time |
| ------- | ------- |
| Adding an item | 0.007s |
| Removing an item | 0.006s |
| Membership | 0.007s |

As you can see, adding and removing an item from a set are about as fast as adding and removing an item from the end of a list, and membership testing is considerably faster for sets than it is for lists.

As you can imagine however, having this extra performance doesn't come for free. The bad news is that sets have some downsides compared to lists that you need to take into account:

- Sets are unordered: you have no control over the order of the items. You can therefore also not index a set, i.e., `my_set[0]` is invalid.
- No duplicates allowed: every item can appear only once in a set. Adding an item a second time has no effect.

Choosing to use a set is typically done for purposes of efficiency: a list can do everything a set can do, just more slowly. The trade-off that we see with sets generally also applies to other datastructures: speed and efficiency for some operations are introduced at the cost of some flexibility.

## Sets in Python

Here is a quick overview of some of the operations you can perform on sets in Python:

| Syntax | Description |
| ------- | ------- |
| `{1, 2, 3}` | Set literal |
| `len(my_set)` | Number of items in set |
| `my_set.add(x)` | Add `x` to set `my_set` |
| `my_set.remove(x)` | Remove `x` from set `my_set` |
| `x in my_set` | Check if `x` is an element of `my_set` |

### `IMPORTANT`

Because sets are unordered, `{1, 2, 3}` and `{3, 2, 1}` are equal sets. Because sets "swallow" duplicates, `{1, 1, 1}` and `{1}` are also equal.

### `IMPORTANT`

Note that the empty set cannot be written `{}`, you have to use `set()`.

The reason for this is that dictionaries use the same curly braces. If there's at least one element, the syntax gives away if the literal is meant to denote a set or a dictionary:

- `{value}` is a set.
- `{key: value}` is a dictionary.

However, when no items are listed, both set and dictionary syntax reduce to `{}`. To resolve this ambiguity, Python chooses to see `{}` as a dictionary and lets you use `set()` for an empty set.

The fact that sets do not allow duplicates can be a very convenient feature if you want to know how many unique instances you have of something. For example, let's say you have a large group of wedding visitors all coming from different countries. If you want to know which unique nationalities have been present on the wedding, we would have to do it as follows if we were to use a normal list:

````
guests_origin_country = ['Belgium', 'France', 'Belgium', 'Peru', 'Australia', 'France'] # The country of origin for each guests
unique_nationalities = [] # We want to know which unique nationalities we have

for country in guests_origin_country:
    if country not in unique_nationalities:
        unique_nationalities.append(country)
print(unique_nationalities) # 'Belgium', 'France', 'Peru', 'Australia'
````

With a list, it's inconvenient that we constantly have to check that we don't add any duplicates to our list of unique nationalities. With a set however, this becomes much easier as a set never stores duplicates so we can always just add the country:

````
guests_origin_country = ['Belgium', 'France', 'Belgium', 'Peru', 'Australia', 'France'] # The country of origin for each guests
unique_nationalities = set() # We want to know which unique nationalities we have

for country in guests_origin_country:
    unique_nationalities.add(country)
print(unique_nationalities) # 'Peru', 'Belgium', 'Australia', 'France'
````

Note that our ordering is different now because sets are unordered.

This example we can actually make even simpler by converting our guest country list to a set, which automatically removes all duplicates. Our code then looks like this:

````
guests_origin_country = ['Belgium', 'France', 'Belgium', 'Peru', 'Australia', 'France']
unique_nationalities = set(guests_origin_country)

print(unique_nationalities) # 'Peru', 'Belgium', 'Australia', 'France'
````
