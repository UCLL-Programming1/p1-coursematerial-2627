# Tuples

So far, we've seen how lists allow us to to maintain a sequence of values. We've also seen that lists can be modified: we can add or remove elements from the list without having to make a new one. Sometimes, however, this can be undesirable.

Let's say that we have a list where we're keeping track of the days of the week. The days of the week never change and we always know that we have 7 different days.

````
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
print(len(days)) # We have 7 days
````

Let's say I have another list `numbers` which I want to add a new element to, but I make a small mistake and accidentally add it to my `days` list instead.

````
numbers = [1,5,7,2]
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
days.append(3) # We want to add an element to numbers but accidentally add it to days
print(len(days)) # We have 7 days ... or do we?
````

If we now want to check again how many days there are in the week, we end up with 8 days. This is because our list of weekdays is still mutable, even though we would prefer to have this fixed so mistakes like this can't happen.

Fortunately, python has another type for this, the **tuple**. A tuple functions mostly the same as a list, except you can't modify it. You can't update any values, or add or remove any values to it. It is **immutable**.

To make a tuple, you can use the same syntax as with lists, except now we use round brackets `()` instead of square ones `[]`.

````
my_tuple = (6, 7)
````

If we then look back at our previous example with the days of the week, we can replace our list with a tuple and notice that our code now gives an error, pointing out our mistake:

````
numbers = [1,5,7,2]
days = ('Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday')
days.append(3) # Can't append to tuple
print(len(days))
````

This allows us to really ensure that this collection of days of the week is always what we expect of it: That it contains exactly 7 items and that they are the days which we specified. This makes our code more readable and more predictable.

#### `INFO`

The syntax for creating a tuple with a single item is a bit exceptional. When we try to do it as one would expect, we get the following:

````
my_t = ("a")
print(my_t)
````

We see that we don't actually get a tuple, but just the value. This is because the round brackets used for tuples are also parantheses used in other statements (like in mathematical expressions). Here it's seen as redundant parentheses instead of the tuple initializer, so instead we have to initialize it as follows:

````
my_t = ("a",)
print(my_t)
````

Fortunately it's quite uncommon to use tuples with only a single value as it defeats the purpose of tuples, so you will rarely bump into this issue.

As mentioned before, most operations that work on lists also work on tuples. They main thing you should remember is that tuples function in the same manner as lists, except you can't perform any operations to modify them. Here are some example operations that you can perform on tuples, just like you can on lists:

- slicing (my_t[2:5])
- membership testing (if x in xs)
- iteration using for loops
- min/max, sum, sorted, len() ...

And some operations that you can perform on lists but not on tuples:

- update: lst[index] = value
- append, insert
- pop, remove, del

## Destructuring

Let's say we have a tuple with three entries:

````
my_tuple = ("one", "two", "three")
````

So far, if we wanted to retrieve the individual items, we always had to access them individually:

````
my_tuple = ("one", "two", "three")
first = my_tuple[0]
second = my_tuple[1]
third = my_tuple[2]
print(f" We have a {first}, a {second} and a {third} item")
````

This is a bit cumbersome. However, python also allows us to write it more elegantly:

````
my_tuple = ("one", "two", "three")
first, second, third = my_tuple
print(f" We have a {first}, a {second} and a {third}")
````

This is called **destructuring**. Tuples make it nice to use destructuring, because you know they are immutable and will thus always have a fixed number of items. However, lists and other iterables can just as well be destructured like this:

````
my_list = ["a", "b", "c"]
first, second, third = my_list
print(f" We have a {first}, a {second} and a {third}")
````

They are a bit riskier to destructure though, because if we add another element to the list our code will throw an error:

````
my_list = ["a", "b", "c"]
my_list.append("d")
first, second, third = my_list # Too many values!
print(f" We have a {first}, a {second} and a {third}")
````

## Conclusion: When to use tuples

At first glance tuples may seem a bit pointless as they are a more limited form of a list, which might make you inclined to always opt for lists instead. However, as the previous examples have demonstrated, using tuples can make your code more predictable and less error prone. This might not seem as important now, but when you're working with larger code examples it can really make your life a lot easier.

In general, it's a good idea to use tuples when you know beforehand that your sequence of elements is fixed. The extra features that a list has over a tuple won't be of any use in this case and can only introduce errors.

To summarize:

- Use a **list** if: you don't know how many elements you need in your sequence and if this can change
- Use a **tuple** if: you know how many elements your sequence has and this remains fixed
