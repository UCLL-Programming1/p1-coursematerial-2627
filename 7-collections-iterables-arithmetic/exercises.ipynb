## Exercise 7.1: Potion Brewing Game

Welcome, apprentice alchemist!

Your goal is to discover and brew powerful potions using magical ingredients.

Every ingredient has:
- A name
- An elemental type
- A power level

For example:

````
("Fire", 8)
````

means that the ingredient belongs to the Fire element and contributes 8 power when brewing.

The available ingredients are stored in a dictionary:

````
ingredients = {
    "Mandrake Root": ("Nature", 5),
    "Dragon Scale": ("Fire", 8),
    "Mermaid Tear": ("Water", 6),
    "Storm Crystal": ("Electric", 7),
    "Phoenix Feather": ("Fire", 10),
    "Moon Flower": ("Nature", 4)
}
````

The key is the ingredient name.
The value is a tuple:

```
(element_type, power)
```

The ingredients currently added to the cauldron are stored in a set.

Example:

````
cauldron = {
    "Dragon Scale",
    "Storm Crystal"
}
````

The set ensures that the same ingredient cannot be added twice.

Write the following functions:

- `show_ingredients(ingredients)` which prints all available ingredients
- `add_to_cauldron(ingredients, cauldron, ingredient_to_add)` which adds an ingredient to the cauldron, given that it is known in the ingredients list. Returns True if ingredient has been successfully added and False if it was already present in the cauldron or if it is an unknown ingredient.
- `show_cauldron(cauldron)` which prints all ingredients currently in the cauldron
- `brew_potion(ingredients, cauldron)` which brews a potion using the ingredients currently in the cauldron. The function should return a tuple containing on the one hand a list containing all elemental types used and on the other hand the calculated total power. Every elemental type appears only once in this list, even when several ingredients share it. The list with the elemental types should be sorted alphabetically (you can use `sorted()` for this, which does it for you). When a potion is brewed, the cauldron is cleared.

Example function usages:


```python .noeval
ingredients = {
    "Mandrake Root": ("Nature", 5),
    "Dragon Scale": ("Fire", 8),
    "Mermaid Tear": ("Water", 6),
    "Storm Crystal": ("Electric", 7),
    "Phoenix Feather": ("Fire", 10),
    "Moon Flower": ("Nature", 4)
}

cauldron =  {
  "Dragon Scale",
  "Storm Crystal"
}

show_ingredients(ingredients)
# Prints
# Dragon Scale (Fire) - Power 8
# Moon Flower (Nature) - Power 4
# etc ...

# if there is something in the cauldron
show_cauldron(cauldron)
# Prints
# Ingredients in the cauldron:
# Dragon Scale
# Storm Crystal

# if the cauldron is empty
cauldron.clear()
show_cauldron(cauldron)
# Prints
# The cauldron is empty.

# Add Dragon Scale to cauldron
ingredient_added = add_to_cauldron(ingredients, cauldron, "Dragon Scale")
print(ingredient_added)
# Prints True

# Add Storm Crystal to cauldron
ingredient_added = add_to_cauldron(ingredients, cauldron, "Storm Crystal")
print(ingredient_added)
# Prints True

# Add Dragon Scale to cauldron again, even though it's already in it
ingredient_added = add_to_cauldron(ingredients, cauldron, "Dragon Scale")
print(ingredient_added)
# Prints False

# Add Dragon Tooth to cauldron, which is not known in the ingredients list
ingredient_added = add_to_cauldron(ingredients, cauldron, "Dragon Tooth")
print(ingredient_added)
# Prints False

types_used, total_power = brew_potion(ingredients, cauldron)
print(types_used)
print(total_power)
# Prints
# ['Electric', 'Fire']
# 15
```
