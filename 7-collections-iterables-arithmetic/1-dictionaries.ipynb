# Dictionaries

Consider the following situations:

- You need to translate a text and would like having a dictionary that, given a word in English gives the corresponding word in Dutch.

  | English | Dutch |
  | ------- | ----- |
  | cat | kat |
  | dog | hond |

- You have found a recipe for a delicious chocolate cake. It comes with a list of ingredients and the required quantities for each.

  | Ingredient | Quantity |
  | ---------- | -------- |
  | chocolate | 250g |
  | eggs | 5 |
  | sugar | 125g |
  | flour | 75g |
  | butter | 175g |

- Given a student number, you wish to know the name of the student associated with that number.

  | Student Number | Student Name |
  | -------------- | ------------ |
  | r0123456 | Karen Eiffel |
  | r0654321 | Harold Crick |
  | r0121212 | Ana Pascal |
  | r0954321 | Jules Hilbert |

In each cases, we can represent this data as a table. Each row associated some data in the left column with some other data in the right column:

- `cat` is associated with `kat`.
- `chocolate` is associated with `250g`.
- `r0123465` is associated with `Karen Eiffel`.

We call the data in the left columns the *keys* and the corresponding data in the right column the *values*.

Python allows us to easily model such data using *dictionaries*, or `dict`s for short. Our *key* is then what you use to look up the data, the *value*, in the dictionary.

Here's what the translation dictionary would look like:

````
translating_dictionary = {'cat': 'kat', 'dog': 'hond'}
````

If we then have a `key` (in this case the English word for our pet), we can easily find the corresponding translation, the `value`, by doing a lookup in the dictionary:

````
translating_dictionary = {'cat': 'kat', 'dog': 'hond'}
print(translating_dictionary['cat']) # We look up the dutch translation of the English word 'cat', returning 'kat'
````

So far, all the keys and values in our dictionairies were strings. However, dictionaries are not limited to this: both keys and values can have any type. Dictionaries can even contain other dictionaries recursively. For example, this is a perfectly valid `dict`:

````
{
    1: True,
    'a': [1, 2, 3],
    False: {1: 2, 3: 4}
}
````

### `INFO`

Actually, there *is* a limitation on the types of keys: the objects have to be *hashable* and comparable using `==`. We'll delve further into this later. For now, you can imagine there are no constraints on key types.

## Operations

We'll now delve into the typical operations you would perform on a dictionary.

## Lookup

The most common operation for a dictionary is to, given a key, _look up_ the associated value.

We've seen this example before when we were using the translating dictionary. The lookup looked like this:

````
translation = translating_dictionary["cat"]
````

We looked up the Dutch translation `kat` of the English word `cat`. Here, `cat` is the key, and `kat` is the value we wish to look up in our dictionary.

After this code example is executed, `translation` is equal to `"kat"`.

## Insertion

`dict`s are *mutable*, i.e., they can be modified. For example, we can add new key/value pairs to an existing `dict` as follows:

````
# We start with empty dict
my_dict = {}

my_dict['a'] = 1
my_dict['b'] = 2
my_dict['c'] = 3
````

After executing this code, `my_dict` equals `{'a': 1, 'b': 2, 'c': 3}`.

It is also possible to overwrite existing associations:

````
my_dict['a'] = 5
my_dict['b'] *= 2
````

Now `my_dict` equals `{'a': 5, 'b': 4, 'c': 3}`.

## Deletion

Apart from adding and overwriting key/value pairs, it is also possible to delete them with `pop(key)` like you've seen with lists. It will remove the pair from the dictionary and return the value associated with the key:

````
my_dict = {'a':5}
my_dict.pop('a') # 5
````

After calling `pop(key)`, `dictionary` will no longer associate `key` with a value. If `key` did not appear in `dictionary` at the moment of calling `pop`, you will get an error.

````
my_dict = {}
my_dict.pop('a') #KeyError: 'a'
````

## Membership Testing

Given a dictionary `d`, you can look up the value associated with some `key` using `d[key]`. However, it is an error to look up a `key` that is not included in the dictionary, just like with deletion:

````
d = {'a': 1, 'b': 2}
d['c'] #KeyError: 'c'
````

In order to prevent this error from occurring, we might want to first check if the dictionary does indeed contain an entry with `key`. You can do this with the `in` operator:

````
d = {'a': 1, 'b': 2}
print('a' in d) # True

print('b' in d) # True

print('c' in d) # False
````

Note that `in` only works on keys. You cannot look for values.

````
d = {'a': 1, 'b': 2}
print(1 in d) # False
````

## Enumerating

As of yet, we've seen two ways to retrieve information from a dictionary:

- `d[key]` looks up the value associated with `key`.
- `key in d` tells you whether `d` contains a certain `key`.

Remark how you always need a key to start with. If you have no key handy, you can't do much with the `dict`, apart from trying to guess which keys it contains.

### `keys()`

Luckily, `dict`s allow you to ask for a list of keys:

````
d = {'a': 1, 'b': 2}
print(d.keys()) # dict_keys(['a', 'b'])
````

While `keys()` does not really return a `list`, you can think of the result as one.

### `INFO`

Why does `keys()` not simply return a `list`? Consider the following code:

````
d = {'a': 1, 'b': 2}
keys = d.keys()
print(keys) # dict_keys(['a', 'b'])

d['c'] = 3
print(keys) # dict_keys(['a', 'b', 'c'])
````

If `keys()` were to return `list`, it would contain a copy of the keys that are present at the moment you call `keys()`. Adding new keys to the `dict` later would not affect the `list`. The `list` is "dumb": it is not aware that its contents come from a `dict`.

A `dict_keys` object is very much *like* a list, but it is "smarter": it *knows* its items are keys of a `dict` and will keep itself synchronized. In other words, if you were to add or remove keys after calling `keys()`, this change would be reflected in the `dict_keys` object.

### `values()`

`dict`s also provide you with a `values()` method, which returns a "list" of values:

````
d = {'a': 1, 'b': 2}
d.values() # dict_values([1, 2])
````

### `items()`

Say you want to iterate over all key/value pairs. You can achieve this using

````
for key in d.keys():
    value = d[key]
    print(f"Key: {key}, value: {value}")
````

However, you achieve the same more efficiently using `items()`:

````
for key, value in d.items():
    print(f"Key: {key}, value: {value}")
````

This way, you don't have to manually look up every key.

### `INFO`

In essence, `items()` returns a list of pairs, i.e., tuples of size 2. The loop above relies on destructuring as we've seen in the previous chapter to make our code more readable. It is equivalent with

````
for pair in d.items():
    key = pair[0]
    value = pair[1]
    # ...
````

### `len`

If you simply need to know how many key/value pairs are present in a `dict`, you can rely on `len`:

````
d = {'a': 1, 'b': 2, 'c': 3}
len(d) # 3
````

## Lookup With Defaults

Say you have a `dict` that represents the contents of your pantry:

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
````

You intend to bake some dessert and need to know if you have enough eggs. The dessert requires four eggs, so you perform the following check:

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
if pantry['eggs'] >= 4:
    print("We have enough eggs!")
````

But what if you check for vanilla beans?

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
if pantry['vanilla beans'] >= 1:
    print("We have enough vanilla beans!")
````

This will lead to a crash: you can only look up using an existing key. To prevent a crash, you would need to write

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
if 'vanilla beans' in pantry and pantry['vanilla beans'] >= 1:
    print("We have enough vanilla beans!")
````

That's a lot of typing for something relatively trivial. Luckily, there exist better solutions.

### `get`

`dict`s have a method named [`get`](https://docs.python.org/3/library/stdtypes.html#dict.get). `d.get(key)` does not raise an error if it does not have the key like with `d[key]`, but instead it returns `None`. It also takes an optional second parameter: a default return value in case `key` does not appear in `d`.

Our situation from before was:

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
pantry['vanilla beans'] #KeyError: 'vanilla beans'
````

When we switch to `get`, we no longer get an error but instead we get None:

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
pantry.get('vanilla beans') #None
````

If we add a default value to our `get`, we now get 0 instead of None:

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
pantry.get('vanilla beans', 0) # We now get 0
````

As you can see, `get` allows you to succinctly express that a missing ingredient in the ingredient is the same as having 0 of that item. Therefore, the check above can be rewritten as

````
pantry = {'sugar': 200, 'eggs': 5, 'flour': 750}
if pantry.get('vanilla beans', 0) >= 1:
    print("We have enough vanilla beans!")
````

### `INFO`

More advanced students might like to know about [`defaultdict`](https://docs.python.org/3/library/collections.html#collections.defaultdict). This is a special kind of `dict` that allows you to define an `if_missing` value at creation. Every lookup `d[key]` is then equivalent to `d.get(key, if_missing)`. The main advantage is that you only need to define the `if_missing` value once (i.e., at creation) instead at every lookup.
