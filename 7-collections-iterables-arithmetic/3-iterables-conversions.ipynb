# Conversions between iterables (list, str, set, tuple)

When we introduced the concept of **iterables** when discussing `for`-loops, we noted that strings and lists were iterables and thus shared some similarities. The similarities we noted were:

- They represent a sequence of items.
- They have a length: `len(xs)`.
- Many operations (`min`, `max`, etc.) can be applied on them.

Now we've seen a few extra: `tuple` and `set` are also iterables.

Using `str(x)`, `tuple(x)`, `list(x)` and `set(x)` you can convert `x` to a string, tuple, list and set respectively, as long as `x` is an iterable (not necessary for `str(x)`).

### `str`

We already discussed `str`. It can be applied on (as good as) any value `x` to build a string representation of `x`.

```
# Converts list to string
from_list = str([1, 2, 3])
print(from_list) # "[1, 2, 3]"


# Converts tuple to string
from_tuple = str((1, 2, 3))
print(from_tuple) # "(1, 2, 3)"
```

### `tuple`

Given an iterable value `x`, `tuple(x)` will take all values stored in `x` and put them in a tuple.

```
# Converts string to tuple
from_str = tuple("abcd")
print(from_str) #("a", "b", "c", "d")

# Converts tuple to tuple, not very useful
from_tuple = tuple((1, 2, 3, 4))
print(from_tuple) #(1, 2, 3, 4)

# Converts list to tuple
from_list = tuple([1, 2, 3, 4])
print(from_list) #(1, 2, 3, 4)

# Converts set to tuple
from_set = tuple({1, 2, 3, 4})
print(from_set) #(1, 2, 3, 4)
```

### `list`

We can use `list` in the same way as we used `tuple`: `list` takes all items in `x` and puts them in a list.

```
# Converts string to list
from_str = list("abcd")
print(from_str) #["a", "b", "c", "d"]

# Converts tuple to list
from_tuple = list((1, 2, 3, 4))
print(from_tuple) #[1, 2, 3, 4]

# Converts list to list, can be used to make a copy
from_list = list([1, 2, 3, 4])
print(from_list) #[1, 2, 3, 4]

# Converts set to list
from_set = list({1, 2, 3, 4})
print(from_set) #(1, 2, 3, 4)
```

### `set`

In the chapter on sets earlier on we've already seen how we could convert a `list` into a `set`. Important to note was that converting to a `set` removes duplicates and rearranges the ordering, as sets are unordered.

We can also convert the other iterables we've seen so far into sets.

```
# Converts string to set
from_str = set("abcda")
print(from_str) #["b", "c", "d", "a"], duplicate "a" removed and order changed

# Converts tuple to set
from_tuple = set((1, 2, 3, 4))
print(from_tuple) #[1, 2, 3, 4]

# Converts list to set
from_list = set([1, 2, 3, 4])
print(from_list) #[1, 2, 3, 4]

# Converts set to set, can be used to make a copy
from_set = set({1, 2, 3, 4})
print(from_set) #(1, 2, 3, 4)
```

## Our final iterable: dict

Dictionaries are technically also iterables, as they also allow you to iterate over them, index them, etc ... They are not just a simple sequence of items though, so we can't convert them into any of the other iterables like we've seen here.
